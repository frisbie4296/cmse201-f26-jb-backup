**1. Variables**

```python
int_var = 2   # Integer variable
float_var = 2.34  # floating point variable
str_var = 'bananas'  # string variable

print('1:', 'An integer plus a float works in python:',int_var+float_var)
#You can not do math with strings, but you can concatenate strings (if you turn your variables into strings first)
new_str_var = str(int_var)+' '+str_var+' is '+str(float_var)+' times the fun.'
print('2:',new_str_var)
# or you can just use a print statement with commas to make meaningful debugging and result statements
print('3:',int_var,'or',float_var, str_var,'are not all that fun.')
```

**2. Lists**

A list stores a series of items in a particular order. You access items using an index, or within a loop.

```python
list_ex = []   # initialize an empty list
list_ex.append('we')  # append an item to a list
list_ex.append('have')
list_ex.append('no')
list_ex.append('apples')
print('Print 1:',list_ex)  # print contents of variable or list
list_ex.remove('apples')  # remove specific entry from list, but only first entry with this value
print('Print 2:',list_ex)  # print contents of variable or list
list_ex.append('bananas')
print('Print 3:',list_ex)
print('Print 4:',list_ex[3])  # print the 4th value in the list 'list_ex'
```

**3. Loops**

So far, we have learned:
    - for loops  (repeats a block of code the number of times described in the "for" statement)
    - while loops (repeats a block of code as long as a certain condition is true.)

```python
for value1 in list_ex:   # loop through all the entries in list "list_ex"
    print('Current entry in variable value is:', value1)  # for each iteration, variable named "value1" 
                                                        #      will be assigned the next entry in "list_ex"
```

```python
for index1 in range(len(list_ex)): # loop through integers from 0 to length of list "list_ex"
                                # for each iteration, variable named "index1"
                                #      will be assigned the next integer in 0 to length of list "list_ex"
    str_now = list_ex[index1]   # assign a variable the content of the index1-th entry of list "list_ex"
    print('The',index1,'entry in list_ex is',str_now)
```

```python
index1 = 0
while index1 < len(list_ex):  # perform a while loop until index1 is equal to or greater than the length of list "list_ex"
    str_now = list_ex[index1]
    print('The',index1,'entry in list_ex is',str_now)
    index1 += 1               # increment whatever is in index1 by +1
    # Note this is the identical result as the for loop in cell above
```

**4. Conditionals**

If statements allow you to examine the current state of an algorithm and respond to that state.

```python
height = 5
if height >= 4:
    print("you are tall enough to ride the roller coaster")
```

```python
height = 2.23        # modify this variable to satisfy different conditions below
if height >= 4:
    print("you are tall enough to ride the roller coaster")
elif height > 3:   #  If height does not meet above conditions, and is higher than 3, do this
    print('you can ride the tea cups')
elif height >= 2:   #  If height does not meet above conditions, and is greater than or equal to 2, do this
    print('you can ride the choo-choo')
else:               #  If height does not meet any of the above conditions, do this
    print('you need to grow some first')
```

```python
# Conditionals with lists
if 'no' in list_ex:
    print('they do not have bananas today')
else:
    print('bananas for everyone')
    
```

```python
# Using multiple conditions
if 'no' in list_ex and height>4:
    print('They have no bananas and you are tall enough to ride the roller coaster')
elif 'no' in list_ex or height>4:
    print('They either have no bananas or you are tall enough to ride the roller coaster')
else:
     print('This is a silly example, yes bananas and too short for roller coaster')   
```

**5. Functions**

Functions are defined blocks of code designed to do one specific job. Functions allow you to write code once that can then be run whenever you need to accomplish the same task. Functions are defined with specific inputs and can return the information generated.  What happens in a function, stays in a function.  Only items on the `return` line leave the function.

```python
def say_happy_bday():        # Define a simple function that just runs and returns nothing
    print("Happy Birthday")  # While this function is running, print this phrase
    

say_happy_bday()       # Call the function and see that it did its job, it printed this phrase
say_happy_bday()       # Call the function again
```

```python
def say_happy_bday_2(name1):        # Define a simple function with one input
    print("Happy Birthday", name1)  # While this function is running, print this phrase
    
say_happy_bday_2('aardvark')  

# To be clear, 'name1' as a variable does not exist out here, outside the function
print(name1) # this will fail, try it
```

```python
def say_happy_bday_3(name1='giraffe'):        # Define a simple function with one input, assign a default value in case this is called without a variable
    print("Happy Birthday", name1)  # While this function is running, print this phrase
    

say_happy_bday_3('aardvark')
say_happy_bday_3()           # Call function without a input, will revert to default value
say_happy_bday_3('giraffe2')
```

```python
def do_some_math(v1,v2,v3):    # Define a function with 3 inputs
    # do some fun math and make some new variables
    v4 = v1+v2*v3
    v5 = v4+1
    return v4, v5   # Return two variables
```

```python
# Call this new function and assigns whatever it returns to new variables
outside_func_v1, outside_func_v2 = do_some_math(32,123,12)
print('Function returned:',outside_func_v1, 'and',outside_func_v2)

list1 = [32, 123, 12]
outside_func_v1, outside_func_v2 = do_some_math(list1[0],list1[1],list1[2])
print('2nd time, function returned:',outside_func_v1, 'and',outside_func_v2)

print('Function returned:',v4, 'and',v5)  # this line will fail because v4 and v5 only exist inside the function
```
