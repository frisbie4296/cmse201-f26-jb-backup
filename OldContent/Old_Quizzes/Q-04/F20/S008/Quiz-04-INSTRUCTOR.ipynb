One of the earliest applications of differential equations and optimization in modern marketing was in 1969 when Frank Bass published a paper modeling the spread of a product through a population. This technique, now known as the Bass diffusion model, has been used for decades as a forecasting technique in marketing, advertising, and economics. It takes the following mathematical form:

$$\frac{f(t)}{1-F(t)} = p + q F(t)$$

The model indicates that the number of new adopters of a product at a given time, will be proportional to the number of adopters at the previous time, as well as the number of innovators, or people trying to spread the product. These effects are represented by the constants $p$ and $q$ above.

The **coefficient of innovation**, $p$, is the effect from advertising and "influencers" and the **coefficient of imitation**, $q$, is the effect from word-of-mouth and followers.

The evolution of new adopters might then look like the following:

<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/d/d0/Bass_new_adopters.svg/1200px-Bass_new_adopters.svg.png" width=300px>
</div>

For this quiz, you'll be using data from [**Google Trends**](https://trends.google.com/trends/) to gauge the spread of popularity for something over a period of time.

The data has been provided for you and should be available on D2L. Specifically, you will be working with cumulative search volume index (CSVI), which is the total search volume for a specific term up until a certain time. The CSVI can be an indicator of the number of adopters.

It turns out the Bass model can predict this behavior in the following way:

$$A(t) = M\frac{1 - e^{-(p+q)t}}{1 + \frac{q}{p}e^{-(p+q)t}}$$

This uses the same $p$ and $q$ values from above, as well as a new parameter $M$, which is the expected market cap, or maximum number of adopters that can be reached.  For this model, we expect the evolution of adopters to behave like so:

A popular and contagious product in 2017 was the fidget spinner. You're going to use the provided `fidget.csv` file from Google Trends and the Bass "Adopter" model to observe the actual and expected spread of the product in terms of cumulative search volume index (CSVI)!

```python
# Read in the data using pandas here, and show the first few rows
```

```python
### ANSWER ###
df = pd.read_csv("fidget.csv")
df.head(10)
```

&#9989;&nbsp; **Question 2 [3 points]**: Now that you've loaded the data, **define a function for "$A(t)$"**, which represents the Bass diffusion model for adopters like so:

$$A(t) = M\frac{1 - e^{-(p+q)t}}{1 + \frac{q}{p}e^{-(p+q)t}}$$

**Write a function called `bass` in the cell below that returns the value of $A$ at a given time, $t$. The function should take as input t, p, q, and M**. Make sure to use the NumPy `np.exp()` function to ensure that your function will work for an *array* of time values!

```python
# Define the bass function here
```

```python
### ANSWER ###
def bass(t, p, q, m):
    return m * ((1 - np.exp(-1 * (p + q) * t)) / (1 + (q / p) * np.exp(-1 * (p + q) * t)))
```

&#9989;&nbsp; **Question 3 [5 points]**: Your next step is to use Scipy's `curve_fit` function to estimate your parameter values using the Google Trends data. If necessary, you should review the documentation for `curve_fit`, which you can find [here](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.curve_fit.html).

You can use the "Day" column as the time value and "CSVI" as your "$A$" value.

**Use `curve_fit` to estimate the values of the parameters in the model: p, q, and M.**

Once you have your estimated parameter values, **print them**.

(Note: if you get an "RuntimeWarning: overflow" message. this can sometimes safely be ignored, so if the it seems like the code ran fine otherwise, move on) 

```python
# Call the curve_fit function here

# Print your best-fit parameter values here
```

```python
### ANSWER ###
popt, pcov = curve_fit(bass, df["Day"], df["CSVI"])

innovation = popt[0]
imitation = popt[1]
max_potential = popt[2]
print("Innovation: %f, Imitation: %f, Market Cap: %f" %(innovation,imitation,max_potential))
```

&#9989;&nbsp; **Question 4 [3 points]**: Now that you've used `curve_fit` to estimate the parameters values, compute the expected values for $A(t)$ using these best-fit parameters, **plot the actual data along with the expected the model values in the cell below.** Label your axes and include a legend to make it clear what part of the plot represents the data and what part represents the model.

```python
# Put your plot code here
```

```python
### ANSWER ###
sns.set_style("ticks")
sns.set_context("notebook")
y_pred = bass(df['Day'], innovation, imitation, max_potential)
plt.figure(figsize=(8,5))
plt.scatter(df["Day"], df["CSVI"],label="Data")
plt.plot(df["Day"], y_pred, color="orange",label="Model")
plt.xlabel("Time [days]")
plt.ylabel("CSVI")
plt.legend()
plt.grid()
sns.despine()
```
