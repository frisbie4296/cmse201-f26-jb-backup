```python
### ANSWER ###
def coupled_ho(t, curr_vals, constants):
    """
    """
    
    x1, v1, x2, v2 = curr_vals
    m1, m2, k1, k2, b1, b2, q = constants
    
    dx1dt = v1
    dv1dt = -(b1*v1 + (k1 + q) * x1 - q * x2 )/m1
    
    dx2dt = v2
    dv2dt = -(b2*v2 + (k2 + q)* x2 - q * x1)/m2
    
    return [dx1dt, dv1dt, dx2dt, dv2dt]
```

## Part 2.3 Fitting [10 points]

Let's now check that physics textbooks are correct! They claim that the solution of damped coupled harmonic oscillators is 

$$ x_{1,2}(t) = e^{- t/\tau } \left [  A_{1,2} \cos( \Omega_+ t) + B_{1,2} \cos( \Omega_- t)\right ] $$

&#9989;&nbsp; **Task:** Write a function called `dho_sol` that computes the equation of $x_1(t)$. You will be fitting the above equation to your `solve_ivp` solution. Note that the equation for $x_1$ and $x_2$ are the same the only thing that changes are the constants in front. 

**IMPORTANT:** This function will be used by `curve_fit`, hence, make sure the order of the inputs is the correct one.

- Make sure the inputs are in the correct order **[2 points]**.
- Write the correct equations **[2 points]**.
- Make sure you are returning the correct function **[2 points]**.

```python
# Put your code here
```

```python
### ANSWER ###
def dho_sol(t, tau, omega1, omega2, A, B):
    positive = A * np.exp(-t/tau) * np.cos( omega1 * t)
    negative = B * np.exp(-t/tau) * np.cos( omega2 * t)
    return positive + negative
```
