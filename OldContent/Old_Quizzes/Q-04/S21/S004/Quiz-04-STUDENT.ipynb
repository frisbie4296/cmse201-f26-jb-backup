### The Bass Diffusion Model


One of the earliest applications of modeling in modern marketing is  Frank Bass' 1969 paper describing the spread of a product through a population. This technique, now known as the Bass diffusion model, has been used for decades as a forecasting technique in marketing, advertising, and economics. The  model takes the mathematical form:

$$A(t) = M\frac{1 - e^{-(p+q)t}}{1 + \frac{q}{p}e^{-(p+q)t}}$$

where $A(t)$ is the number of adopters of the product after time $t$, the **coefficient of innovation**, $p$, is the effect from advertising and "influencers",  the **coefficient of imitation**, $q$, is the effect of  word-of-mouth and followers, and $M$ is the expected market cap, or maximum number of adopters that can be reached.  For this model, we expect the evolution of adopters to behave like so:
