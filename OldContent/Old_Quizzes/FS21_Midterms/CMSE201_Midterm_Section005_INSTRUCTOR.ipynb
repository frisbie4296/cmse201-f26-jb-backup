### &#9989;&nbsp; Task 2.2: Calculating the Golden Angle (5 points)

You learn that the sunflower pattern is based on the Fibonacci sequence because the angle separating each petal from the next is the Golden Angle. 

To calculate the Golden Angle, you first start with the Golden Ratio, which is approximated by dividing element `i` by element `i-1`, becoming more accurate the farther `i` is in the sequence. From the approximation of the Golden Ratio, you approximate the Golden Angle as `360 / (golden_ratio)**2` (in degrees). Again, the Golden Angle approximation becomes more accurate the larger the value of `i`.

The Golden Angle in degrees is `137.50776405003785`

Fix the function below!

````python
# code for a function to calculate the golden angle and print out the calcualted value

def calculate_golden_angle(n): # the input n is how many elements of the Fibonacci sequence to calculate
    
    fibonacci = [0,1] # list to store Fibonacci sequence and first two elements
    golden_angle = [] # list to store approximations of the Golden Angle

    for i in range(2,n): # iterate from 2 up to n

        fibonacci.append(fibonacci[i-2] + fibonacci[i-1]) # calculate the next Fibonacci sequence element and append to list
        golden_ratio = fibonacci[i]/fibonacci[i-1] # calculate each element of the Golden Ratio
        golden_angle.append(360/golden_ratio**2) # calculate and append Golden Angle estimates

    golden_angle # return the list golden_angle
    
golden_angle = calculate_golden_angle(100) # use the function for 100 elements of Fibonacci sequence
print(golden_angle[-1]) # print the last, most accurate estimate of the Golden Angle from list golden_angle
````
