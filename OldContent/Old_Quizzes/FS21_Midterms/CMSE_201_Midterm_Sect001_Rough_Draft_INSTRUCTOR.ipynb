### Part 1: Calculating the Growth Rate of MSAs

First, calculate the growth rate of the different MSAs over the last ten years. The equation for calculating growth rate is:

\begin{equation}
\mathrm{Growth~Rate} = \frac{\mathrm{2020~Population} - \mathrm{2010~Population}}{\mathrm{2010~Population}}
\end{equation}

So, for example, the growth rate of New York would be:
\begin{equation}
\mathrm{NY Growth~Rate} = \frac{20140470 - 18897109}{18897109} = 0.0658
\end{equation}
Or 6.58%.

**Do this for all MSAs and append all growth rate values to a new list, titled `msa_growth`.**

````python
#Write your code here
````

````python
###ANSWER
msa_growth= []
for jj in range(len(msa_pop_2020)):
    growth_rate = (msa_pop_2020[jj]-msa_pop_2010[jj])/msa_pop_2010[jj]
    msa_growth.append(growth_rate)
````

### Part 3: Selecting MSAs by Region
Next, create a function that takes in your `all_msa_info` and a region, and returns all of the MSA lists that are in that region. So, for example, if I passed this function `all_msa_info` and `'East'`, it would return a list that looked like:

```
[['New York City', 20140470, 18897109, 'East', 0.0658],
['Philidelphia',...]
etc.
```

````python
#Write your code here
````

````python
###ANSWER

def get_regional_msas(msa_list,region):
    regional_msa_list = []
    for jj in range(len(msa_list)):
        if msa_list[jj][3] == region:
            regional_msa_list.append(msa_list[jj])
    return regional_msa_list
````
