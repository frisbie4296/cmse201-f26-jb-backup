### Improve your `chat_bot` function to deal with user questions (10 points)

Add an `if` block to your function that does the following:  
- if the query contains "why", respond "I'm not sure.. why do think?"
- if the query contains "what", respond "That's so interesting.  People have been pondering that for a long time."
- if the query contains "who", respond "It could be anybody, really."
- if the query contains "where", respond "Start by looking within yourself."
- if the query contains "how", respond "The world is indeed mysterious."
- if the query contains "when":
   - if the query contains "did" or "were", respond "Long ago."
   - if the query contains "will", respond "Someday!  I'm sure of that."
   - otherwise respond "I'm not sure."
- if the query didn't contain any of the above question words, respond "Wow!  Tell me more!"
   
**Hint #1: to check if a string is contained in another string, remember the `in` operator!**  

**Hint #2: to make things easier, convert the query to lowercase at the beginning of your function:**  
`query = query.lower()`

```python
# your code here
```

```python
### ANSWER ###
def chat_bot(query):
    query = query.lower()
    if "why" in query:
        response = "I'm not sure.. why do think?"
    elif "what" in query:
        response = "That's so interesting. People have been pondering that for a long time."
    elif "who" in query:
        response = "It could be anybody, really."
    elif "where" in query:
        response = "Start by looking within yourself."
    elif "how" in query:
        response = "The world is indeed mysterious."
    elif "when" in query:
        if "did" in query or "were" in query:
            response = "Long ago."
        elif "will" in query:
            response = "Someday! I'm sure of that."
        else: 
            response = "I'm not sure."
    else:
        response = "Wow!  Tell me more!"
        
    return response
```

```python
### ANSWER ###
import numpy as np
x = np.linspace(0,1,20)
y1 = np.exp(x)
y2 = 2*np.exp(x)
y3 = np.exp(2*x)
```

### Show them on the same plot, label them appropriately, as well as the axes (7 points)

```python
# code here
```

```python
### ANSWER ###
import matplotlib.pyplot as plt
plt.plot(x,y1,label='y=exp(x)')
plt.plot(x,y2,label='y=2*exp(x)')
plt.plot(x,y3,label='y=exp(2*x)')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
```
