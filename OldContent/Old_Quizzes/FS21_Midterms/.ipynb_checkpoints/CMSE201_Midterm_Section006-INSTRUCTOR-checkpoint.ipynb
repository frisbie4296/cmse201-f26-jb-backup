
### Improve your `age_fun` function to return more meaningful answers (10 points)

Add an `if` block to your function that does the following:  
- if the age is less than 5, respond "Play, play, and play!"
- if the age is between 5 (including 5) and 16 (not including 16), respond "Sleep, study, and take exams."
- if the age is between 16 (including 16) and 22 (not including 22), respond "Sleep, study, drive, and take exams."
- if the age is between 22 (including 22) and 65 (not including 65), respond "Working, working, working, working..."
- if the age is between 65 (including 65) and 200 (not including 200), respond "Boring, boring, boring, boring..."
- if the age isn't in any of the above interval, respond "Wow!  Call the scientis, now!!!"
 

```python
# your code here
5 <= 6
```

```python
### ANSWER ###
def age_fun(age):
    if age < 5:
        response = "Play, play, and play!"
    elif  5 <= age and age < 16:
        response = "Sleep, study, and take exams."
    elif 16 <= age and age < 22:
        response = "Sleep, study, drive, and take exams."
    elif 22 <= age and age < 65:
        response = "Working, working, working, working..."
    elif 65 <= age and age < 200:
        response =  "Boring, boring, boring, boring..."    
    else:
        response = "Wow! Call the scientis, now!!!"
        
    return response
```

## Part 4: Plotting exponential growth (14 points)

### Evaluate the following functions on the range $[-3,3]$ using at least 40 linearly spaced points for x (5 points)

$$y = \exp(x)$$
$$y = \frac{1}{\sqrt{2\pi}}*\exp(-\frac{x^2}{2})$$
$$y = \exp(2x)$$



```python
# code here
```

```python
### ANSWER ###
import numpy as np
x = np.linspace(-3,3,40)
y1 = np.exp(x)
y2 = (2 * np.pi)**0.5 *np.exp(-(x**2)/2)
y3 = np.exp(2*x)
```

### Show them on the same plot, label them appropriately, as well as the axes (7 points)

```python
# code here
```

```python
### ANSWER ###
import matplotlib.pyplot as plt
plt.plot(x,y1,label='y=exp(x)')
plt.plot(x,y2,label='y=(2 * np.pi)**0.5 *np.exp(-(x**2)/2)')
plt.plot(x,y3,label='y=exp(2*x)')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
```

### Show the same plot but using a `log` scale for the y axis (2 points)

```python
# code here
```

```python
### ANSWER ###
import matplotlib.pyplot as plt
plt.plot(x,y1,label='y=exp(x)')
plt.plot(x,y2,label='y=(2 * np.pi)**0.5 *np.exp(-(x**2)/2)')
plt.plot(x,y3,label='y=exp(2*x)')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.yscale('log')
```
