### &#9989;&nbsp; Question (10 points)
## Part 3a)
You are interested in how long it would take to watch movies you have been recommended. Given a list of movie lengths in minutes, find the total time to watch all the movies and display this result in both minutes and in hours. 

Your code should do the following:

* Use a loop to iterate through the list
* Keep a running total of minutes and a running total of hours
* Print the current total of minutes and the current total of hours right next to each other
* Display the final total in both minutes and hours


```python
#starter code
movie_lengths = [90, 105, 120, 135, 150] #units of minutes here
```

```python
### ANSWER
# Example starter code
movie_lengths = [90, 105, 120, 135, 150] #units of minutes here

total_minutes = 0
total_hours = 0
for i in movie_lengths:
    total_minutes += i
    total_hours += i/60
    print(total_minutes,total_hours)

# RUBRIC
# points for code that runs and produces the appropriate output
# points for reasoning
# points for resources
```
