# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)
Suppose you are a veterinarian.  You need to determine the appropriate care to give a pet based on the age of the animal.  For cats, the [age ranges](https://www.banfield.com/Senior-hub/What-to-expect/Life-stages-of-cats-and-dogs) are the following:

* **Kitten**: less than 1 year
* **Young adult**: 1-5.9 years
* **Mature adult**: 6-9.9 years
* **Senior**: 10 years or more

In the cell below, you will see a list of cat ages. Using `if`, `elif`, and `else` statements, determine the age category of each cat. Some starter code for navigating the list (and formatting the output) is given!


```python
# Example starter code
cat_ages = [6.0, 1.0, 8.3, 5.3, 0.6, 11.3, 6.4, 3.3, 14.1, 0.8]

for age in cat_ages:
    print(age,"is","---?which category?---")
```

```python
### ANSWER

for age in cat_ages:
    if age < 1:
        print(age,"is","a kitten!")
    elif age < 6:
        print(age,"is","a young adult!")
    elif age < 10:
        print(age,"is","a mature adult!")
    else:
        print(age,"is","a senior!")

    
# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 9 out of 10 for correct output but edge cases (age=6.0 or age=1.0) are incorrect
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You order a collection of nuts, bolts, and washers of different sizes.  You know the unit weight of each individual item.  In order to quickly ensure that you have the correct quantity of each, you will weigh the total order.  You need to find out what the total weight of the order should be, so you can use that to verify against the actual weight.

**Write some code to find the total weight of your order.**  Your code should do the following:
* Use a loop to iterate through the lists
* Multiply the weight of the components by the quantity of the components
* Compute the total weight of all the components you have
* Print the total weight


```python
# Example starter code
product_weights = [3.2, 10.6, 5.5, 2.7, 0.7, 3.7, 4.4, 8.2]  #in grams
product_quantities = [150, 300, 225, 350, 600, 450, 100, 75]
```

```python
### ANSWER

total = 0
for i in range(len(product_weights)):
    total += product_weights[i]*product_quantities[i]
    print(total)

# RUBRIC
# 3 pts - correct for loop
# 3 pts - multiply value * number
# 2 pts - update total
# 2 pts - print final total
```
