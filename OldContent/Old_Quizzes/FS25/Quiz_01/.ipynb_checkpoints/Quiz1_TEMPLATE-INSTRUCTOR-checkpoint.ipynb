```python
# put your code here

batches = ["Batch 1", "Batch 2", "Batch 3"]
for batch in batches:
    print("Preparing", batch)
```

```python
### ANSWER

batches = ["Batch 1", "Batch 2", "Batch 3"]
for batch in batches:
    print("Preparing", batch)

i=0
while i<len(batches):
    print("Preparing",batches[i])
    i+=1

# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)
You are designing a piece of artwork to capture the changing temperatures throughout the year at MSU. You are working in degrees Celsius, and your color scale is:
* "Blue": -8 $^\circ$C to 0.9 $^\circ$C
* "Green": 1 $^\circ$C to 6.9 $^\circ$C
* "Yellow": 7 $^\circ$C to 12.9 $^\circ$C
* "Orange": 13 $^\circ$C to 18.9 $^\circ$C
* "Light Red": 19 $^\circ$C to 24.9 $^\circ$C
* "Dark Red": 25 $^\circ$C to 30.9 $^\circ$C

In the cell below, you will see a list of 12 temperatures. Using `if`, `elif`, and `else` statements, determine what color you should use for each day. Some starter code for navigating the list (and formatting the output) is given!

### &#9989;&nbsp; Metacognition
## Part 2b) Documenting Your Solution Pathway (10 points)

**Describe in detail** how you solved the problem including any external resources you consulted (e.g. generative AI **prompts and outputs**, past assignments, Code Portfolio, Stack Overflow, Google, etc.) and how they informed your approach. You must also describe how you know your solution works (e.g. and calculations you made, tests you ran, etc.)

If you did not use external resources, describe your thought process for solving the problem by answering the following questions:
1. How did you know where to start?
2. What prior knowledge did you recall to solve the problem?
3. How do you know that your solution works?

```python
# Example starter code
temperatures = [-1,2,5,8.5,17,21.5,25,22,20,12,5,1]

for temp in temperatures:
    print(temp,"is","example color","today!")
```

```python
### ANSWER

temperatures = [-1,2,5,8.5,17,21.5,25,22,20,12,5,1]

for temp in temperatures:
    if temp > -8 and temp < 0.9:
        print(temp,"is","Blue","today!")
    elif temp > 0.9 and temp < 7 :
        print(temp,"is","Green","today!")
    elif temp > 7 and temp < 12.9:
        print(temp,"is","Yellow","today!")
    elif temp > 12.9 and temp < 18.9:
        print(temp,"is","Orange","today!")
    elif temp > 19 and temp < 24.9:
        print(temp,"is","Light Red","today!")
    else:
        print(temp,"is","Dark Red","today!")
# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

You're working for a farmer's market and need to calculate the total sales for different products. Write a program to compute the total sales based on the prices and quantities of items sold.

```python
# Example starter code
product_prices = [3.99, 1.99, 5.49, 2.79]
product_quantities = [12, 25, 10, 20]
```

```python
### ANSWER
# Example starter code
product_prices = [3.99, 1.99, 5.49, 2.79]
product_quantities = [12, 25, 10, 20]

total = 0
for i in range(len(product_prices)):
    total += product_prices[i]*product_quantities[i]
    print(total)

# RUBRIC
# points for code that runs and produces the appropriate output
# points for reasoning
# points for resources
```
