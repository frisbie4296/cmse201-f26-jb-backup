```python
# put your answer here
```

```python
### ANSWER

exercises = ["Push-ups", "Sit-ups", "Jumping jacks"]
for exercise in exercises:
    print("Next exercise:", exercise)
    
    
# while loop
i = 0
while i < len(exercises):
    print("Next exercise:", exercises[i])
    i += 1


# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)

You are tracking the heart rates of participants during 12 workout sessions. Your goal is to classify each heart rate into an exercise intensity category based on the following scale:

* “Resting”: less than 60 bpm

* “Light”: 60–99 bpm

* “Moderate”: 100–139 bpm

* “High”: 140–169 bpm

* “Maximum”: 170–190 bpm

* “Above Max”: greater than 190 bpm

Below, you will find a list of 12 heart rates. Using `if`, `elif`, and `else` statements, determine the exercise intensity for each session. Some starter code is provided to help you navigate the list and format the output.

```python
# Example starter code

heart_rates = [55, 78, 102, 120, 145, 160, 180, 135, 110, 95, 70, 60]

for hr in heart_rates:
    print("Heart rate", hr, "bpm is", '"example"', "intensity today!")

```

```python
# put your answer here
```

```python
### ANSWER
# Heart rates for 12 workout sessions
heart_rates = [55, 78, 102, 120, 145, 160, 180, 135, 110, 95, 70, 60]

for hr in heart_rates:
    if hr < 60:
        intensity = "Resting"
    elif 60 <= hr <= 99:
        intensity = "Light"
    elif 100 <= hr <= 139:
        intensity = "Moderate"
    elif 140 <= hr <= 169:
        intensity = "High"
    elif 170 <= hr <= 190:
        intensity = "Maximum"
    else:
        intensity = "Above Max"

    print("Heart rate", hr, "bpm is", intensity, "intensity today!")

# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You are managing a bookstore and need to calculate the total revenue for different books. Each book has a price and the number of copies sold. Write a program that uses a loop to compute the total revenue.

Below, you will find two lists: one for prices and one for quantities sold. Using a loop, calculate the total revenue for all books. Some starter code is provided to help you get started.

```python
# Prices of books in dollars
book_prices = [12.99, 8.5, 15.75, 22.0]

# Number of copies sold for each book
copies_sold = [5, 12, 7, 3]

# TODO: Use a loop to calculate revenue for each book
# and accumulate into total_revenue
```

```python
# put your answer here
```

```python
### ANSWER

# Prices of books in dollars
book_prices = [12.99, 8.5, 15.75, 22.0]

# Number of copies sold for each book
copies_sold = [5, 12, 7, 3]

# Initialize total revenue
total_revenue = 0  # 2 points

# Method 1: Use a for loop to calculate total revenue
for i in range(len(book_prices)):  # 2 points
    revenue = book_prices[i]*copies_sold[i] # 2 points
    total_revenue +=revenue # 2 points
    print(f"Book {i+1}: ${book_prices[i]} x {copies_sold[i]} copies = ${revenue:.2f}") # 2 points
print("Total revenue: $", round(total_revenue,2)) # 2 points
    
#Method 2: Use a while loop to calculate total revenue
c = 0 # 1 points 
total_revenue2 = 0 # 1 points

while c <  len(book_prices): # 2 points
    revenue2 = book_prices[c]*copies_sold[c] # 2 points
    total_revenue2 +=revenue2 # 1 points
    print(f"Book {c+1}: ${book_prices[c]} x {copies_sold[c]} copies = ${revenue2:.2f}") #1 points
    c +=1 # 1 points
print("Total revenue: $", round(total_revenue2,2))# 1 points
```
