# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)
You are categorizing wave frequencies. These are your categories:


- **Infrasonic**: Below 20 Hz  
- **Audible**: 20 Hz to 19,999 Hz    
- **Radio Frequency (RF)**: 20,000 Hz to 299,999,999 Hz  
- **Microwave**: 300,000,000 Hz to 300,000,000,000 Hz


In the cell below, you will see a list of 6 frequencies. Using `if`, `elif`, and `else` statements, determine what category you should use for each frequency. Some starter code for navigating the list (and formatting the output) is given!


```python
# Example starter code

frequencies = [1046, 852219, 138161045091, 177751907582, 203235549782, 12]


for freq in frequencies:
    print(freq,"is","this category")
```

```python
### ANSWER

frequencies = [1046, 852219, 138161045091, 177751907582, 203235549782, 12]


for freq in frequencies:
    cat='none'
    if freq < 20:
        cat = 'Infrasonic'
    elif freq >= 20 and freq < 20000:
        cat = 'Audible'
    elif freq >= 20000 and freq < 300000000:
        cat = 'RF'
    else:
        cat = "Microwave"
    print(freq,"is",cat)

# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```
