# 1. Problem Context: Converting between `for` loops and `while` loops 

### &#9989;&nbsp; Question (10 points)
## Part 1a)
You set a reading goal of finishing **1000 pages this month**. You’ve already chosen the stack of books you want to read, and you know how many pages each one has. You want to figure out how many of those books you need to finish in order to meet your monthly goal.  

**Task: Rewrite the provided `for` loop as a `while` loop, maintaining the same output.**

```python
# list of page counts for different books (DO NOT CHANGE!)
book_pages = [120, 95, 310, 220, 150, 450, 200, 180, 95, 300, 210, 190]
print('I have', len(book_pages), 'books.')
```

```python
# for loop 
total_pages = 0
for i in range(len(book_pages)):
    if total_pages < 1000:
        total_pages = total_pages + book_pages[i]
        print("After finishing Book", i+1, "I have read", total_pages, "pages in total.")
```

```python
# put your answer here
```

```python
### ANSWER
total_pages = 0
i = 0
while total_pages < 1000:
    total_pages = total_pages + book_pages[i]
    print("After finishing Book", i+1, "I have read", total_pages, "pages in total.")
    i += 1

# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

```python
### ANSWER

animals = ["Tortoise", "Elephant", "Pig", "Kangaroo", "Cheetah", "Horse", 
           "Squirrel", "Greyhound", "Cat", "Ostrich", "Rabbit", "Lion"]

speeds = [0.3, 15, 11, 25, 70, 40, 12, 45, 30, 43, 35, 50]  # mph

for i in range(len(animals)):
    if speeds[i] <= 10:
        category = "Slow"
    elif speeds[i] <= 25:
        category = "Moderate"
    elif speeds[i] <= 45:
        category = "Fast"
    else:
        category = "Very Fast"
    print(animals[i], "runs at", speeds[i], "mph and is", category)
    
# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You are running a coffee shop and need to calculate the total revenue for different drinks sold in one day. Each drink has a unit price, and you know how many of each drink was sold.  

Write a program to compute the total revenue. Your code should do the following:  

- Use a `for` loop to iterate through the lists  
- Multiply the price of each drink by the quantity sold  
- Compute the total revenue from all drinks  
- Print the cumulative revenue as you go

```python
# Example starter code
drink_prices = [2.50, 3.75, 4.25, 5.50, 1.95]  # price of drinks in dollars
drink_quantities = [30, 20, 15, 10, 25]        # how many of each drink sold
```

```python
# put your answer here
```

```python
### ANSWER
drink_prices = [2.50, 3.75, 4.25, 5.50, 1.95]
drink_quantities = [30, 20, 15, 10, 25]

total = 0
for i in range(len(drink_prices)):
    revenue = drink_prices[i] * drink_quantities[i]
    print("Revenue from drink", i+1, ":", revenue)
    total += revenue

print("Total revenue from all drinks:", total)

# RUBRIC
# 3 pts - correct for loop
# 3 pts - multiply value * number
# 2 pts - update total
# 2 pts - print final total
```
