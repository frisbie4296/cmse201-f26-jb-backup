# 1. Problem Context: Converting between `for` loops and `while` loops

### &#9989;&nbsp; Question (10 points)
## Part 1a)
You are leading a Great Lakes field team deploying buoys (A buoy is a floating device drift with ocean currents). You need to log each buoy's status report in order. 

Rewrite the **provided `for` loop** as a **`while` loop**, maintaining the same output (same messages, same sequence).

```python
# Provided for loop
buoy_ids = [101, 102, 103, 104, 105, 106]
for bid in buoy_ids:
    print("Logging buoy", bid, "status: OK")
```

```python
### ANSWER

i = 0
buoy_ids = [101, 102, 103, 104, 105, 106]
while i < len(buoy_ids):
    print("Logging buoy", buoy_ids[i], "status: OK")
    i += 1


# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)


Below is a list of 10 Water Quality Index (WQI) daily averages from different sampling sites.   

**Your task:** Write a program using `if`, `elif`, and `else` statements to classify each site based on its WQI value and print the corresponding advisory level. Starter code is provided to help you navigate the list and format the output.

Advisory categories (based on daily WQI):
- Good: WQI 0–39  
- Caution: WQI 40–59  
- Warning: WQI 60–79  
- Hazard: WQI 80 or higher


```python
# Example starter code
wqi_values = [18, 35, 52, 68, 74, 82, 91, 27, 9, 4]

for wqi in wqi_values:
    print(wqi, "WQI -", "?advisory level?")
```

```python
### ANSWER

# Example starter code
wqi_values = [18, 35, 52, 68, 74, 82, 91, 27, 9, 4]

for wqi in wqi_values:
    if wqi >= 0 and wqi <= 39:
        print(wqi, "WQI -", "Good")
    elif wqi >= 40 and wqi <= 59:
        print(wqi, "WQI -", "Caution")
    elif wqi >= 60 and wqi <= 79:
        print(wqi, "WQI -", "Warning")
    else:  # wqi >= 80
        print(wqi, "WQI -", "Hazard")

# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You're managing an online bookstore and need to calculate the total revenue generated from book sales. 

Write a program that computes the **total revenue** based on the **book prices** and the **number of copies sold**.

```python
# Example starter code
per_sample_costs = [7.50, 12.00, 9.25, 6.00]  # per-sample costs at 4 sites ($)
samples_processed = [24, 18, 12, 30]  # number of samples processed at each site

print("Total Processing Cost: ?")
```

```python
### ANSWER

# Example starter code
per_sample_costs = [7.50, 12.00, 9.25, 6.00]  # per-sample costs at 4 sites ($)
samples_processed = [24, 18, 12, 30]  # number of samples processed at each site

total_cost = 0
for i in range(len(per_sample_costs)):
    total_cost += per_sample_costs[i] * samples_processed[i]

print("Total Processing Cost:", total_cost)

# RUBRIC
# 10 out of 10 for a correctly formatted loop that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```
