In statistics,the Pearson correlation coefficient (r) measures the strength and direction of a linear relationship between two variables, x and y. The formular is as follows:
$$ r = \frac{\sum (x_i - \bar{x})(y_i -\bar{y})}{\sqrt{\sum (x_i-\bar{x})^2 * \sum(y_i - \bar{y})^2}} $$
where $\bar{x}$ and $\bar{y}$ represent the mean of array x and y, the $\sum$ (Greek letter sigma) represents the summation operator, indicating that the expression should be summed over all data points.
The value of r ranges between:
  -  +1 → Perfect positive correlation
  - -1 → Perfect negative correlation
  -    0 → No linear correlation

```python
# test data
import numpy as np
test_x = np.arange(0,10,1.0) # indices test data
test_y = [215250., 211845., 515280., 517740., 123660.,  82260., 136275., 147885., 122490., 92880.] # Stream time (minutes) test data


# PUT YOUR FUNCTION HERE
```

```python
### ANSWER ###
def pearson_corr(x, y):
    # TODO: Compute the mean of x and y
    mean_x = np.mean(x)
    mean_y = np.mean(y)

    # TODO: Compute numerator and denominator of the formula
    numerator = np.sum((x - mean_x) * (y - mean_y))
    denominator = np.sqrt(np.sum((x - mean_x)**2) * np.sum((y - mean_y)**2))

    # TODO: Return the correlation coefficient
    r = numerator / denominator
    return r

# --- Test your function ---

print("Pearson correlation coefficient:", pearson_corr(test_x, test_y))

#Rubric:
# 10 points for defining the function and having it return values
# 5 points for calculating the statistics
# 5 points for testing the function
```
