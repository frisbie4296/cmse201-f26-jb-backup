&#9989;&nbsp; **Question 1 (30 points)**: In this question, you're going to simulate a model of the double pendulum. The model is simplified by assuming that the pendula will only swing through small angles. 

The double pendulum system consists of two planar pendulums connected ended to end.  Each of the two components of the pendulum can be approximated as a uniform rectangular bar. The angles in radians that bars $1$ and $2$ make with a vertical reference are $\theta=y_1$ and $\phi=y_3$, respectively, and the corresponding angular velocities (how fast the angles change in time in radians/second) are $\dot{\theta}=y_2$ and $\dot{\phi}=y_4$. Knowing these two angles, we can completely describe the motion of the system at any time instant. 
 Assuming the pendulum system only undergoes small motions, we obtain

$$
\begin{align}
\frac{dy_1}{dt} &= y_2 \\
\frac{dy_2}{dt} & = -C_1 y_1 + C_2 y_3 \\
\frac{dy_3}{dt} &= y_4 \\
\frac{dy_4}{dt} &= C_3  y_1 - C_4 y_3
\end{align}
$$

where the constants are given by (don't worry, you don't have to code these, they will be provided in a code cell)
$$
\begin{align*}
C_1 &= \frac{g(\ell_1 m_2 + m_1 \bar{r}_1)(\bar{I}_2 + m_2 \bar{r}_2^2)}{\bar{I}_2 L_1^2 m_2 + I_1(\bar{I}_2 + m_2 \bar{r}_2^2)}; C_2 &= \frac{g \ell_1 m_2^2 \bar{r}_2^2}{I_1 \bar{I}_2 + \bar{I}_2 L_1^2 m_2 + I_1 m_2 \bar{r}_2^2} \\
C_3 &= \frac{g \ell_1 m_2 (\ell_1 m_2 + m_1 \bar{r}_1)\bar{r}_2}{\bar{I}_2 \ell_1^2 m_2 + I_1(\bar{I}_2 + m_2 \bar{r}_2^2)}; C_4 &= \frac{g m_2 (I_1 + \ell_1^2 m_2) \bar{r}_2}{\bar{I}_2 \ell_1^2 m_2 + I_1 (\bar{I}_2 + m_2 \bar{r}_2^2)}.
\end{align*}
$$
