#### &#9989;&nbsp; **Question 2.1 (10 points):** 

Taking another look at your energy bill, you want to model your gas usage over the course of a year.

Using the `Month` column and the `Gas Usage (kWh)` column from the full dataset from Part 1 and the provided model below, use `curve_fit` to find best fit parameters for *a, b, c* and *d*. Make sure you give `curve_fit` a good starting point for finding the best fit parameters using the `p0` argument and these starting values:
* a = 5
* b = 0.5
* c = 6.0
* d = 8

Once you've found the best values, **print them out**. You should find that they are close to, but not the same as, the starting values.

```python
# Will need to use NumPy for the function
import numpy as np

# Will need to import curve_fit from SciPy
from scipy.optimize import curve_fit

# This function calculates energy usage corresponding to time
def sine_wave(month,a,b,c,d):
    usage = a*np.sin(b*month+c)+d
    return usage
```

```python
# Put your code here
```

```python
### ANSWER
params,ignore = curve_fit(sine_wave,Bill['Month'],Bill['Gas Usage (Mcf)'],p0=[250,0.5,6.0,500])
print(params)
                          
# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for p0 correct
# 5 pts for printing
```

Make sure to label your axes and include a legend to communicate what parts of the plot correspond to the model versus the data.

```python
# Make sure we import matplotlib
import matplotlib.pyplot as plt

# Put your code here
```

```python
### ANSWER ###
plt.scatter(Bill['Month'],Bill['Gas Usage (Mcf)'],label='data')
plt.plot(Bill['Month'],sine_wave(Bill['Month'],params[0],params[1],params[2],params[3]),label='fit')
plt.legend()

# Rubric
# 10 pts for data and best fit
# 10 pts for labels, legend, etc.
```
