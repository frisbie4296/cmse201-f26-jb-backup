#### &#9989;&nbsp; **Question 2.1 (10 points):** 

Taking another look at the daylight hours data, you want to model the hours of daylight over the course of a year.

Using the `DayOfYear` column and the `DaylightHours` column from the full dataset from Part 1 and the provided model below, use `curve_fit` to find best fit parameters for *a, b, c* and *d*. Make sure you give `curve_fit` a good starting point for finding the best fit parameters using the `p0` argument and these starting values:
* a = 3
* b = 0.01
* c = -1
* d = 12

Once you've found the best values, **print them out**. You should find that they are close to, but not the same as, the starting values.

```python
# Will need to use NumPy for the function
import numpy as np

# Will need to import curve_fit from SciPy
from scipy.optimize import curve_fit

# This function calculates energy usage corresponding to time
def sin_wave(day,a,b,c,d):
    usage = a*np.sin(b*day+c)+d
    return usage
```

```python
# Put your code here
```

```python
### ANSWER
params,ignore = curve_fit(sin_wave,EL_sun['DayOfYear'],EL_sun['DaylightHours'],p0=[3,0.01,-1,12])
print(params)
                          
# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for p0 correct
# 5 pts for printing
```
