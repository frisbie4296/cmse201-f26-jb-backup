```python
### ANSWER
def divisible_by_4_and_6(nums):
    new_nums = []
    for n in nums:
        if n % 4 == 0 and n % 6 == 0:
            new_nums.append(n)
    return new_nums

evens = list(range(2,128,2))

answer = divisible_by_4_and_6(evens)

print(answer)
```
