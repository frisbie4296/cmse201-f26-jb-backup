**Hint:** Given two integers `n` and `m`, the Python statement `n%m` returns the remainder of dividing `n` by `m`. For instace, `7%3`  returns the number `1`, because `7 = 3*2 + 1`.

```python
# Write your function here

#1. define the function/name with 2 input parameters : x (list), and n (integer) 
#2. initialize empty list with output answer
#3. for loop: for each element in the input list check if it is divisible by n or by n+3
#4. if so, append values to output list
#5. return answer

def divide(input_list, n):
    output_list =[]
    for val in input_list:
        if val%n==0 or val%(n+3)==0:
            output_list.append(val)
    return output_list

```
