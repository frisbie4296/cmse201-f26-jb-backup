```python
# Write your function here

def div_by_four_six(num_list):
    div_list = []
    for num in num_list:
        if num % 4 == 0 and num % 6 == 0:
            div_list.append(num)
    return div_list
```

```python
# Use this list of even numbers from 2 to 128 as your input to your function
evens = list(range(2,128,2))

# Call your function here

div_list = div_by_four_six(evens)

# Print the new list here

for num in div_list:
    print(num)
```
