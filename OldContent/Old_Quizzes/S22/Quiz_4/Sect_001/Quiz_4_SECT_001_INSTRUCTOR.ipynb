# Finding Planets

You are working for NASA looking for planets outside of our solar system. One way of finding these planets is to measure how fast stars are moving towards you and modeling the motion using a *periodic* function, which looks like the following:

$$ v(t) = A*\sin\left(\omega*t + C \right) + S$$


Where $v$ is the velocity (speed), $t$ is the time, $A$ is the called amplitude, $\omega$ is called the angular velocity, $C$ is called the phase, and $S$ is called the peculiar velocity. Stars moving fast enough (I.e., with a large enough amplitude) likely have a planet around them.

You’ve been given a dataset and you are tasked with finding the best fit amplitude.

#### &#9989;&nbsp; **Part 3 (30 points):**
Fit the provided data to the sinusoid function given above (I.e. find the best fit parameters).

```python
#Fit the sin function to the data here
```

```python
### ANSWER ###
import numpy as np
from scipy.optimize import curve_fit

def sin_func(t,A,P,C,S):
    return A*np.sin(t*P + C) + S


popt,pcov = curve_fit(sin_func,df["Time (Days)"],df["Velocity (m/s)"])
print(popt)
```

#### &#9989;&nbsp; **Part 4 (15 points):**
Create new x-data set (time) that has the same minimum and maximum value as the original x-data, but contains 100 equally spaced points.

```python
#Create a new x-data set here
```

```python
### ANSWER ###
x_new = np.linspace(np.amin(df["Time (Days)"]),np.amax(df["Time (Days)"]),100)
```

#### &#9989;&nbsp; **Part 5 (15 points):**
Now create a plot using the new x data with the fitted function. In the same plot display the original data. Be sure to include labels and a legend.

```python
#Make a plot showing your best fit line here
```

```python
### ANSWER ###
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(9,4))
y_best_fit = sin_func(x_new,*popt)
plt.scatter(df["Time (Days)"],df["Velocity (m/s)"],label='Data Points')
plt.plot(x_new,y_best_fit,c='r',label='Best Fit')
plt.xlabel("Time (Days)")
plt.ylabel("Velocity (m/s)")
plt.legend()
```
