&#9989;&nbsp; **Part 1.1 (10 points)** In the cell below, define a function named `prct_chg` that calculate the **percent change** of a development indicator from the initial year to each year. That is, given an array of N elements, the function should return the percent change of each element relative to the first element. Recall that the percent change from x to y is calculated as $ 100*(y-x)/x $. For example, if you have an array `[1,2,3,4]`, the returned array should be $[100*(1-1)/1, 100*(2-1)/1, 100*(3-1)/1, 100*(4-1)/1]$, which equals to $[0\%,100\%,200\%,300\% ]$.

```python
# define your function here
```

```python
### ANSWER ###
# Rubric:
#  5 points for defining a function with input and return
#  5 points for getting the percent change correct

def prct_chg(val):
    prct_val = (val -  val[0]) / val[0] * 100
    return prct_val
```

```python
# define your function here
```

```python
## ANSWER ##
# option 1: if statement
#  5 points for defining the function and return value
#  3 points for using the if statement correctly
#  2 points for appending the values correctly.
def greater_idx(A,B):
    idx = []
    for i in range(len(A)):
        if A[i]>B[i]:
            idx.append(i)
    return idx

# option 2: masking
def greater_idx(A,B):
    return np.where(A>B)
```
