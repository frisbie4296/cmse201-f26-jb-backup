# 4. While Loops (15 points)

In the cell below, there is a `for` loop and some lists that your friend wrote to keep track of the number of trees of each species they saw in the forest. When they gave you their code, you realized that their code might be more clear and efficient if it were written as a `while` loop instead.

In the empty cell below, rewrite the `for` loop as a `while` loop. The output should remain the same, but you can change any of the variable names.

```python
# a list of tree species you saw in the forest
tree_species = ["aspen", "basswood", "beech", "cedar", "cherry",
         "chestnut", "fir", "hemlock", "hickory", "maple"]

# a list of how many of each tree you saw in the forest
number_of_trees = [2,2,3,3,4,5,7,9,9,12]

count = 0
tree_count = 0

for i in tree_species:
    
    print('There are', number_of_trees[count], tree_species[count], 'in the forest.')
    
    tree_count += number_of_trees[count]
    
    count = count + 1
    
print('THERE ARE',tree_count,'TOTAL TREES IN THE FOREST.')
```

```python
### ANSWER
# One possible solution

# Full credit (15 points) for any while loop that runs while maintaining the same output
# Partial credit for a while loop that runs but does not keep the same output

i = 0
tree_count = 0
while i<len(tree_species):
    print('There are', number_of_trees[i], tree_species[i], 'in the forest.')
    tree_count += number_of_trees[i]
    i+=1
print('THERE ARE',tree_count,'TOTAL TREES IN THE FOREST.')
```
