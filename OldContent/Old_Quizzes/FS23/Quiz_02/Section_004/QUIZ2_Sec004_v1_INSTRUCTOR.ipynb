# 1. Writing a function (8 points)

You love plants! 💚🪴 You decide you want to make models of the beautiful geometries and forms of plants. You decide to start simple by modeling the growth of a leaf.

You believe a logistic growth function might be a good model for the length of a leaf over time. The logistic growth function is provided below:

$$L = \frac{F} {1+e^{-k\dot(t-m)}}$$

Where for **$L$ = leaf length in centimeters** as a function of **$t$ = time in days** there are the following constants:

$F$ = final length (in centimeters)   
$k$ = growth rate, steepness of the growth curve   
$m$ = midpoint of growth (in days)   

✅ **Write a function (6 points)**

- Write a function called `LeafLength` (**2 points**).
- Your function should take the following inputs: `t`, `F`, `k`, and `m` (**1 point**).
- **Use `t`, `F`, `k`, and `m` as your variable names** so that the next part of this question works (**1 point**).
- The function should return the length of a leaf as a function of time using the above equation (**2 points**). 

***Some hints:***
- You can get the value of $e$ using `np.e` or `math.e`
- Remember the operation for exponents is `**`

```python
# write your function LeafLength() here
```

```python
### ANSWER ###

def LeafLength(t, F, k, m):
    
    return F / (1 + np.e**(-k*(t-m)))
```

# 2. Visualizing a model using matplotlib (6 points)

Having developed a model for the growth of a leaf, you are eager to model more advanced aspects of plant morphology! 

You read about ***phyllotaxy***, or the arrangement of leaves and other organs in plants. Sometimes the angle separating leaves is 90$^\circ$, like a mint plant. Or, the angle can be 180$^\circ$ like maize. Many plants have spiral phyllotaxy, where plant organs are separated by the golden angle, approximately 137.507764$^\circ$.

You find a function to model the placement of leaves in a plant and decide to model a 🌻 sunflower (137.507764$^\circ$).

✅ **Do this (6 points):**

- Use the function `phyllotaxy` below to calculate $x$ and $y$ values to plot for your sunflower. Use the values `n = 1000` organs and `deg = 137.507764` degrees **(2 points)**
- Use a ***scatterplot*** **(2 points)**
- Adjust the ***size, color, and/or alpha*** of the points to make a pleasant looking plot. Be creative! **(1 point)**
- Give your plot a title of your choosing **(1 point)**

```python
# phyllotaxy function:
# Inputs: n = number of organs, deg = separation angle in degrees
# Outputs: a list of x and y values: xvals, yvals

def phyllotaxy(n, deg):
    
    xvals = [] # to store x vals
    yvals = [] # to store y vals
    
    for i in range(n): # for the number of organs 
    
        angle = i*deg # angle of current organ
        rads = angle*(np.pi/180) # convert from degrees to radians
        r = np.sqrt(i) # radius from center 

        x = r*np.cos(rads) # calculate x val
        y = r*np.sin(rads) # calculate y val

        xvals.append(x) # append x val to xvals
        yvals.append(y) # append y val to yvals

    return xvals, yvals # return list of x vals, list of y vals
    
```

```python
# make your sunflower SCATTERPLOT here

# keep this code
plt.gca().set_aspect("equal")
plt.axis("off")
```

```python
### ANSWER ###

xvals, yvals = phyllotaxy(1000, 137.507764)
plt.scatter(xvals, yvals, s=20, c="green", alpha=1)

plt.title("My sunflower")

# keep this code
plt.gca().set_aspect("equal")
plt.axis("off")
```
