# 3. For Loops and Debugging: Calculating 2022 Michigan crop values (50 points)

You work in the Department of Agriculture for the State of Michigan. Your boss wants you to calculate the value of some crops that they are interested in. You have looked up the crops they requested, the number of units harvested in 2022, and the price of each unit using this [resource](https://www.nass.usda.gov/Quick_Stats/Ag_Overview/stateOverview.php?state=MICHIGAN).

You have created 3 lists from the data (see below):

- One with the name of the crop your boss requested
- One with the number of units produced in Michigan in 2022 (in 1000s)
- One with the price per unit in Michigan in 2022 (in US dollars)

```python
# a list of crop names
crops = ["maple syrup", "oats", "pumpkins", "peppers", "asparagus",
         "beans", "cabbage", "tart cherries", "squash", "cucumbers"]

# a list of units (in 1000s) of the crop harvested in 2022
units = [188, 1830, 931, 450, 273.9, 1650, 1892, 178.9, 1642.5, 4920]

# a list of prices per unit (in dollars)
prices = [37.1, 4.9, 17.9, 50.3, 96.2, 18.8, 16.9, 0.203, 24.2, 13]
```

**Your boss needs the overall value of these crops to the State of Michigan in 2022 so they can complete a report.** To calculate the total value *for one crop*, the crops's price must be multiplied by the units produced. For example, the total value of maple syrup is \\$37.1 per unit $\times$ 188 units (in thousands) $=$ \\$6974.8 (in thousands).

Your boss started to write this bit of code (below), but part way through they **ran into an error** they couldn’t figure out. 

```python
crop_value_list = []

for i in len(crops):
    
    crop_value = units[i]*prices[i]
    
    crop_value_list.append(round(crop_value))
    
print(crop_value_list)
```
