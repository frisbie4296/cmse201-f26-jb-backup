# Drawing Star Shapes  (15 points)

One of your friend wants to have a Python function that allows him to draw various star shapes. To compute the points $(x,y)$ that determine a star shape, he wants to use this set of equations 
\begin{equation*}
\begin{aligned}
x =&\,\, t\,\cos(0.25\,\pi)-r\,\sin(0.25\,\pi) + 0.5, \\
y =&\,\, t\,\sin(0.25\,\pi)+r\,\cos(0.25\,\pi) + 0.5,
\end{aligned}
\end{equation*}
where 
\begin{equation*}
\begin{aligned}
t =&\,\, (0.5+0.15\,\sin(\alpha\,s))\,\cos(s), \\
r =&\,\, (0.5+0.15\,\sin(\alpha\,s))\,\sin(s),\\
&\,\,
\end{aligned}
\end{equation*}
$s$ goes from $0$ to $2\,\pi$ radians and $\alpha$ is a parameter. 
You propose to him to write down a Python code that will allow him to visualize these shapes.

&#9989;&nbsp; **Part 1 (1 point):** Import all the modules that you use in the cell below.

```python
# Put your code here! 
import math 
import matplotlib.pyplot as plt 
%matplotlib inline
import numpy as np
```

&#9989;&nbsp; **Part 2 (5 points):** 
You want to create a Python function, called `compute_s`, that computes the values of $s$. Your Python function takes the following input: the increment between two values of $s$. Your Python function returns the number of points and a list (or array) containing the values of $s$. 

__Hint__: the starting and ending values of $s$ are provided in the context above. 

```python
# Put your function here!
def compute_s(ds):
    s_start = 0
    s_end = 2*np.pi
    s = np.arange(s_start,s_end+ds,ds)
    nb_points = len(s)
    return nb_points, s
```

&#9989;&nbsp; **Part 3 (8 points):** 
You want to create a Python function, called `compute_star_shapes`, that computes the points $(x,y)$ of a given star shape. Your Python function takes the following inputs: the increment between two values of $s$ and the value of $\alpha$. The default value of $\alpha$ is $5$, meaning that your Python function should consider $\alpha = 5$ if this parameter is not specified.
Your Python function returns the number of points, a list (or array) containing the $x$ values and a list (or array) containing the $y$ values. 

__Hint__: you should use your function `compute_s`.

```python
# a possible solution with numpy arrays
def compute_star_shapes(ds, alpha = 5):
    nb_points, s = compute_s(ds)
    t = (0.5+0.15*np.sin(alpha*s))*np.cos(s)
    r = (0.5+0.15*np.sin(alpha*s))*np.sin(s) 
    x = t*np.cos(0.25*np.pi)-r*np.sin(0.25*np.pi) + 0.5 
    y = t*np.sin(0.25*np.pi)+r*np.cos(0.25*np.pi) + 0.5
    return nb_points, x, y
```

&#9989;&nbsp; **Part 3 (6 points):** In this part, you want to generate an example of a figure to your friend. Make multiple plots in a single figure that displays the star shape for $\alpha = 2,5,8,10$. We will use the increment between two values of $s$ to be $0.01$.
So, you should have four different plots. Each plot should contain a $x$-axis label, a $y$-axis label, and a title that specifies the value of $\alpha$. Each star shape should be of different color. In the **figure**'s title, you will specify the number of points that are used to generate the curves. Complete the code in the cell below. 



__Hint__: you should use your function `compute_star_shapes`.

```python
ds = 0.01 
# Put your code here!
plt.figure(figsize=(12,12))
nb_points, x, y = compute_star_shapes(ds, 2)
plt.suptitle('Star shapes with '+str(nb_points)+' points')  
plt.subplot(2,2,1) 
plt.plot(x, y, color='k') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('alpha=2')
plt.subplot(2,2,2) 
nb_points, x, y = compute_star_shapes(ds)
plt.plot(x, y, color='Purple') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('alpha=5')
plt.subplot(2,2,3) 
nb_points, x, y = compute_star_shapes(ds, 8)
plt.plot(x, y, color='g') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('alpha=8')
plt.subplot(2,2,4) 
nb_points, x, y = compute_star_shapes(ds, 10)
plt.plot(x, y, color='g') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('alpha=10')
```
