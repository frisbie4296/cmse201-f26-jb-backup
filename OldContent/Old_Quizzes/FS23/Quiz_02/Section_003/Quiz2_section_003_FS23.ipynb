Pearson's correlation coefficient is the covariance of the two variables divided by the product of their standard deviations and the formula is:
### $$ r = \frac{\sum\limits_{i=1}^{N} (x_{i}-\overline{x})(y_{i}-\overline{y})}{\sqrt{\sum\limits_{i=1}^{N} (x_{i}-\overline{x})^2(y_{i}-\overline{y})^2}}$$

where the symbols in this equation represent the following:

* $r$: Pearson correlation coefficient
* $\overline{x}$: Mean of the value of x-variable
* $\overline{y}$: Mean of the value of y-variable
* $x_{i}$: The value of x-variable at position $i$
* $y_{i}$: The value of y-variable at position $i$
