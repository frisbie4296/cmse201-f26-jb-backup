```python
brunch_items = ['coffee','donuts','cereal','juice','bagels','fruit','bacon','eggs','granola','yogurt']
item_cost = [20,30,4,8,24,10,12,5,6,9]

count = 0
total_cost = 0
for i in brunch_items:
    print('I spent',item_cost[count],'dollars on',brunch_items[count])
    total_cost += item_cost[count]
    count = count + 1
print('I spent a total of',total_cost,'dollars on brunch.')
```
