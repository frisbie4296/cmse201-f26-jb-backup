# 3. Loops [10 points]

Write code using a loop to compute the element-wise sum of two lists and store the result in a new list.

For example, given the lists `[1, 2, 3]` and `[4, 5, 6]`, the correct output should be `[1 + 4, 2 + 3, 3 + 6] = [5, 7, 9]`.

You're provided two example lists to check your work, but your code should work with any two lists of numbers.

```python
xs = [1, 2, 3]
ys = [4, 5, 6]

zs = []
# Your code here.
for i in range(len(xs)):
    z = xs[i] + ys[i]
    zs.append(z)

zs
```
