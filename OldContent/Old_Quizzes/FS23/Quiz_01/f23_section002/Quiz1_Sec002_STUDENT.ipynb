# 3. For Loops and Debugging: Calculating Tips (50 points)

You went out to have dinner with your friends. Each of you ordered different things and decided to tip differently. Now you need to figure out how much you tip all together. Relevant information are stored in three lists:
- One with everyone's name 
- One with dinner bill for each one of you
- One with preferred amount of tips for each one of you

```python
names = ["Me", "Joseph", "Lauren", "Carla", "Mike", "Qin"]
dinner_bills = [23, 40, 31, 37, 18, 26]
precent_tips = [20, 18, 20, 15, 25, 18]
```

**You and your friends need a piece of code that will calculate the total amount of tips you should pay.** To calculate the tip *for one person*, the percentage of the tip must be multiplied by the bill (e.g., Joseph tips 18% for a dinner bill of \\$40, which is 18% $\times$ \\$40 $=$  \\$7.2.) The total tip is the sum of the tip from each person.

Your friends started to write this bit of code (below), but part way through they **ran into an error** they couldn’t figure out. 

```python
total_tip = 0

for jj in range(names):
    individual_tip = dinner_bills[jj]*precent_tips[jj]/100
    total_tip += individual_tip
print(total_tip)
```
