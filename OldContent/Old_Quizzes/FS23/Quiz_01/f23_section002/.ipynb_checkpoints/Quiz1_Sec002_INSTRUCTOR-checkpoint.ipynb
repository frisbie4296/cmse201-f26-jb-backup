# 3. For Loops and Debugging: Calculating Tips (50 points)

You went out to have dinner with your friends. Each of you ordered different things and decided to tip differently. Now you need to figure out how much you tip all together. Relevant information are stored in three lists:
- One with everyone's name 
- One with dinner bill for each one of you
- One with preferred amount of tips for each one of you

```python
names = ["Me", "Joseph", "Lauren", "Carla", "Mike", "Qin"]
dinner_bills = [23, 40, 31, 37, 18, 26]
precent_tips = [20, 18, 20, 15, 25, 18]
```

**You and your friends need a piece of code that will calculate the total amount of tips you should pay.** To calculate the tip *for one person*, the percentage of the tip must be multiplied by the bill (e.g., Joseph tips 18% for a dinner bill of \\$40, which is 18% $\times$ \\$40 $=$  \\$7.2.) The total tip is the sum of the tip from each person.

Your friends started to write this bit of code (below), but part way through they **ran into an error** they couldn’t figure out. 

```python
total_tip = 0

for jj in range(names):
    individual_tip = dinner_bills[jj]*precent_tips[jj]/100
    total_tip += individual_tip
print(total_tip)
```

# 4. While Loops (15 points)

You and your friends are playing a card game. In the cell below, there is a `for` loop and some lists that your friend wrote to keep track of how many cards has been played. When they gave you their code, you realized that their code might be more clear and efficient if it were written as a `while` loop instead.


```python
card_ranks = ['Ace','2','3','4','5','6','7','8','9','10','Jack','Queen','King']
counts_per_rank = [1,3,4,0,1,1,4,2,3,0,1,1,0]

rank_count = 0
card_count = 0
for i in card_ranks:
    print('There are',counts_per_rank[rank_count],card_ranks[rank_count],'in the discard pile.')
    card_count+=counts_per_rank[rank_count]
    rank_count = rank_count + 1
print('There are',card_count,'total cards in the discard pile.')
```

In the empty cell below, rewrite the `for` loop as a `while` loop. The output should remain the same, but you can change any of the variable names.

```python
# Put your modified code here
```

```python
### ANSWER
# One possible solution

# Full credit (15 points) for any while loop that runs while maintaining the same output
# Partial credit for a while loop that runs but does not keep the same output

rank_count = 0
card_count = 0
while rank_count < len(card_ranks):
    print('There are',counts_per_rank[rank_count],card_ranks[rank_count],'in the discard pile.')
    card_count+=counts_per_rank[rank_count]
    rank_count += 1
print('There are',card_count,'total cards in the discard pile.')
```
