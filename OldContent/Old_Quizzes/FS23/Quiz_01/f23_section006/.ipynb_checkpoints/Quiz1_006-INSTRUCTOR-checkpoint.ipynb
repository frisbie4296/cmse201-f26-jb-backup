```python
item_names = ["Eggs", "Milk", "Bananas", "Poptarts", "Ground Beef", "Honey Crisp Apple","Black Beans"]
item_quantity = [2, 1, 5, 4, 3, 12, 6]
item_unit_prices = [1.89, 2.79, 0.56, 2.79, 5.69, 2.39, 0.79]
```

**You need a piece of code that will calculate the total cost for each shopping trip, so they can make up the budget.** To calculate the total cost *for one item*, the item's price must be multiplied by the quantity needed (E.g., the total cost of Eggs is \\$1.89 per dozen $\times$ 2 dozen eggs $=$ \\$3.78.) The total cost of the shopping trip is the sum of the costs of each item.

You started to write this bit of code (below), but part way through they **ran into an error** you couldn’t figure out. 

```python
all_items_total_cost = []

for j in range(item_quantity):
    item_total_cost = item_quantity[0]*item_unit_prices[0]
    all_items_total_cost.append(item_total_cost)
print(all_items_total_cost)
```

# 4. While Loops (15 points)

In the cell below, there is a `for` loop and some lists that your friend wrote to keep track of the number of species of birds they saw while bird watching. When they gave you their code, you realized that their code might be more clear and efficient if it were written as a `while` loop instead.

In the empty cell below, rewrite the `for` loop as a `while` loop. The output should remain the same, but you can change any of the variable names.

```python
bird_species = ['Cardinal', 'Chickadee','Blue Jay','Bald Eagle','Red-tailed Hawk','Great Blue Heron','America Gold-Finch','Robin','House Sparrow','Common Starling']
bird_counts = [4,6,2,1,3,2,9,5,20,12]

count = 0
bird_count = 0
for i in bird_species:
    print('I saw',bird_counts[count],bird_species[count],'birds today!')
    bird_count+=bird_counts[count]
    count = count + 1
print('I saw',bird_count,'total birds today!')
```

```python
### ANSWER
# One possible solution

# Full credit (15 points) for any while loop that runs while maintaining the same output
# Partial credit for a while loop that runs but does not keep the same output

i = 0
bird_count = 0
while i<len(bird_species):
    print('I saw',bird_counts[i],bird_species[i],'birds today.')
    bird_count+=bird_counts[i]
    i+=1
print('I saw',bird_count,'total birds today.')
```
