## 2. Reading and visualizing data [10 points]

There's a file on D2L called `weather.csv` which contains min and max temperatures (in Farenheit) in Michigan during the 1980s (Source: A subset of this [UMBS data](https://mfield.umich.edu/dataset/university-michigan-biological-station-weather-observations-1980-present).)

- Read in this data using Pandas.
- Plot the min and max temperatures on the same plot.
- Include appropriate axis labels and a legend.

If you can't read in this data, you can use the data from Question 1 for partial credit. In this case, plot temperature and snowfall on the same plot.

```python
# Your code here.
import matplotlib.pyplot as plt

weather = pd.read_csv("weather.csv")

plt.figure(figsize=(15, 5))
plt.plot(weather.Date, weather.Min_Temperature, label="Min Temp")
plt.plot(weather.Date, weather.Max_Temperature, label="Max Temp")
plt.legend()
plt.xlabel("Time")
plt.ylabel("Temperature (F)")
plt.xticks(rotation=75);
```

## 3. Fitting curves to data [10 points]

Use `curve_fit` from SciPy to fit the max temperature T as a function of time t to a function of the form

$$ T(t) = a \sin(b t + c) + d .$$

Plot the fit $T$ using the best-fit coefficients found by `curve_fit` in the same plot with the raw data.

If you couldn't load in the data, fit the temperature data from Question 1 to a line $T(t) = a t + b$, then plot the best-fit line with the raw data.

In both cases, take first temperature to be at time $t = 0$, the next temperature to be at time $t = 1$, etc. You can do this with `np.arange(len(dataframe))` where `dataframe` is the name of your Pandas data frame.

```python
# Your code here.
from scipy.optimize import curve_fit

ts = np.arange(len(weather))
ys = weather["Max_Temperature"]

def fit(t, a, b, c, d):
    return a * np.sin(b * t + c) + d

opt_params, _ = curve_fit(fit, ts, ys)

plt.plot(ts, ys)
plt.plot(ts, fit(ts, *opt_params));
```
