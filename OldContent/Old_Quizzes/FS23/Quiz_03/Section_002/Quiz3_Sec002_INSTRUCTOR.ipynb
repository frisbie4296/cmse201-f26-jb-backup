# 2. Fitting a model to data with `curve_fit` and visualizing the results (40 points)

You heard that the brain size is not proportional to the animal's body size. Rather, body mass and brain mass follow a [*power law*](https://en.wikipedia.org/wiki/Power_law) relationship:

$$
(brain\ mass) = a\times (body\ mass)^{k}
$$

where the power `k` is less than 1. 

You wonder if this is really true. To test this, you decided to fit the power law model to a subset of the brain size data. 

**Note**: for your convenience, you can simply ran the cell below to load this subset of data, stored in a dataframe `bb`.  

```python

# RUN THIS CELL to make sure you have the dataset loaded

import pandas as pd

bb = pd.read_json('{"brain_g": [  2.7       ,   7.8       ,  24.06      ,  83.9       ,95.3       ,  97.5       , 130.        , 159.2       ,   2.995447  ,   7.2       ,  14.2       ,  23.        ,31.08391333,  57.        ,  57.19      ,  59.08333333, 60.        ,  63.6       ,  93.7       ,  94.6       ],\
"body_g":[   77.6  ,   480.   ,   750.   , 10281.25 ,  4700.   ,  3452.   ,7710.   , 11500.   ,   113.   ,   400.   ,  1202.71 ,  1129.785,3470.   ,  4817.   ,  5065.   ,  6611.04 ,  5630.   ,  9525.   ,9000.   ,  6800.   ] }')
bb.sort_values("body_g",inplace=True)
bb.head()
```

#### &#9989;&nbsp; **Question 2.1 (15 points):** Fitting the power law model


Use `curve_fit` to the fit the power law model (provided in the cell below `brainmodel`) to the data provided above (contained in dataframe `bb`). **Print out** the fitted parameters for *a* and *k*. 


```python
# Will need to use NumPy for the function
import numpy as np

# Will need to import curve_fit from SciPy
from scipy.optimize import curve_fit

# This function calculates expected brain mass from the body mass of the animal
def brainmodel(body,a,k):
    return a*(body**k)
```

```python
# Put your code here
```

```python
### ANSWER
params, ignore = curve_fit(brainmodel,bb.body_g, bb.brain_g)
print(params)

# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for printing
```

&#9989;&nbsp; **Part 2.2 (20 points): Visualizing your model**

Now that you have a set of best fit parameters, **use those parameters to calculate the expected model values and plot the expected model values on top of the actual data values**. Make sure to label your axes and include a legend to communicate what parts of the plot correspond to the model versus the data.

```python
# Make sure we import matplotlib
import matplotlib.pyplot as plt

# Put your code here
```

```python
### ANSWER ###
plt.scatter(bb.body_g, bb.brain_g,label='data')
plt.plot(bb.body_g,brainmodel(bb.body_g,params[0],params[1]),label='fit')
plt.xlabel("Body mass (g)")
plt.ylabel("Brain mass (g)")
plt.legend()

# Rubric
# 5 pts for plotting data
# 5 pts for plotting the best fit
# 5 pts for axes labels
# 5 pts for legend
```
