The United States Geological Survey ([UGSC](https://www.usgs.gov/)) is a government agency that monitors and publishes data related to the natural resources of the United States. One of the data that UGSC collects is related to water quality. In this question you will be working with data collected from Lake Lansing, which has Hydrologic Unit Code HUC 12-040500040504. The measurements are pooled from approximately 3 sites as shown below and across different depth locations. In this problem, for simplicity the data does not show which location measured which data, which depth it was taken at, nor does it show the time stamp of the measurements. 

The data is structured as follows:
* Column 0: Index column
* Column 1: Dissolved Oxygen (milligrams/Liter)
    - For a healthy freshwater lake, the ideal dissolved oxygen level is generally between 6.5 and 8 milligrams per liter (mg/L), [Source](https://www.fondriest.com/environmental-measurements/parameters/water-quality/dissolved-oxygen/).
* Column 2: Specific conductance (measured in microsiemens per centimeter ($\mu$ S/cm))
    - Is typically in the range 161-419 $\mu$ S/cm for Michigan's inland lakes, [Source](https://pubs.usgs.gov/sir/2011/5233/pdf/sir2011-5233_web.pdf).
    - Conductivity of 1000+ $\mu$ S/cm indicates saline (salty) conditions, [Source](https://datastream.org/en-ca/guidebook/conductivity#:~:text=Conductivity%20in%20lakes%20and%20streams,of%20saline%20(salty)%20conditions.).
* Column 3: Water temperature (Celsius)
    - In summer, the top layer can reach 18.8-24.5 Celsius. The middle layer can drop to 7.4-18.8 Celsius, [Source](https://www.mbgnet.net/fresh/lakes/pond.htm#:~:text=Water%20temperatures%20in%20lakes%20during,7.4%E2%80%9318.8%20degrees%20C).
    - In winter, the temperatures can be cooler.
    * Swimming temperatures, [Source](https://www.redcross.org/content/dam/redcross/docs/Water%20Temperature%20for%20Aquatics%E2%80%94ADVISORY.pdf?srsltid=AfmBOopjWwRRJv8HJS_ZJuGsDN2836_IslVvJ-szIFXyhRAaeWMBcayz):
        - General swimming temperature is 25.5-29.4 C.
        - Young and the elderly temperature 28.9-34.4 C.
        - Fitness swimming temperature 25.5-28.9 C.
* Column 4: pH acidity level (On a scale from 0 to 14. See [this link for scale visualization](https://blog.orendatech.com/hs-fs/hubfs/pH%20scale%202.png?width=820&height=312&name=pH%20scale%202.png))
    - For freshwater lake, pH is between 6.5 and 8, [Source](https://www.epa.gov/caddis/ph).
