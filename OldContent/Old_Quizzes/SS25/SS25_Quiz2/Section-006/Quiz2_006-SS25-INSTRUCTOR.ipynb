# Part 1a) (20 points) Writing and Testing a Function

In statistics, the Pearson correlation coefficient (PCC) is a correlation coefficient that measures linear correlation between two sets of data. It has a value between −1 and 1 and the formula is defined as:
### $$ r = \frac{\sum\limits_{i=1}^{N} (x_{i}-\overline{x})(y_{i}-\overline{y})}{\sqrt{\sum\limits_{i=1}^{N} (x_{i}-\overline{x})^2(y_{i}-\overline{y})^2}}$$

where the symbols in this equation represent the following:

* $r$: Pearson correlation coefficient
* $\overline{x}$: Mean of the value of x-variable
* $\overline{y}$: Mean of the value of y-variable
* $x_{i}$: The value of x-variable at position $i$
* $y_{i}$: The value of y-variable at position $i$

```python
### ANSWER
### RUBRIC
import numpy as np
## 10/10 for a function that runs/will run and produce correct output
## 5/10 for a function that runs/will run but has errors (mathematical or otherwise)
## 0/10 for a function that will not run AND has error(s)

# Pearson's correlation coefficient function
#Solution #1
def Corr_coeff(x_value, y_value):
    mu = np.mean(x_value)
    nu = np.mean(y_value)
    num1 = []
    num2 = []
    for jj in range(len(x_value)):
        num1.append(x_value[jj] - mu)
        num2.append(y_value[jj] - nu)
    
    return sum(np.array(num1)*np.array(num2))/np.sqrt(sum(np.array(num1)**2)*(sum(np.array(num2)**2)))

#print(Corr_coeff(x_values,y_values))

#Solution #2
def Corr_coeff2(x_value, y_value):
    return np.corrcoef(x_value,y_value)

#print(Corr_coeff2(x_values,y_values))
```
