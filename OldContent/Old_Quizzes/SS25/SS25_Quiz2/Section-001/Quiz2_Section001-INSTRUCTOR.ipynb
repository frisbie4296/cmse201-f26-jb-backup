# Part 1a) (20 points) Writing and Testing a Function

Write a Python function that takes in an array of numerical data below and performs the following tasks:

1. Computes and returns the mean and standard deviation of the data.
2. Generates a histogram of the data. 

### Task 1a): (10 points)
Use the following data (test_data) that generates 100 random numbers from a normal distribution by 

```python
# put your code here!
import numpy as np

test_data = np.random.normal(loc=50, scale=10, size=100)
```

```python
### ANSWER
### RUBRIC

## 10/10 for a function that runs/will run and produce correct output
## 5/10 for a function that runs/will run but has errors (mathematical or otherwise)
## 0/10 for a function that will not run AND has error(s)

import matplotlib.pyplot as plt

def analyze_data(data):
    mean_val = np.mean(data)
    std_dev = np.std(data) 
    return mean_val, std_dev

# Test the function with sample data
mean, std = analyze_data(test_data)
print(f"Mean: {mean:.2f}, Standard Deviation: {std:.2f}")

# Plot histogram
plt.figure(figsize=(6,4))
plt.hist(test_data, bins=10, alpha=0.7, color='blue', edgecolor='black')
plt.axvline(mean, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean:.2f}')
plt.title("Histogram of Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
### ANSWER

# Rubric
# 5 points for reading in only the desired columns of data and defining variables
# 5 points for plotting the data
# 5 points for axis labels
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset using NumPy (excluding headers)
data = np.loadtxt("City_Climate_Data.csv", delimiter=",", skiprows=1)

# Extract columns
city_index = data[:, 0]  # City_Index
population = data[:, 1]  # Population
avg_temperature = data[:, 2]  # Avg_Temperature
annual_rainfall = data[:, 3]  # Annual_Rainfall

# Visualization 1: Histogram of Population
plt.figure(figsize=(6,4))
plt.hist(population, bins=20, color='blue', alpha=0.7, edgecolor='black')
plt.xlabel("Population")
plt.ylabel("Number of Cities")
plt.title("Distribution of City Populations")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Visualization 2: Line Plot of Average Temperature
plt.figure(figsize=(6,4))
plt.plot(city_index, avg_temperature, marker='o', linestyle='-', color='red', markersize=3, alpha=0.7)
plt.xlabel("City Index")
plt.ylabel("Average Temperature (°C)")
plt.title("Average Temperature by City")
plt.grid(True)
plt.show()

# Visualization 3: Line Plot of Annual Rainfall
plt.figure(figsize=(6,4))
plt.plot(city_index, annual_rainfall, marker='s', linestyle='-', color='green', markersize=3, alpha=0.7)
plt.xlabel("City Index")
plt.ylabel("Annual Rainfall (mm)")
plt.title("Annual Rainfall by City")
plt.grid(True)
plt.show()
```
