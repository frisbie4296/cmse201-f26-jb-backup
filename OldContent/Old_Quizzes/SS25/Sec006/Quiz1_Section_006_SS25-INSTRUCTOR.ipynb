```python
### ANSWER
grades = [89,92,55,81.5,77,95.5,65,70,82,60,85,81]

for grade in grades:
    if grade > 90:
        print("The GPA of",grade,"is","4 point!")
    elif grade > 85 and grade <=90 :
        print("The GPA of",grade,"is","3.5 point!")
    elif grade > 80 and grade <= 85:
        print("The GPA of",grade,"is","3.0 point!")
    elif grade > 75 and grade <= 80:
        print("The GPA of",grade,"is","2.5 point!")
    elif grade > 70 and grade <= 75 :
        print("The GPA of",grade,"is","2.0 point!")
    elif grade > 65 and grade <= 70:
        print("The GPA of",grade,"is","1.5 point!")
    elif grade > 60 and grade <= 65 :
        print("The GPA of",grade,"is","1.0 point!")
    else:
        print("The GPA of",grade,"is","0 point!")
# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```
