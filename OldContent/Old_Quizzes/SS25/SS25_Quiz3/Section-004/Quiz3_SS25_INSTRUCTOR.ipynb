# 2. Fitting a model to data with `curve_fit` and visualizing the results (35 points)

You have been in CMSE 201 for almost a full semester, and you have ~hated~ *thoroughly enjoyed* Dr. Termuhlen's lectures before the in-class assignments.  You want to see if you can predict how long you will ~have to endure~ *get to listen* to lectures in the remainder of the CMSE 201 class periods.

```python

# RUN THIS CELL to make sure you have the dataset loaded

import pandas as pd
import numpy as np

data = {'Day': np.arange(1,21), 
        'Minutes': [43+54/60, 33+44/60, 39+43/60, 28+29/60, 32+28/60, 29+36/60, 19+52/60, 19+2/60, 20+38/60, 27+11/60, 15+37/60, 11+57/60, 15+15/60, 29+11/60, 20+20/60, 16+8/60, 28+5/60, 29+44/60, 17+29/60, 21+16/60]}
lectures = pd.DataFrame(data)
lectures.head()
```

---

#### &#9989;&nbsp; **Question 2.1 (10 points):** 

In this question, you will fit an exponential decay model $y = A e^{-Bt}+C$ to the data using `curve_fit`.  In this case, the independent variable $t$ is the Day and the dependent variable $y$ is the Minutes.  (Don't worry, the function to define the model has already been defined for you!)

1. **Use `curve_fit` to find best fit parameters for $A$, $B$, and $C$**, using the `lectures` dataframe above and the `exp_decay` model below.
2. **Make sure you give `curve_fit` a good starting point for finding the best fit parameters** using the `p0` argument and these starting values:
    * $A = 30.0$
    * $B = 0.3$
    * $C = 20.0$
3. **Once you've found the best values, print them out**. You should find that they are not far off from, but not the same as, the initial values you provide.

```python
## RUN THIS CELL!!!
# Will need to import curve_fit from SciPy
from scipy.optimize import curve_fit

# This function calculates minutes of lecture corresponding to a given day
def exp_decay(t,A,B,C):
    return A*np.exp(-B*t) + C
```

```python
# Put your code here
```

```python
### ANSWER
params,ignore = curve_fit(exp_decay,lectures['Day'],lectures['Minutes'],p0=(40.0,0.3,20.0))
print(params)

# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for p0 correct
# 5 pts for printing
```
