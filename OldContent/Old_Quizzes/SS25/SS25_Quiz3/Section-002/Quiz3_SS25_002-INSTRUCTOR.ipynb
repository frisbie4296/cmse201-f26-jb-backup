# 2. Curve fitting underwater lake temperatures and visualizing the results (50 points)

You likely observed some periodicity of temperatures in the summary plot from Quiz $2$. Suppose you now want to investigate the relationship between temperature and time of year in your underwater lake temperatures dataset. One way to do this is by modeling the temperature observations [using a *periodic* function](https://www.examples.com/ap-precalculus/modeling-data-and-scenarios-with-sinusoidal-functions), which looks like the following:

$$ T(t) = A*\sin\left(\omega*t + C \right) + S$$


where $T$ is the temperature (degrees Celsius), $t$ is time (days), $A$ is the amplitude, $\omega$ is the period, $C$ is the phase shift, and $S$ is the vertical shift. Higher temperatures likely occur during the summer months.

#### &#9989;&nbsp; **Task 1 (10 points):** 
Pick one column of measurements in your dataset. Using the data and information provided above, write into code the sinusoid function which you will be fitting for this column of the data. Use the `curve_fit` to find best fit values for the parameters of the model. Make sure you give `curve_fit` a good starting point for finding the best fit parameters using the `p0` argument and these starting values:
* A = $20$
* $\omega$ = $2\pi$
* C = $3$
* S = $30$

Once you've found the best values, **print them out**. You should find that they are not far off from, but not the same as, the starting values.

```python
#Will need to use NumPy for the function
import numpy as np
from scipy.optimize import curve_fit

# Put your code here
```

```python
#ANSWER
#Will need to use NumPy for the function
import numpy as np
from scipy.optimize import curve_fit

#This sinusoid function models the temperature trends
def sin_func(t,A,w,C,S):
    return A*np.sin(t*w + C) + S


#initial parameter estimates
A_init=20
w_init=2*np.pi
C_init=3
S_init=30

popt,pcov = curve_fit(sin_func,temp["Day of year"],temp["-4.19"],p0=[A_init, w_init, C_init, S_init])
print(popt)

# popt_2,pcov_2 = curve_fit(sine,temp_new["Day of year"],temp_new["-4.19"],p0=[A_init])
# print(popt_2)

# rubric
# 3 pts for using model function correctly
# 3 pts for using data correctly
# 2 pts for p0 correct
# 2 pts for printing
```

```python
### ANSWER ###
import matplotlib.pyplot as plt


#fig = plt.figure(figsize=(9,8))

#model values
y_best_fit = sin_func(temp['Day of year'],*popt)

#plots
plt.scatter(temp['Day of year'],temp['-4.19'],label='Data Points')
plt.plot(temp['Day of year'],y_best_fit,c='r',label='Best Fit')
plt.xlabel("Time (Days)")
plt.ylabel("Temperature (Celsius)")
plt.legend()


# fig = plt.figure(figsize=(9,8))
# y_best_fit = sine(temp_new['Day of year'],*popt_2)
# plt.scatter(temp_new['Day of year'],temp_new['-4.19'],label='Data Points')
# plt.plot(temp_new['Day of year'],y_best_fit,c='r',label='Best Fit')
# plt.xlabel("Time (Days)")
# plt.ylabel("Temperature (Celsius)")
# plt.legend()

# Rubric
# 5 pts for data and best fit
# 5 pts for labels, legend, etc.
```
