# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)
In boxing, athletes compete in weight classes.  A (simplified) list of classifications is below:
* "Heavyweight": 175.1 pounds and above
* "Middleweight": 135.1 to 175 pounds
* "Lightweight": 118.1 to 135 pounds
* "Flyweight": 118 pounds and below

In the cell below, you will see a list of 12 weights. **Using `if`, `elif`, and `else` statements, print which weight class each athlete will compete in.** Some starter code for navigating the list (and formatting the output) is given!

```python
# Example starter code
weights = [111, 113, 182, 164, 120, 122, 143, 102, 223, 174, 158, 143] #pounds

for weight in weights:
    print(weight,"competes in","[example class]")
```

```python
### ANSWER

weights = [111, 113, 182, 164, 120, 122, 143, 102, 223, 174, 158, 143] #pounds

for weight in weights:
    if weight > 175:
        print(weight,"competes in","Heavyweight")
    elif weight <= 175 and weight > 135:
        print(weight,"competes in","Middleweight")
    elif weight <= 135 and weight > 118:
        print(weight,"competes in","Lightweight")
    else:
        print(weight,"competes in","Flyweight")
    
# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You have a collection of coins, each worth a separate amount.  The number and value of the coins correspond to the lists given below.  **Write some code to find the total value of the coins that you have.**  Your code should do the following:
* Use a loop to iterate through the lists
* Multiply the value of the coin by the quantity of that coin
* Compute the total value of all the coins you have
* Print the total value

```python
# Example starter code
value_of_coins = [0.01, 0.05, 0.10, 0.25, 0.50, 1.00] #in dollars
number_of_coins = [8, 3, 5, 14, 2, 1] #quantity of coins for each value amount

# write your code here
```

```python
### ANSWER
# Example starter code

total = 0
for i in range(len(value_of_coins)):
    total += value_of_coins[i]*number_of_coins[i]
    print(total)

# RUBRIC
# 3 pts - correct for loop
# 3 pts - multiply value * number
# 2 pts - update total
# 2 pts - print final total
```
