# 1. Problem Context: Converting between `for` loops and `while` loops 

### &#9989;&nbsp; Question (10 points)
## Part 1a)
You are a production engineer designing the code for a printer of serial number srtickers. Rewrite the provided `for` loop as a `while` loop, maintaining the same output.

````python
# 1a) put your code here

stickers = ["S/N 6138i31", "S/N 6138i32", "S/N 6138i33", "S/N 6138i34"]

for sticker in stickers:
    print("Printing", sticker)
````

````python
### ANSWER

stickers = ["S/N 6138i31", "S/N 6138i32", "S/N 6138i33", "S/N 6138i34"]

for sticker in stickers:
    print("Printing", sticker)

i = 0
while i < len(stickers):
    print("Printing", stickers[i])
    i += 1

# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
````

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You're writing code for an inventory system for a construction material warehouse. The software keeps track of the total price of the stock in the warehouse. 

The entries in the same location within the given lists refer to the same item; for example, the first entry in ```material``` corresponds to the item whose unit price is in the first entry of ```unit_price``` and whose quantity is in the first entry within ```product_quantities```, etc. 


````python
# Example starter code

material = ["steel", "lumber", "concrete", "brick"]
unit_prices = [450, 590, 165, 30]
product_quantities = [12, 200, 350, 5000]

# 3a) Write your code for computing the total price of the material here
````

````python
### ANSWER
# Example starter code
material = ["steel", "lumber", "concrete", "brick"]
unit_prices = [450, 590, 165, 30]
product_quantities = [12, 200, 350, 5000]

total = 0
for i in range(len(material)):
    total += unit_prices[i] * product_quantities[i]

print(total)

# RUBRIC
# points for code that runs and produces the appropriate output
# points for reasoning
# points for resources
````
