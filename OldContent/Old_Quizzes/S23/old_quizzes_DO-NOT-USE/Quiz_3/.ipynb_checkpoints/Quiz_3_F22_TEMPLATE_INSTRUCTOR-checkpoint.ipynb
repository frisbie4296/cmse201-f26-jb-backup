### ANSWER ###

Data file contains five columns: year, monh, date_of_month, day_of_week, and births. There are 5479 rows, and it contains information about the number of births in the US for every day of the year between January 1st, 2000 and December 31st, 2014.

#### &#9989;&nbsp; **Part 2 (20 points):** Select all the data for the first of the month (i.e., where the date_of_month=1). Print out the mean (average) number of *births* that occurred on the first of the month.


```python
# Write any code you need here
```

```python
### ANSWER ###

df_mondays = df[df["date_of_month"] == 1]
monday_mean = df_mondays["births"].mean()
print(monday_mean)
```

#### &#9989;&nbsp; **Part 3 (40 points):** Calculate the mean number of births for every day of the month. Make a plot showing the average number of births for each day of the month. 


```python
# Write any code you need here
```

```python
### ANSWER ###

import matplotlib.pyplot as plt

all_days_mean = []
days_of_week = range(1,31)

for day in days_of_week:
    df_temp = df[df["date_of_month"] == day]
    all_days_mean.append(df_temp["births"].mean())
    
plt.plot(days_of_week,all_days_mean)
```
