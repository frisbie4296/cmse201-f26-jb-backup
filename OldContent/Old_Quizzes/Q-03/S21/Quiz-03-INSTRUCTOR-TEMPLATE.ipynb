```python

# Load data
bb = pd.read_csv("US_births_2000-2014_SSA.csv")
# Add column with coded year+month information
bb['datecode'] = bb['year']+(bb['month']-1)/12

# make pandas dataframe of just births on the 12, 13 and 14-th day of month
b12  = bb[bb['date_of_month']==12]
b13 = bb[bb['date_of_month']==13]
b14  = bb[bb['date_of_month']==14]

```
