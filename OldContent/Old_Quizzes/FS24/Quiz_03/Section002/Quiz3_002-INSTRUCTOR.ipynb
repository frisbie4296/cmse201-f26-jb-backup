* A = 27.5
* B = 0.017
* C = 215
* D = 57.5

Once you've found the best values, **print them out**. You should find that they are not far off from, but not the same as, the original values that your friend gave you.

```python
# Will need to use NumPy for the function
import numpy as np

# Will need to import curve_fit from SciPy
from scipy.optimize import curve_fit

# This function calculates temperature over a range of days
def calculate_temperature(t, A, B, C, D):
     return A*np.cos(B*(t - C)) + D
```

```python
# Put your code here
```

```python
### ANSWER
best_fit_params, covariance = curve_fit(calculate_temperature, temp_data['days'], temp_data['TMAX'], p0=(27.5,0.017,215,57.5))
for p in best_fit_params:
    print('%.3f' % p)

# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for p0 correct
# 5 pts for printing
```
