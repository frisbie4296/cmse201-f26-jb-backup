# 2. Fitting a model to data with `curve_fit` and visualizing the results (45 points)

## Finding Planets

You are working for NASA looking for planets outside of our solar system. One way of finding these planets is to measure how fast stars are moving towards you and modeling the motion using a *periodic* function, which looks like the following:

$$ v(t) = A*\sin\left(\omega*t + C \right) + S$$


Where $v$ is the velocity (speed), $t$ is the time, $A$ is the called amplitude, $\omega$ is called the angular velocity, $C$ is called the phase, and $S$ is called the peculiar velocity. Stars moving fast enough (I.e., with a large enough amplitude) likely have a planet around them.

You’ve been given a dataset and you are tasked with finding the best fit amplitude.

```python
### HERE IS THE DATA
import numpy as np
import pandas as pd

data_numpy = np.loadtxt("velocity_data.csv",delimiter=',',skiprows=1)
time = data_numpy[:,0]
velocity = data_numpy[:,1]

#See decriptive statistics
df_describe = pd.DataFrame(data_numpy)
df_describe.describe()
```

#### &#9989;&nbsp; **Question 2.1 (20 points):** 

Using the data and information provided above, write into code the sinusoid function which you will be fitting. Use the `curve_fit` to find best fit values for the parameters. Make sure you give `curve_fit` a good starting point for finding the best fit parameters using the `p0` argument and these starting values:
* A = -268
* $\omega$ = 2.0
* C = 8.0
* S = 9.0

Once you've found the best values, **print them out**. You should find that they are not far off from, but not the same as, the starting values.

```python
# Make sure we import numpy:
import numpy as np

# Put your code here
```

```python
#ANSWER
#Will need to use NumPy for the function
import numpy as np
from scipy.optimize import curve_fit

#This sinusoid function models the motion of stars
def sin_func(t,A,w,C,S):
    return A*np.sin(t*w + C) + S


popt,pcov = curve_fit(sin_func,time,velocity,p0=[-268,2.0,8.0,9.0])
print(popt)

# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for p0 correct
# 5 pts for printing
```

#### &#9989;&nbsp; **Question 2.2 (20 points)** 

Now that you have a set of best fit parameters, **use those parameters to calculate the expected model values and plot the expected model values on top of the actual data values**. Make sure to label your axes and include a legend to communicate what parts of the plot correspond to the model versus the data.

```python
# Make sure we import matplotlib
import matplotlib.pyplot as plt

# Put your code here
```

```python
### ANSWER ###
fig = plt.figure(figsize=(9,4))
y_best_fit = sin_func(time,*popt)
plt.scatter(time, velocity, label='Data Points')
plt.plot(time,y_best_fit,c='r',label='Best Fit')
plt.xlabel("Time (Days)")
plt.ylabel("Velocity (m/s)")
plt.legend()

# Rubric
# 10 pts for data and best fit
# 10 pts for labels, legend, etc.
```
