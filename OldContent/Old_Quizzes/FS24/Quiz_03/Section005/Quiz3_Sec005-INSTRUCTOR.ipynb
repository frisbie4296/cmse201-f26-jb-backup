#### &#9989;&nbsp; **Question 2.1 (20 points):** 

Using the data and provided model below, use `curve_fit` to find best fit parameters for `v0` and `g`. Make sure you give `curve_fit` a good starting point for finding the best fit parameters using the `p0` argument and these starting values:
* `v0 = 12`
* `g = 10`

Once you've found the best values, **print them out**. You should give your result in a human-readable way: "The best fit value for v0 is _____ and the best fit value for g is _____". 

You should find that they are not far off from, but not the same as, the original values that your friend gave you.

```python
# Will need to use NumPy for the function
import numpy as np

# Will need to import curve_fit from SciPy
from scipy.optimize import curve_fit

# This function calculates height corresponding to time for projectile motion
def projectile(t,v0,g):
    return v0*t - 0.5*g*t**2
```

```python
# Put your code here
```

```python
### ANSWER
params,ignore = curve_fit(projectile,baseball_data['time'],baseball_data['height'],p0=[12,10])
print(params)
print("The best value for v0 is", params[0], "and the best value for g is", params[1])

# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for p0 correct
# 3 pts for printing
# 2 pts for human-readable
```

```python
### ANSWER ###
plt.scatter(baseball_data['time'],baseball_data['height'],label='data')
plt.plot(baseball_data['time'],projectile(baseball_data['time'],params[0],params[1]),label='fit')
plt.legend()
plt.xlabel('time')
plt.ylabel('height')

# Rubric
# 5 pts for data and best fit
# 5 pts for title, labels, legend, etc.
```
