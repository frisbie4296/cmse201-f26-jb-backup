```python
### ANSWER ###
import numpy as np
x_values = np.random.normal(2, 2, 200)# 200 sample poitns from normal distribution with mean 2 and standard deviation 2.
y_values = np.linspace(1,4,200)# 200 evenly distributed points between 1 and 4.

def do_stats(my_yvals):
    list_mean = np.mean(my_yvals)
    list_std = np.std(my_yvals)
    list_median = np.median(my_yvals)
    return list_mean,list_std,list_median

print(do_stats(y_values))

#Rubric:
# 10 points for defining the function and having it return values
# 5 points for calculating the statistics
# 5 points for testing the function
```
