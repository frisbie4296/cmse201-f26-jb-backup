```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('openpowerlifting-2024-10-12-b58b8e08.csv')
```

```python
equip_mask = data['Equipment'] == 'Raw'
data = data[equip_mask]
country_mask = data['Country'] == 'USA'
data = data[country_mask]
#state_mask = data['State'] =='MI'

data = data[data['State']=='MI']
```

```python
data = data.drop(columns=['Name','Sex', 'Event', 'Equipment', 'AgeClass',
       'BirthYearClass', 'Division', 'WeightClassKg',
       'Squat1Kg', 'Squat2Kg', 'Squat3Kg', 'Squat4Kg', 
       'Bench1Kg', 'Bench2Kg', 'Bench3Kg', 'Bench4Kg', 
       'Deadlift1Kg', 'Deadlift2Kg', 'Deadlift3Kg', 'Deadlift4Kg', 'TotalKg', 'Place', 'Dots', 'Wilks', 'Glossbrenner',
       'Goodlift', 'Tested', 'Country', 'State', 'Federation',
       'ParentFederation', 'Date', 'MeetCountry', 'MeetState', 'MeetTown',
       'MeetName', 'Sanctioned'])
```

```python
all_clean = data.dropna()
```
