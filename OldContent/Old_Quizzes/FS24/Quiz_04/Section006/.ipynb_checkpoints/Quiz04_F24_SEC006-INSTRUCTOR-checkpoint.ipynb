## Part 1: Modeling a system using `solve_ivp` (40 points)

#### Objective:

We will model and visualize the [Rössler attractor](https://en.wikipedia.org/wiki/R%C3%B6ssler_attractor), which is a set of three coupled nonlinear ordinary differential equations (ODEs). The Rössler system is a prime example of a chaotic dynamical system, where small variations in initial conditions can lead to vastly different outcomes—a phenomenon known as the [Butterfly Effect](https://en.wikipedia.org/wiki/Butterfly_effect). This makes it an invaluable tool for understanding the fundamental principles underlying complex systems and [chaos theory](https://en.wikipedia.org/wiki/Chaos_theory).

The insights gained from studying the Rössler attractor can contribute to broader scientific understanding. Chaos theory has applications in various fields, including physics, engineering, biology, and economics. Understanding chaotic systems like the Rössler attractor also enhances our ability to analyze real-world phenomena that exhibit similar characteristics, from weather patterns and financial markets to biological processes. By exploring these systems, researchers can better predict and control complex behaviors, which is crucial for developing more accurate models and making informed decisions.

Knowledge about chaotic systems like the Rössler attractor can foster a greater appreciation of the beauty and complexity inherent in nature. It demonstrates how even simple mathematical equations can give rise to intricate patterns and behaviors that are both mesmerizing and profound. 

#### Mathematical Model:
The Rössler equations are given by:

$$
\begin{align*}
\frac{dx}{dt} &= -y - z \\
\frac{dy}{dt} &= x + ay \\
\frac{dz}{dt} &= b + xz - cz
\end{align*}
$$

where \(a = 0.2\), \(b = 0.2\), and \(c = 5.7\) are parameters that can be adjusted to explore different behaviors of the system. Another common set of parameters is \(a = 0.1\), \(b = 0.1\), and \(c = 14\).

#### Initial Conditions:

When we start the modeling process, we will use the following initial conditions for each variable:
- $x(0) = 0.1$
- $y(0) = 0.1$
- $z(0) = 0.1$

#### Time Range and Timestep:
- Evolve the model for 100 time units.
- Use a timestep of $\Delta t = 0.01$ time units.

#### Tasks:
1. **Write a Python function that calculates the derivatives for the Rössler equations** using these parameters.
2. **Set up the initial conditions and time range**.
3. **Use `solve_ivp` from SciPy to solve the system of ODEs**.
4. **Unpack the solution and plot the attractor in 2D** to visualize its behavior.
5. (BONUS) **Plot the attractor in 3D** to explore its structure further.

### Code you need to write to solve this problem:
* A function that takes in a set of time values as well as the current values for $x$, $y$, and $z$ and **returns** the derivatives for each variable, $\frac{dx}{dt}$, $\frac{dy}{dt}$, and $\frac{dz}{dt}$. This would be your "derivatives" function that `solve_ivp` needs.
* A set of variables for storing your initial conditions, $x_0$, $y_0$, and $z_0$, set to the correct initial values.
* A variable that stores all the times for which you want to solve the system of ODEs, you can generate these times using `np.arange()`. The times should match the time range and timestep size specified above.
* Code that calls `solve_ivp` using all of the appropriate arguments and stores the result. 
* Code that unpacks the solution so that you can plot it.

**Note**: you will not need the "`args`" argument for `solve_ivp` for this case because any model parameters that might be needed for this system are provided directly as numbers in the set of  ODEs include above.

&#9989;&nbsp; **Question 1 (30 points)**:

For 50 time units, solve the Rössler equations using the parameters \(a = 0.2\), \(b = 0.2\), and \(c = 5.7\). Your code should use a function for the derivatives, set up the initial conditions and time range, use `solve_ivp` to solve the system of ODEs, and unpack the solution.

**Show that you can print the first few values of the solution to demonstrate that you have successfully solved the system of ODEs.**

```python
# The import commands you need are provided here
from scipy.integrate import solve_ivp
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D ## Only for the BONUS question

# Put your code here
```

```python
### ANSWER ###

# Define the derivatives function
def rossler_derivatives(t, state, a=0.2, b=0.2, c=5.7):
    x, y, z = state
    dx_dt = -y - z
    dy_dt = x + a * y
    dz_dt = b + x * z - c * z
    return [dx_dt, dy_dt, dz_dt]

# Initial conditions
x0, y0, z0 = 0.1, 0.1, 0.1

# Time range and timestep
t_span = (0, 100)
t_eval = np.arange(t_span[0], t_span[1], 0.01)

# Solve the system of ODEs
sol = solve_ivp(rossler_derivatives, t_span, [x0, y0, z0], t_eval=t_eval)

# Extract the solution
x = sol.y[0,:]
y = sol.y[1,:]
z = sol.y[2,:]

# Print the solution
print(x[0:10])
```
