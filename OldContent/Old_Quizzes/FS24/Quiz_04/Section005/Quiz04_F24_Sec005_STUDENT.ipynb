This system of equations was originally developed by Edward Lorenz, Ellen Fetter, and Margaret Hamilton as a simplified model to use in weather modeling.  It was later observed that the system is very sensitive to initial conditions, and it has been used to demonstrate [turbulence](https://en.wikipedia.org/wiki/Turbulence#) and the so-called "[butterfly effect](https://en.wikipedia.org/wiki/Butterfly_effect)".

$$\frac{dx}{dt} = 10 (y-x)$$
$$\frac{dy}{dt} = 28 x - y - xz$$
$$\frac{dz}{dt} = -2.67 z + xy$$
