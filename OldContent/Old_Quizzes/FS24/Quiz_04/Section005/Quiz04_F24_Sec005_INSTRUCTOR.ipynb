```python
# The import commands you need are provided here
from scipy.integrate import solve_ivp
import numpy as np
import matplotlib.pyplot as plt

# Put your code here

# PSEUDOCODE TO GET YOU STARTED!
# Define the derivatives function

# Initial conditions

# Define a time array

# Call solve_ivp

# Unpack the solution from solve_ivp

# Plot x vs. y
```

```python
### ANSWER ###

# Import needed modules
from scipy.integrate import solve_ivp
import numpy as np

# Define the derivatives function
def derivs(t, vals):
    x, y, z = vals
    dxdt = 10*(y-x)
    dydt = 28*x - y - x*z
    dzdt = -2.67*z + x*y
    return dxdt, dydt, dzdt

# Initial conditions
x0 = 0.1
y0 = 0.1
z0 = 0.1
ICs = [x0, y0, z0]

# Define a time array
t = np.arange(0, 100, 0.01)

# Call solve_ivp
sol = solve_ivp(derivs, (t[0], t[-1]),ICs, t_eval=t)

# Unpack the solution from solve_ivp
x = sol.y[0,:]
y = sol.y[1,:]
z = sol.y[2,:]

# Plot x vs. y
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.grid('on')
# plt.savefig('Lorenz_X_Y')
```
