# 2. Problem Context: Converting between `for` loops and `while` loops (10 points)

```python
# put your code here

name_list = ["Vivian", "Ryan", "Steven"]
for name in name_list:
    print("Making attendance for", name)
```

```python
### ANSWER

name_list = ["Vivian", "Ryan", "Steven"]
for name in name_list:
    print("Making attendance for", name)

i=0
while i<len(name_list):
    print("Making attendance for",name_list[i])
    i+=1

# RUBRIC
# 5 pts for correctly converting the for loop to a while loop with the same output
# 5 pts for providing resources and describing approach in detail
```
