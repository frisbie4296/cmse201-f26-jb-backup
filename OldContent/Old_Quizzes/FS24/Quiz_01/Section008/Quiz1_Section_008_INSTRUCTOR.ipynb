```python
### ANSWER
# Example starter code
course_assignment_weights = [0.15, 0.15, 0.2, 0.25, 0.25]
assignment_score_total = [80/100, 100/100, 67/100, 71/100, 79.5/100]

total = 0
for i in range(len(course_assignment_weights)):
    total += course_assignment_weights[i]*assignment_score_total[i]
    print(total)

# RUBRIC
# points for code that runs and produces the appropriate output
# points for reasoning
# points for resources
```
