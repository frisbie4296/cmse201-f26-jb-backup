```python
# put your code here

stitches=["satin","back","running","split","feather"]

i=0
while i<(len(stitches)):
    print(stitches[i], "stitch")
    i+=1
```

```python
### ANSWER

stitches=["satin","back","running","split","feather"]

i=0
while i<(len(stitches)-1):
    print(stitches[i], "stitch")
    i+=1

for i in stitches:
    print(i,"stitch")

# RUBRIC
# 5 pts for correctly converting the for loop to a while loop with the same output
# 5 pts for providing resources and describing approach in detail
```
