### &#9989;&nbsp; Pirate Party Invitations (10 points)
You're organizing a pirate-themed party, and you have a list of pirate names that you want to print on invitations. Rewrite the following `for` loop code as a `while` loop to achieve this while maintaining the original printed output.

```python
pirate_names = ["Captain Blackbeard", "Anne Bonny", "Long John Silver", "Mary Read"]
invitation_message = "Ahoy! You are invited,"

for pirate in pirate_names:
    print(invitation_message, pirate)
```

```python
# put your code here
```

```python
### ANSWER ###
x=0
while x < len(pirate_names):
    print(invitation_message, pirate_names[x])
    x+=1

### RUBRIC
# 10 pts for code that runs and give correct output and uses a while loop
# 5 pts for code that is missing any one of the above items
# 2 pts for an attempt that didn't work
# 0 pts for no attempt
```

```python
# put your code here
```
