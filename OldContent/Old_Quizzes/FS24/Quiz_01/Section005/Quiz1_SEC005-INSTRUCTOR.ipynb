# 2. Converting between `for` loops and `while` loops (10 points)

#### Problem context: A rather famous thought experiment poses the question: would you rather have \\$1,000,000 all at once or 1 penny (\\$0.01) that doubles each day for 30 days?  It turns out that the penny will turn into more than \\$1,000,000 at the end of 30 days.  Below is a `for` loop that determines how many days it takes for the penny to become \$1,000,000.

```python
# for loop
total = 0.01
number_of_days = 1
for i in range(30):
    if total < 1000000:
        total *= 2
        number_of_days += 1
print("At the end of ", number_of_days, "days, you will have ", total, "dollars" )
```

### &#9989;&nbsp; Rewrite the provided `for` loop as a `while` loop.  The output from your `while` loop should be the same as the above `for` loop.  (Hint: The loop should end once the total reaches \\$1,000,000!) 

```python
# put your while loop here!

```

```python
### ANSWER

total = 0.01
number_of_days = 1
while total < 1000000:
    total *= 2
    number_of_days += 1
print("At the end of ", number_of_days, "days, you will have ", total, "dollars" )

# RUBRIC
# 5 pts for correctly converting the for loop to a while loop with the same output
```

#### Problem context: You are working in the sales department for a fencing company.  To simplify the problem, we can assume that the fences will always be built around a rectangular area of land. ### &#9989;&nbsp; Given the base and width of multiple rectangles, write a `for` loop to determine the total perimeter of each the rectangles (so you know how much fencing to order!)

```python
base = [15.5, 20.6, 17.8, 25.2]
width = [18.5, 17.3, 25.4, 30.2]

# Write your code here!
```

```python
### ANSWER

# one solution
total = 0
for i in range(len(base)):
    total += 2*(base[i]+width[i])
print(total)

# I would also accept this, depending on how they interpret the question
perim = []
for i in range(len(base)):
    perim.append(2*(base[i]+width[i]))
print(perim)

# RUBRIC
# 5 points for code that runs and produces the appropriate output
# 5 points for reasoning and resources
```
