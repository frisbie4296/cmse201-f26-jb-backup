### &#9989;&nbsp; You're working in a scientific laboratory that needs to purchase some chemical reagents. You need to calculate the total cost for different samples of chemicals. Write a program to compute the total cost based on the unit prices and weights (in tonnes) of items bought.
You can ssume that the $i^{th}$ entry in the both lists correspond.
### &#9989;&nbsp; Metacognition
How do you know that your code is working?

What resources did you consult, and how did they inform your approach? Include any past assignments, urls, generative AI prompts/outputs, as well as how you used them.

```python
# Example starter code
chemical_prices = [47650, 694, 691770, 41650, 65.86, 45545]
chemical_quantities = [0.25, 3, 0.12, 1.6, 2.13, 0.75]
```

```python
### ANSWER
# Example starter code
chemical_prices = [47650, 694, 691770, 41650, 65.86, 45545]
chemical_quantities = [0.25, 3, 0.12, 1.6, 2.13, 0.75]

total = 0
for i in range(len(chemical_prices)):
    total += chemical_prices[i]*chemical_quantities[i]
print(total)

# RUBRIC
# 7 points for code that runs and produces the appropriate output
# 2 points for reasoning
# 1 points for resources
```
