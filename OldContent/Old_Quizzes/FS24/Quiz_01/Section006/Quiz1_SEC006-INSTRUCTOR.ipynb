# 1. Organizing a Punk Legends Concert Lineup (10 points)
## Manipulating Lists with Loops

*Problem Context*: You are managing a huge punk concert lineup featuring bands from **1980 to 2020**. You need to split the bands into two categories: **old bands** (formed before 2000) and **newer bands** (formed in or after 2000). However, you’re not given the founding years directly, but you know which bands fit into each category.

Your task is to filter out the **old bands** and add them to a new list called `old_bands`, which will represent the headliners at the concert.

### &#9989;&nbsp; Write a `for` loop that uses `if` conditions with `or` statements to identify punk bands from the `bands` list that were formed **before the year 2000**, based on their **names**. Add the old bands to the `old_bands` list.

### &#9989;&nbsp; Metacognition  
Describe your approach and how using `if` conditions with `or` statements handles the identification of old bands effectively. Be sure to note any resources you consulted, such as past assignments, web searches, generative AI outputs, etc.

### Old Bands
The following bands are old (formed **before** 2000) and should be added to the `old_bands` list:
- NOFX
- Bad Religion
- Lagwagon
- Rancid
- Strung Out

### Newer Bands
These newer bands (formed **in or after** 2000) **do not** belong in the `old_bands` list:
- The Menzingers
- Rise Against
- The Flatliners
- Against Me!
- PUP

```python
# put your code here!

bands = ["NOFX", "Lagwagon", "Rise Against", "The Menzingers", "Against Me!", 
         "Bad Religion", "Strung Out", "Rancid", "The Flatliners", "PUP"]

old_bands = []   # New list for old bands formed before 2000
```

```python
### ANSWER

bands = ["NOFX", "Lagwagon", "Rise Against", "The Menzingers", "Against Me!", 
         "Bad Religion", "Strung Out", "Rancid", "The Flatliners", "PUP"]

old_bands = []   # New list for old bands formed before 2000

# Using if-else with or statements to identify old bands
for band in bands:
    if band == "NOFX" or band == "Bad Religion" or band == "Lagwagon" or \
       band == "Rancid" or band == "Strung Out":
        old_bands.append(band)

print(old_bands)

# Expected output:
# ['NOFX', 'Lagwagon', 'Bad Religion', 'Strung Out', 'Rancid']

# RUBRIC
# 5 pts for correctly creating a creating a new list with herbivores using a loop
# 5 pts for providing resources and describing approach in detail
```

# 2. **Coaching Soccer Practice Attendance** (10 points) 
## Converting between `for` loops and `while` loops

*Problem Context:* As a coach, you're tracking attendance for your soccer team's practice session. You have the names of the players, and you need to call out each name to confirm they are present for the session. You want to change the attendance tracking system from using a `for` loop to a `while` loop.

### &#9989;&nbsp; Rewrite the provided `for` loop as a `while` loop, maintaining the same output.

### &#9989;&nbsp; Metacognition  
Explain the differences between using a `for` loop and a `while` loop in this context. What made your `while` loop achieve the same result?  

What resources did you consult, and how did they inform your approach? Include any past assignments, URLs, generative AI prompts/outputs, as well as how you used them.

```python
# put your code here

players = ["Sam", "Chris", "Taylor"]
for player in players:
    print("Marking attendance for", player)
```

```python
### ANSWER

players = ["Sam", "Chris", "Taylor"]
for player in players:
    print("Marking attendance for", player)

# Converting for loop to while loop
i = 0
while i < len(players):
    print("Marking attendance for", players[i])
    i += 1

# RUBRIC
# 5 pts for correctly converting the for loop to a while loop with the same output
# 5 pts for providing resources and describing approach in detail
```

# 3. Problem Context: Categorizing Dinosaur Fossils by Geological Era (10 points)
## Using if/else Statements to Control Flow

*Problem Context:* You are helping a paleontologist sort dinosaur fossils into their respective geological periods for a museum exhibit. The fossils need to be categorized into the **Triassic**, **Jurassic**, or **Cretaceous** periods based on their age.  The **Triassic** period ranged from **250 to 201 million years ago**, the **Jurassic** period spanned **201 to 145 million years ago**, and the **Cretaceous** period covered from **145 to 66 million years ago**. 

### &#9989;&nbsp; Use this information to classify fossils based on their estimated ages. Rewrite the following series of nested `if` statements as `if-elif-else` statements to classify fossils.

### &#9989;&nbsp; Metacognition  
How do you know that your code is working?

What resources did you consult, and how did they inform your approach? Include any past assignments, URLs, generative AI prompts/outputs, as well as how you used them.

```python
# Example starter code

fossil_age_in_mya = 160  # Fossil age in million years ago (mya)

if fossil_age_in_mya < 250:
    if fossil_age_in_mya > 201:
        print("Triassic")
    else:
        if fossil_age_in_mya > 145:
            print("Jurassic")
        else:
            if fossil_age_in_mya > 66:
                print("Cretaceous")
            else:
                print("Post-dinosaur era")
else:
    print("Invalid age")  # The fossil age can’t be greater than 250 (upper range of Mesozoic)
```

```python
### ANSWER

fossil_age_in_mya = 160  # Fossil age in million years ago (mya)

if fossil_age_in_mya > 250:
    print("Invalid age")
elif fossil_age_in_mya > 201:
    print("Triassic")
elif fossil_age_in_mya > 145:
    print("Jurassic")
elif fossil_age_in_mya > 66:
    print("Cretaceous")
else:
    print("Post-dinosaur era")

# RUBRIC
# Correctly transforms the  nested if into if-elif-else structure statements (5 pts)
# Provides a well-reasoned explanation on how the solution works (3 pts)
# Lists relevant resources used to inform the answer (2 pts)
```

# 4. **Punk Legends Concert Ticket Sales Calculation** (10 points)
## Using Loops for Calculations

*Problem Context:* You're managing the ticket sales for the **Punk Legends Concert**, featuring some of the greatest punk bands of all time! The concert offers five different types of tickets ranging from economy to VIP. You're given the prices for each ticket type and the quantities sold for each corresponding type.

### &#9989;&nbsp;  Write a program to compute the total ticket revenue for the concert.

### &#9989;&nbsp; Metacognition  
Explain how you know your code is computing the total revenue correctly.

What resources did you consult, and how did they inform your approach? Include any past assignments, URLs, generative AI prompts/outputs, as well as how you used them.

### **Ticket Prices** and **Quantities Sold**:
- **VIP** tickets cost **$150** and **85** were sold.
- **Premium** tickets cost **$100** and **160** were sold.
- **General Admission** tickets cost **$60** and **220** were sold.
- **Balcony** tickets cost **$40** and **320** were sold.
- **Economy** tickets cost **$25** and **500** were sold.

```python
# Example starter code
ticket_prices = [150, 100, 60, 40, 25]  # Prices for VIP, Premium, GA, Balcony, Economy
ticket_quantities = [85, 160, 220, 320, 500]  # Quantities sold for each type
```

```python
### ANSWER
ticket_prices = [150, 100, 60, 40, 25]  # Prices for VIP, Premium, GA, Balcony, Economy
ticket_quantities = [85, 160, 220, 320, 500]  # Quantities sold for each type

# Initial total
total_revenue = 0

# Loop through ticket prices and quantities to compute the total revenue
for i in range(len(ticket_prices)):
    total_revenue += ticket_prices[i] * ticket_quantities[i]  # Add the revenue for each ticket type

# Final total revenue
print(f"\nTotal revenue from Punk Legends Concert ticket sales: ${total_revenue:.2f}")

# RUBRIC
# Correctly implements a loop to calculate total revenue for the concert (5 pts)
# Provides valid reasoning on confirming the correctness of the solution and lists resources consulted (3 pts)
# Includes any relevant resources used to inform the approach (2 pts)
```
