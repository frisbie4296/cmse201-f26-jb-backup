# 3. Problem Context: Categorizing Dinosaur Fossils by Geological Era (10 points)
## Using if/else Statements to Control Flow

*Problem Context:* You are helping a paleontologist sort dinosaur fossils into their respective geological periods for a museum exhibit. The fossils need to be categorized into the **Triassic**, **Jurassic**, or **Cretaceous** periods based on their age.  The **Triassic** period ranged from **250 to 201 million years ago**, the **Jurassic** period spanned **201 to 145 million years ago**, and the **Cretaceous** period covered from **145 to 66 million years ago**. 

### &#9989;&nbsp; Use this information to classify fossils based on their estimated ages. Rewrite the following series of nested `if` statements as `if-elif-else` statements to classify fossils.

### &#9989;&nbsp; Metacognition  
How do you know that your code is working?

What resources did you consult, and how did they inform your approach? Include any past assignments, URLs, generative AI prompts/outputs, as well as how you used them.

```python
# Example starter code

fossil_age_in_mya = 160  # Fossil age in million years ago (mya)

if fossil_age_in_mya < 250:
    if fossil_age_in_mya > 201:
        print("Triassic")
    else:
        if fossil_age_in_mya > 145:
            print("Jurassic")
        else:
            if fossil_age_in_mya > 66:
                print("Cretaceous")
            else:
                print("Post-dinosaur era")
else:
    print("Invalid age")  # The fossil age can’t be greater than 250 (upper range of Mesozoic)
```
