Now you are helping the instructor to send messages to the students with their grades and the feedback messages. The rules will be the following:

grade >= 90: 
- message = 'Great!'.

80 <= grade < 90: 
- message = 'Good job~'

60 <= grade < 80: 
- message = 'Not bad, keep going...'

40 <= grade < 60:

&#9989;&nbsp; Please write a function, names `feedback`, and the input will be one `int`, which is the quiz grade, and the output will be one `str`, which is: 

    'Your quiz grade is (grade). (message)'. 

Example: 

When you run `feedback(45)`, it returns you the string:

    Your quiz grade is 45. Please attend more office hours.

In this `feedback` function, you are allowed to use only **one** `if` branch and **no** `elif`'s. (That means, the keyword `if` can only appear once in your function definition). If you are not able to do so, you may use multiple `if`'s and `elif`'s in the `feedback` function, but you will lose some points.

```python
# ANSWER
def feedback(grade):
    points_seg = [101, 90, 80, 60, 40, 20, 0]
    message \
        = ['Great!',
           'Good job~',
           'Not bad, keep going...',
           'Please attend more office hours.',
           'You are not doing well, do you need help?',
           'Please contact the instructor ASAP for your quiz!']
    for point_seg_idx in range(len(points_seg)):
        if grade < points_seg[point_seg_idx] and grade >= points_seg[point_seg_idx + 1]:
            return 'Your quiz grade is ' + str(grade) + '. ' + message[point_seg_idx]

# 30 pts
# 20 pts if with multiple ifs.
```

&#9989;&nbsp; Please test your function in the cases `grade = 25, 40, 100`.

```python
# Put your code here
```

```python
### ANSWER
print(feedback(25))
print(feedback(40))
print(feedback(100))
# 5 pts
```
