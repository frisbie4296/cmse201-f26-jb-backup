# Drawing Star Shapes  (15 points)

One of your friend wants to have a Python function that allows him to draw various star shapes. To compute the points $(x,y)$ that determine a star shape, he wants to use this set of equations 
\begin{equation*}
\begin{aligned}
x =&\,\, t\,\cos(0.25\,\pi)-r\,\sin(0.25\,\pi) + 0.5, \\
y =&\,\, t\,\sin(0.25\,\pi)+r\,\cos(0.25\,\pi) + 0.5,
\end{aligned}
\end{equation*}
where 
\begin{equation*}
\begin{aligned}
t =&\,\, (0.5+\epsilon\,\sin(\alpha\,s))\,\cos(s), \\
r =&\,\, (0.5+\epsilon\,\sin(\alpha\,s))\,\sin(s),\\
&\,\,
\end{aligned}
\end{equation*}
$s$ goes from $0$ to $2\,\pi$ radians, and $\epsilon$ and $\alpha$ are parameters. 
You propose to him to write down a Python code that will allow him to visualize these shapes.

&#9989;&nbsp; **Part 1 (1 point):** Import all the modules that you use in the cell below.

```python
# Put your code here! 
import math 
import matplotlib.pyplot as plt 
%matplotlib inline
import numpy as np
```

&#9989;&nbsp; **Part 2 (8 points):** 
You want to create a Python function, called `compute_star_shapes`, that computes the points $(x,y)$ of a given star shape. Your Python function takes the following inputs: the increment between two values of $s$, the value of $\epsilon$ and the value of $\alpha$. The default value of $\epsilon$ and $\alpha$ are respectively $0.15$ and $3$, meaning that your Python function should consider $\epsilon = 0.15$ and $\alpha = 3$ if these parameters are not specified.
Your Python function returns the number of points, a list (or array) containing the $x$ values and a list (or array) containing the $y$ values. 

```python
# Put your function here!
```

```python
# Put your function here! a possible solution with lists
def compute_star_shapes(ds, epsilon = 0.15, alpha = 3):
    s_start = 0 
    s_end = 2*math.pi
    x_vector = list()
    y_vector = list()
    s = s_start
    while s <= s_end:
        t = (0.5+epsilon*math.sin(alpha*s))*math.cos(s)
        r = (0.5+epsilon*math.sin(alpha*s))*math.sin(s) 
        x = t*math.cos(0.25*math.pi)-r*math.sin(0.25*math.pi) + 0.5 
        x_vector.append(x)
        y = t*math.sin(0.25*math.pi)+r*math.cos(0.25*math.pi) + 0.5
        y_vector.append(y)
        s += ds 
    return len(x_vector), x_vector, y_vector
```

```python
# a possible solution with numpy arrays
def compute_star_shapes(ds, epsilon = 0.15, alpha = 3):
    s_start = 0 
    s_end = 2*np.pi
    s = np.arange(s_start,s_end+ds,ds)
    t = (0.5+epsilon*np.sin(alpha*s))*np.cos(s)
    r = (0.5+epsilon*np.sin(alpha*s))*np.sin(s) 
    x = t*np.cos(0.25*np.pi)-r*np.sin(0.25*np.pi) + 0.5 
    y = t*np.sin(0.25*np.pi)+r*np.cos(0.25*np.pi) + 0.5
    return len(x), x, y
```

So, you should have four different plots. Each plot should contain a $x$-axis label, a $y$-axis label, and a title that specifies the values of $\epsilon$ and $\alpha$. Each star shape should be of different color. In the **figure**'s title, you will specify the number of points that are used to generate the curves. Complete the code in the cell below. 

```python
ds = 0.01 
# Put your code here!
plt.figure(figsize=(12,12))
nb_points, x, y = compute_star_shapes(ds, 0.15, 3)
plt.suptitle('Star shapes with '+str(nb_points)+' points')  
plt.subplot(2,2,1) 
plt.plot(x, y, color='k') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('epsilon=0.15, alpha=3')
plt.subplot(2,2,2) 
nb_points, x, y = compute_star_shapes(ds, 0.15, 5)
plt.plot(x, y, color='Purple') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('epsilon=0.15, alpha=5')
plt.subplot(2,2,3) 
nb_points, x, y = compute_star_shapes(ds, 0.3, 3)
plt.plot(x, y, color='g') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('epsilon=0.1, alpha=3')
plt.subplot(2,2,4) 
nb_points, x, y = compute_star_shapes(ds, 0.3, 5)
plt.plot(x, y, color='red') 
plt.xlabel('x')
plt.ylabel('y')
plt.title('epsilon=0.1, alpha=5')
```
