```python
### ANSWER ###
#Rubric:
#Definition of the function: 5 points
#Having *some* function arguments (ideally filename, at least): 5 points
#Incorporating code from previous question into the function: 15 points
#Runs correctly: 5 points

def read_and_plot_data(filename,y_label,plot_title):
    time, brightness= np.loadtxt(filename, usecols = (0,1), unpack=True, skiprows = 1, delimiter=',')
    transit_count = 0
    for kk in range(len(time)):
        if brightness[kk] < 0.99:
            transit_count+=1
    print('There are',transit_count,'points below 0.99')
    plt.plot(time, brightness,linewidth=3.0)
    plt.xlabel("Time")
    plt.ylabel(y_label)
    plt.title(plot_title)
    plt.show()

read_and_plot_data("hatp32.csv","Brightness","Exoplanet transit observation")
```
