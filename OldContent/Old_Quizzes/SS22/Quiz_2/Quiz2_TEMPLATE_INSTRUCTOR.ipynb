Thinking about this a little more, you realize you’re probably going to make many of these kinds of plots, so you decide to turn this code into a function (that way, you can easily reuse it!).

**Turn your code from the previous question--which reads in data, removes negative values, and plots it--into a function that you can easily reuse.**

```python
#Write your function for reading in data, selecting positive y values, and plotting data here
```

```python
### ANSWER ###
#Rubric:
#Definition of the function: 5 points
#Having *some* function arguments (ideally filename, at least): 5 points
#Incorporating code from previous question into the function: 15 points
#Runs correctly: 5 points

def read_and_plot_data(filename,y_label,plot_title):
    time, cell_count= np.loadtxt(filename, usecols = (0,1), unpack=True, skiprows = 1, delimiter=',')
    new_time = []
    new_cell_count = []
    for jj in range(len(time)):
        if cell_count[jj] > 0.0:
            new_time.append(time[jj])
            new_cell_count.append(cell_count[jj])
    plt.plot(new_time, new_cell_count,linewidth=3.0)
    plt.xlabel("Time")
    plt.ylabel(y_label)
    plt.title(plot_title)
    plt.show()

read_and_plot_data("bacterial_dataset.csv","Bacterial Cell Count","Growth of Bacterial Culture in Chlorine")
```
