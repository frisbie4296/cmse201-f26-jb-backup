### &#9989;&nbsp; Question 5': (10 bonus points)

Following Question 5, please **calculate** how many outliers (circles) are there in the box-plot.

```python
# Put your code here
```

```python
### Answer ###

Q1 = polymers_pd.describe()['VY'].loc['25%']
Q3 = polymers_pd.describe()['VY'].loc['75%']
lower = polymers_pd['VY'].loc[polymers_pd['VY'] < Q1 - (Q3 - Q1) * 1.5]
upper = polymers_pd['VY'].loc[polymers_pd['VY'] > Q3 + (Q3 - Q1) * 1.5]

print('There are ' + str(lower.size + upper.size) + ' outliers.')
```
