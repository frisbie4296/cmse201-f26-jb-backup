```python
# Copying and pasting the above data and handling it in the correct way is part of the evaluation.
countries = ["Brazil", "Canada", "France", "Spain", "Italy", "Mexico", "Netherlands", "Sweden", "United Kingdom", "United States"]
cases = [ 4544629, 145750, 467614, 640040, 298156, 697663, 98217, 88237, 396744, 6804814]
deaths = [ 136895, 9267, 31257, 30495, 35707, 73493, 6324, 5865, 41866, 199509 ]

for i in range( len(countries)):
    
    ratio = deaths[i] / cases[i] * 100
    # Format as what done in the example!
    print("{} has a case fatality ratio of {:.2f}".format(countries[i], ratio) )
```
