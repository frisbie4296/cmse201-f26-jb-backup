#### &#9989;&nbsp; Question 1.1 (4 points): Writing a function to sum elements in a list

Write a function to return the sum of a list of integers.

Examples:
- The sum of `[1, 2, 3]` is `1 + 2 + 3 = 6`.
- The sum of `[-1, 0, 1]` is `-1 + 0 + 1 = 0`.

Note: Write the function "from scratch" - don't use any built-in Python functions.

````python
# Your code here.
````

````python
### ANSWER
def compute_sum(values):
    total = 0
    for value in values:
        total += value
    return total
````

#### &#9989;&nbsp; Question 1.4 (4 points): Counting certain elements in a list

Write a function to return the number of elements in a list that are greater than or equal some `threshold` value. The function should take the list of numbers and the `threshold` value as input parameters, and set the default value of `threshold` to `0.0`. Include at least one check to show your function is correct.

Examples: 
- The number of elements in `[-1.0, 2.1, 5.0]` greater than or equal to `threshold=0.0` is `2`.
- The number of elements in `[-1.0, 2.1, 5.0]` greater than or equal to `threshold=3.2` is `1`.

````python
# Your code here.
````

````python
### ANSWER
def count_elements_above_threshold(values, threshold=0.0):
    count = 0
    for value in values:
        if value >= threshold:
            count += 1
    return count


assert count_elements_above_threshold([1, 2, 3]) == 3  # For example.
````

#### &#9989;&nbsp; Question 1.5 (6 points): Is this a probability distribution?

A *probability distribution* is a list of non-negative numbers that sums to one. For example, we can describe a fair coin by the probability distribution `[0.5, 0.5]` corresponding to the probability of heads and tails.

Write a function that inputs a list of numbers and returns `True` if the list of numbers is a probability distribution, else returns `False`. This function should use at least one of the functions you've written above.

Examples:
- `[0, 1/3, 1/3, 1/3]` is a probability distribution.
- `[-1, 1, 1]` is not a probability distribution because it has a negative number.
- `[1/2, 1/2, 1/2]` is not a probability distribution because its values don't sum to one.

Note: Checking if the sum of values is exactly `1` can be unsafe using `==` with floating point numbers. Instead you should use the function `math.isclose` from the `math` library to do this.

````python
# Your code here.
````

````python
### ANSWER
import math


def is_probabiliy_distribution(values):
    if count_elements_above_threshold(values, threshold=0.0) == len(values) and math.isclose(sum(values), 1.0):
        return True
    return False
````

````python
assert is_probabiliy_distribution([0, 1 / 3, 1 / 3, 1 / 3])
assert not is_probabiliy_distribution([-1, 1, 1])
assert not is_probabiliy_distribution([1 / 2, 1 / 2, 1 / 2])
````

## 2. Exploring matrices (20 points)

A matrix (plural: matrices) is a two-dimensional array of numbers such as

$$ A = \left[ \begin{matrix} 1 & 2 & 3 \\ -3 & 5 & 4 \end{matrix} \right] . $$

Matrices are used in many areas of mathematics, computer science, physics, machine learning, economics, and more.

In Python, matrices can be represented as lists of lists. For example, the above matrix can be represented in Python as below.

````python
# Example of a matrix as a list of lists.
matrix = [
    [1, 2, 3],
    [-3, 5, 4]
]
````

This matrix has two *rows* and three *columns*. We index elements in a matrix first by row $r$ and then by column $c$, and use the notation $A_{rc}$ for the element in the $r$th row and $c$th column. For the above matrix:
- $A_{00} = 1$,
- $A_{01} = 2$,
- $A_{02} = 3$,
- $A_{10} = -3$,
- $A_{11} = 5$,
- and $A_{12} = 4$.

In Python, the indexing follows the same format.

````python
# Example of indexing a matrix.
print("A_00 =", matrix[0][0])
print("A_01 =", matrix[0][1])
````

#### &#9989;&nbsp; Question 2.1 (5 points): Are these matrices equal?

Two matrices $A$ and $B$ are equal if:
- They have the same *shape*: that is, the number of rows and number of columns are equal.
- Every element is equal: that is, $A_{rc} = B_{rc}$ for every $r, c$.

1. Write a function which inputs two matrices (as lists of lists of numbers) and returns `True` if they are equal, else returns `False`. Check your function on at least two example inputs to show its working correctly.

````python
# Your code here.
````

````python
### ANSWER
def is_equal(matrixA, matrixB):
    # Check the number of rows are equal.
    if len(matrixA) != len(matrixB):
        return False

    # Check the number of columns are equal.
    for i in range(len(matrixA)):
        if len(matrixA[i]) != len(matrixB[i]):
            return False

    # Check the elements are equal.
    for i in range(len(matrixA)):
        for j in range(len(matrixA)):
            if not math.isclose(matrixA[i][j], matrixB[i][j]):
                return False
    return True


# Some example tests.
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]
assert is_equal(matrix1, matrix1)

matrix2 = [
    [1, 0],
    [0, 1]
]
assert not is_equal(matrix1, matrix2)
assert is_equal(matrix2, matrix2)

matrix3 = [
    [-999, 2, 3],
    [4, 5, 6]
]
assert not is_equal(matrix3, matrix1)

not_a_matrix = [
    [1],
    [4, 5, 6]
]
assert not is_equal(matrix1, not_a_matrix)  # This test may fail in students' code if they didn't check the shapes.
````

#### &#9989;&nbsp; Question 2.2 (5 points): The sum of two matrices

Given two matrices $A$ and $B$ with the same shape, we define the sum $C = A + B$ by $C_{rc} = A_{rc} + B_{rc}$ for all $r$ and $c$. For example,

$$ \left[ \begin{matrix} 1 & 2 \\ -3 & 1 \end{matrix} \right] + \left[ \begin{matrix} 0 & 1 \\ 2 & 4 \end{matrix} \right] =  \left[ \begin{matrix} 1 + 0 & 2 + 1 \\ -3 + 2 & 1 + 4 \end{matrix} \right] = \left[ \begin{matrix} 1 & 3 \\ -1 & 5 \end{matrix} \right] . $$

Write a function that inputs two matrices (as lists of lists of numbers) and returns the sum of the two matrices. You can assume the matrices have the same shape. Test your function on at least two example inputs to show its working correctly.

````python
# Your code here.
````

````python
### ANSWER
def compute_matrix_sum(matrixA, matrixB):
    matrix_sum = []
    
    for i in range(len(matrixA)):
        row = []
        for j in range(len(matrixA[i])):
            row.append(matrixA[i][j] + matrixB[i][j])
        matrix_sum.append(row)

    return matrix_sum


# Example tests.
matrix1 = [
    [1, 2, 3],
    [-3, 5, 4]
]
print(compute_matrix_sum(matrix1, matrix1))

matrix2 = [
    [-1, -2, -3],
    [3, -5, -4]
]
print(compute_matrix_sum(matrix1, matrix2))
````

#### &#9989;&nbsp; Question 3.2 (10 points): Convert a color image to grayscale

This image is currently a color image, meaning there are values for red, green, and blue pixels at each location. Convert this to a grayscale image by applying the formula ```0.25 * red + 0.50 * green + 0.25 * blue``` **without using any loops**. (Solutions using loops will receive at most 50% of the total points.) Use Matplotlib to visualize your image in grayscale.

Example: Consider the following color image with two pixels

$$ [[40, 20, 100], [10, 200, 40]] . $$

To convert this to a grayscale image using the formula, we do

$$ [0.25 * 40 + 0.50 * 20 + 0.25 * 100, 0.25 * 20 + 0.50 * 200 + 0.25 * 40] = [45, 115] . $$

````python
# Your code here.
````

````python
### ANSWER
def convert_to_grayscale(image, weights = (0.25, 0.50, 0.25)):
    reds, greens, blues = image[:, :, 0], image[:, :, 1], image[:, :, 2]
    return weights[0] * reds + weights[1] * greens + weights[2] * blues


image = convert_to_grayscale(image)
plt.imshow(image, cmap="gray");
````

#### &#9989;&nbsp; Question 3.3 (15 points): Writing an image blurring algorithm

Oh no! One of the puppies is embarrased about how their whiskers look, and politely asked to have their face blurred out in the image.

Write an algorithm to blur out the face of the puppy on the right by doing the following:

- Replace every pixel `p` in this region by the average of all pixels in a `20 x 20` window centered at `p`.
  - For example, if the indices of `p` are `(60, 100)`, the window should be the square of pixels with corners `(20, 80)`, `(20, 120)`, `(80, 80)`, and `(80, 120)`.

Take the the region `(100, 300) x (450, 625)` to be the region of the face of the puppy on the right.

````python
# Your code here.
````

````python
### ANSWER
def blur(image, xregion  = (100, 300), yregion = (450, 625), window: int = 10):
    image = np.copy(image)
    for i in range(*xregion):
        for j in range(*yregion):
            image[i, j] = np.average(image[i - window: i + window, j - window: j + window])
    return image


plt.imshow(blur(image), cmap="gray");
````
