## 1. Let it Snow!

Snowfall is dependent on relative humidity as well as temperature [Jennings 2018](https://www.nature.com/articles/s41467-018-03629-7). High relative humidity (lots of moisture in the air) means that snowfall requires lower temperatures than in places with low relative humidity.

**Table 1: 50% snow likelihood**

|Relative Humidity (%)| Temperature (&deg; C)|
|:---:|:---:|
|90 |1.0|
|80 |2.0|
|70 |2.5|
|60 |3.1|
|50 |4.0|
|40 |5.0|

Your task is to take the following list of temperature data and determine if snowfall is likely for each value of relative humidity using a loop and if/else statements. Your code should create a new list containing the result of "Snow" or "Rain" for each temperature data point.

```python
temperature = [0.0, 1.1, 3.0, 5.0, 7.5, -0.5, 0.4] # temperature in degrees celsius
relative_humidity = [40,50,60,70,80,90]
```
