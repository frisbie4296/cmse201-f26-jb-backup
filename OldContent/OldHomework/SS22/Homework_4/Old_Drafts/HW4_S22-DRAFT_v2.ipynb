### &#9989;&nbsp; Question 1.4: (2 points)

To improve the fitting, we can try to use quintic (5th order) polynomials.

Please **plot** your new fitting functions based on the plots in Question 1.2.

```python
# Put your code here
```

```python
### ANSWER ###

# 2pts for subplots

x_pick = x[3 : 18]
y_pick = y[3 : 18]
z_pick = z[3 : 18]

def my_fit_quintic(coord, para0, para1, para2, para3, para4):
    return para4 * coord ** 4 + para3 * coord ** 3 + para2 * coord ** 2 + para1 * coord + para0

popt_y1, _ = curve_fit(my_fit_linear, x_pick, y_pick)
popt_z1, _ = curve_fit(my_fit_linear, x_pick, z_pick)
popt_y5, _ = curve_fit(my_fit_quintic, x_pick, y_pick)
popt_z5, _ = curve_fit(my_fit_quintic, x_pick, z_pick)

plt.figure(figsize = (10, 7))

plt.subplot(1, 2, 1)
plt.title('pure')
plt.plot(x, y, 'o', label = 'data')
plt.plot(x, my_fit_linear(x, popt_y1[0], popt_y1[1]), label = 'linear fit')
plt.plot(x, my_fit_quintic(x, popt_y5[0], popt_y5[1], popt_y5[2], popt_y5[3], popt_y5[4]), label = 'quintic fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.subplot(1, 2, 2)
plt.title('noisy')
plt.plot(x, z, 'o', label = 'data')
plt.plot(x, my_fit_linear(x, popt_z1[0], popt_z1[1]), label = 'linear fit')
plt.plot(x, my_fit_quintic(x, popt_z5[0], popt_z5[1], popt_z5[2], popt_z5[3], popt_z5[4]), label = 'quintic fit')
plt.xlabel('x')
plt.ylabel('z')
plt.legend()

plt.tight_layout()
plt.show()
```

### &#9989;&nbsp; Question 1.6: (7 points)

A simple approach that evaluates the quality of the fitting is to measure the mean squared error (MSE). That is, given the data $\{ (a_i, b_i) \} _{i = 1} ^N$, and the fitted model $f$ (i.e., fitted model parameters), we shall calculate the  mean squared error
$$ MSE := \frac{1}{N}\sum_{i=1}^{N} (f(a_i) - b_i) ^2.$$

First, please **write** functions *MSE_linear* and *MSE_quintic*, which has several inputs and one output:
- Input 1: the data $a$ as a *numpy.ndarray*,
- Input 2: the data $b$ as a *numpy.ndarray* (with the same size as Input 1),
- Input 3, ...: the fitted model parameters (representating the fitted model $f$) as a *float*,
- Output: the mean squared error $MSE$ as a *float*.

Then, **calculate** the MSE with respect to the data and fitted models above:
1. central noisy data, with linear fit,
1. central noisy data, with quintic fit,
1. all noisy data, with linear fit,
1. all noisy data, with quintic fit.

Here *central* means the central 16 data points out of 21 data points.

Finally, **answer** the following questions:
1. When considering the noisy data, which fit model has less MSE?
1. When considering all the data, which fit model has less MSE?

```python
# Put your code here
```

```python
### Answer ###

# 3pts for all MSE functions
# 2pt for all MSE calculations
# 1pt for each answer

def MSE_linear(a, b, para0, para1):
    return np.sum((my_fit_linear(a, para0, para1) - b) ** 2) / len(a)
def MSE_quintic(a, b, para0, para1, para2, para3, para4):
    return np.sum((my_fit_quintic(a, para0, para1, para2, para3, para4) - b) ** 2) / len(a)

print('For central noisy data, with linear fit, the MSE is '  + str(MSE_linear(x_pick, z_pick, popt_z1[0], popt_z1[1])))
print('For central noisy data, with quintic fit, the MSE is ' + str(MSE_quintic(x_pick, z_pick, popt_z5[0], popt_z5[1], popt_z5[2], popt_z5[3], popt_z5[4])))
print('For all noisy data, with linear fit, the MSE is '      + str(MSE_linear(x, z, popt_z1[0], popt_z1[1])))
print('For all noisy data, with quintic fit, the MSE is '     + str(MSE_quintic(x, z, popt_z5[0], popt_z5[1], popt_z5[2], popt_z5[3], popt_z5[4])))

# - When considering the noisy data, the quintic fit model has less MSE.
# - When considering all the data, the linear fit model has less MSE.
```

### &#9989;&nbsp; Question 2.2: (4 points)

This data set comes from an unknown model prediction and stores a snapshot of a partical system flowing between two walls.
In this data set, there stores the coordinates (distance against to one of the wall) of a part of the particles (*coord*) as well as their velocities in three spatial directions (*vel1*, *vel2*, and *vel3*). We are going to investigate these quantities.

First, you may notice one of the velocity data is quite different from two others, please **answer** the question:
- Which velocity is special, *vel1*, *vel2*, or *vel3*? Why?

Then, we investigate the relationship between partical coordinates and their velocities, by data fitting with *scipy.optimize.curve_fit*. We just work on that special velocity data. Please **fit** any function to the coordinates and the special velocity. The restriction is, you may have at most 3 parameters in fitting, and the MSE should be less than 0.1. <font size=+3>&#9998;</font> *Put your answers here.*

```python
# Put your code here
```

```python
### ANSWER ###

# 1pt for answer
# 1pt for curve_fit
# 1pt for MSE calculation
# 1pt for plot

# - The second velocity (vel2) is special since the mean value is quite different.

def my_fit3(coord, para0, para1, para2):
    return para2 * np.square(coord) + para1 * coord + para0
def MSE3(a, b, test_fun, para0, para1, para2):
    return np.sum((test_fun(a, para0, para1, para2) - b) ** 2) / len(a)

popt3, _ = curve_fit(my_fit3, polymers['coord'], polymers['vel2'])
print('The MSE is ', str(MSE3(polymers['coord'], polymers['vel2'], my_fit3, popt3[0], popt3[1], popt3[2])))

plt.figure(figsize = (10, 7))
plt.scatter(polymers['coord'], polymers['vel2'], label = 'data')
plt.plot(polymers['coord'].sort_values(), my_fit3(polymers['coord'].sort_values(), popt3[0], popt3[1], popt3[2]), 'r', label = 'fit')
plt.xlabel('coordinate')
plt.ylabel('the second velocity')
plt.legend()
plt.show()
```

### &#9989;&nbsp; Question 2.3: (2 points)

The data we load contains only a part of particles on the left half of the domain.
