### 2.2.3 Developing your function (5 points)

&#9989;&nbsp; In the cell below, write your percent difference function and test it using the test data. Feel free to add additional cells as needed.

```python
# test data
test_open = [40,42,48,36,40,45]
test_close = [41,46,37,38,44,45,42]
test_days = list(range(len(test_open)))
```

```python
#put your code here
```

```python
### ANSWER

def pct_diff(open_list,close_list,days_list):
    import matplotlib.pyplot as plt # they may forget this or call matplotlib outside
    percent_diff_list = []
    for i in range(len(open_list)):
        percent_diff_list.append((close_list[i]-open_list[i])/open_list[i])
    plt.plot(days_list,percent_diff_list)
    plt.xlabel('days')
    plt.ylabel('% difference')
    plt.title('Stock Prices')
    return percent_diff_list

pct_diff(test_open,test_close,test_days)

# Rubric
# 1 pt for def
# 1 pt for calculating pct diff
# 1 pt for plot
# 1 pt for correct and readable plot labels
# 1 pt for test
```

```python
### ANSWER

def my_plots(x1,y1,x2,y2,x3,y3,label1,label2,label3,xlabel,ylabel,title):
    plt.suptitle(title)
    plt.subplot(1,3,1)
    plt.plot(x1,y1,color="forestgreen",label=label1)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend(loc='best')
    
    plt.subplot(1,3,2)
    plt.plot(x2,y2,color="goldenrod",label=label2)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend(loc='best')
    
    plt.subplot(1,3,3)
    plt.plot(x3,y3,color="teal",label=label3)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend(loc='best')
    
    plt.tight_layout()
    
my_plots(x_values,sin,x_values,cos,x_values,tan,'sin','cos','tan','x','y','Trigonometric Functions')

# Rubric
# 1pt for inputs
# 1pt for labels
# 1pt for subplot
# 2pts for correctly plotting data
# 1pt for test
```

```python
### ANSWER
# Another possible solution

import matplotlib.pyplot as plt
import numpy as np
import math

def my_plots(x,y,label,xlabel,ylabel,color,dims,num):
#     plt.suptitle(title)
    plt.subplot(dims[0],dims[1],num)
    plt.plot(x,y,color=color,label=label)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend(loc='best')

x_values = np.linspace(-math.pi,math.pi,100)

sin = np.sin(x_values)
cos = np.cos(x_values)
tan = np.tan(x_values)
    
plt.figure(figsize=(12,6))
my_plots(x_values,sin,'sin','x','y',"forestgreen",[1,3],1)
my_plots(x_values,cos,'cos','x','y',"goldenrod",[1,3],2)
my_plots(x_values,tan,'tan','x','y',"teal",[1,3],3)
plt.tight_layout()
plt.show()
```
