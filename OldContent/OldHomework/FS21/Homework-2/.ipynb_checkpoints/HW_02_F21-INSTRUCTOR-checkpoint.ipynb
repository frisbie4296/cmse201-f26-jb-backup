## Part 3: Modules and Plotting

The math module contains the basic trigonometric functions (sine, cosine, tangent, arcsine, arcosine,and arctangent, etc.). 

&#9989;&nbsp; In the cell below, choose three of the trigonometric functions and generate x and y values for each. You should use the `math` module and/or the `numpy` module.

```python
# put your answer here
```

```python
### ANSWER

import math
import numpy
x_values = np.linspace(-math.pi,math.pi,100)

sin = np.sin(x_values)
cos = np.cos(x_values)
tan = np.tan(x_values)
```

### 3.2 Plotting

&#9989;&nbsp; In the cell below, make a plot for one of your trigonometric functions. Make sure to include axis labels, a title, and a legend.

```python
# put your answer here
```

```python
### ANSWER

import matplotlib.pyplot as plt

plt.plot(x_values,sin,label='sin')
plt.legend(loc='best')
plt.xlabel('x')
plt.ylabel('y')
plt.tight_layout()
plt.title('sin(x)')
```

### 3.3 Streamlining plots

In CMSE 201, you will be making plots often. To help future you save time, you are going to write a function to use when you make plots!

&#9989;&nbsp; In the cells below, write a function that takes the x and y values from all three of your trigonometric functions above and appropriate labels and returns a plot with a panel for each trigonometric function. You should use plt.subplot() and make use of labels, line shapes, and colors to make your plots clear.

```python
# put your answer here
```

```python
### ANSWER

def my_plots(x1,y1,x2,y2,x3,y3,label1,label2,label3,xlabel,ylabel,title):
    plt.suptitle(title)
    plt.subplot(1,3,1)
    plt.plot(x1,y1,color="forestgreen",label=label1)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend(loc='best')
    
    plt.subplot(1,3,2)
    plt.plot(x2,y2,color="goldenrod",label=label2)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend(loc='best')
    
    plt.subplot(1,3,3)
    plt.plot(x3,y3,color="teal",label=label3)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend(loc='best')
    
    plt.tight_layout()
    
my_plots(x_values,sin,x_values,cos,x_values,tan,'sin','cos','tan','x','y','Trigonometric Functions')
```
