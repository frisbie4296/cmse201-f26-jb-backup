## The 2D Schelling Model

In this Homework you're going to implement  a simple version of the 2D Schelling model. Two kinds of polygons (triangles and squares) are living on a 2-dimensional grid.  You're essentially going to be coding up a version of the [Parable of Polygons](http://ncase.me/polygons/) that you've reviewed previously. 

The steps to initialize the model are as follows:

* Create a 2D Numpy array of $N\times N$ cells (default $N=20$), which will be the board/world.  In this implementation, the number 0 will represent an empty cell, 1 will represent a triangle, and 2 will represent a square.
* Initialize the array with an approximate  fraction $f_{tri}$ of triangles (default $f_{tri} = 1/3$),   $f_{sqr}$ of squares (default $f_{sqr} = 1/3$) and $1 - (f_{tri} + f_{sqr})$ empty cells, with each cell having a random chance of being one of these options.

Then, evolve this model using the following rules:

* A polygon's **neighbors** are the polygons in adjacent cells. In previous assignments we have worked with von Neumann neighborhoods (just left/right/up/down), but for this 2D Shelling model we will use Moore neighborhoods  (left/right/up/down **and the diagonals**).

* A polygon is **happy** if at least `happy_min_frac` (default 1/3) of its neighbors are like it, or if it has no neighbors.
* A polygon is considered to be **segregated** if all of its neighbors are like it, or if it has no neighbors.

The board is evolved by **randomly selecting a single *unhappy* polygon and moving it to a random empty cell**.  This continues until every polygon is happy.  The level of segregation of the board is tracked with each step.

##  The plotting function 

Similar to the `plotgrid` function we have used in previous assingments, we need a function that plots our happy/unhappy polygonal world. 

The function `plot_board`, provided below, takes two "boards" of the same size as inputs and plots the world. 
The first input is `game_board`, a 2D numpy array whose entries are either 0 (empty cell), 1 (a triangle) or 2 (a square).
The second input is `unhappy_board`, a 2D numpy array of the same size as `game_board`, whose entries are either 0 (the cell is either empty, or the resident polygon is happy) or 1 (the resident polygon is unhappy).
**Review the code below and make sure you understand how it works**

```python
def plot_board(game_board, unhappy_board):
    '''
    This function takes  two 2D arrays and uses pyplot to make a plot of the
    polygon "world"
    
    It returns no value.
    '''
    # First we create masks do determine where the triangles are, and where the squares are
    triangle_grid = game_board == 1
    square_grid = game_board == 2
    
    # Next, we create masks to find where the happy/unhappy polygons live
    happy_grid = unhappy_board == 0
    unhappy_grid = unhappy_board == 1
    
   
    # You might not be completely familiar with some of the code in this section
    # It uses "logical_and" to ...
    # It uses "nonzero" to ...
    happy_tr_i,happy_tr_j = np.nonzero(np.logical_and(triangle_grid, happy_grid))
    happy_sq_i,happy_sq_j = np.nonzero(np.logical_and(square_grid, happy_grid))
    
    unhappy_tr_i,unhappy_tr_j = np.nonzero(np.logical_and(triangle_grid, unhappy_grid))
    unhappy_sq_i,unhappy_sq_j = np.nonzero(np.logical_and(square_grid, unhappy_grid))
    
    
    # Turn the [i,j] indices into (x,y) coordinates to plot
    happy_tr_x = happy_tr_j
    happy_tr_y = game_board.shape[0] - 1 - happy_tr_i
    happy_sq_x = happy_sq_j
    happy_sq_y = game_board.shape[0] - 1 - happy_sq_i
    
    unhappy_tr_x = unhappy_tr_j
    unhappy_tr_y = game_board.shape[0] - 1 - unhappy_tr_i
    unhappy_sq_x = unhappy_sq_j
    unhappy_sq_y = game_board.shape[0] - 1 - unhappy_sq_i

    # Plot the polygons
    # Make them larger than the default so they're easy to see!
    plt.plot(happy_tr_x, happy_tr_y, 'g^', markersize=10)
    plt.plot(happy_sq_x, happy_sq_y, 'bs', markersize=10)

    plt.plot(unhappy_tr_x, unhappy_tr_y, 'r^', markersize=10)
    plt.plot(unhappy_sq_x, unhappy_sq_y, 'rs', markersize=10)
    
    
    # Do a bit of formatting to get the plot to look the way we want
    plt.tick_params(axis='both', which='both',
                    bottom=False, top=False, left=False, right=False,
                    labelbottom=False, labelleft=False)
```

#  Who is unhappy?

The next step is to **write a function that goes through a 2D board and determines what  polygons are happy/unhappy**. 
The function, which we will name `is_unhappy`, should take as inputs a board (e.g., generated with `set_board`), as well as a float between 0 and 1 defining what percentage of your neighbors need to be like you, in order for you to be a happy polygon. 

The output of `is_unhappy` will be a 2D numpy array of the same size as the input board, and each entry is either a 1 to indicate that there is a polygon there who is unhappy, or 0 which would indicate that there is either a happy polygon there or that the location is empty.

To begin we define two auxiliary functions, `onBoard` and `getNeighborValues`, which we have used in previous assingments. **Do you remember what they do?**

```python
def onBoard(i,j,board):
    if i <= board.shape[0]-1 and i >= 0 and j <= board.shape[1]-1 and j >= 0:
        return True
    else:
        return False
```

```python
def getNeighborValues(i,j, board):
    # Von Neumman neighborhood
    neighborhood = [(i-1, j), (i, j-1), (i+1, j), (i, j+1)]
    
    neighbor_values = []
    for neighbor in neighborhood:
        if onBoard(neighbor[0], neighbor[1], board):
            neighbor_values.append(board[neighbor[0], neighbor[1]])
    
    return neighbor_values
```

Next we define the `is_unhappy` function below. Notice how it uses `getNeighborValues`.

```python
def is_unhappy(game_board, happy_min_frac):
    '''
    This function should take in:
    
    1. The game board with the current state of the polygon "world"
    2. The fraction of your neighbors must be like you in order for you to be happy
      
    The function should return a 2D board of the same shape and size of the
    polygon "world" but containing 0s and 1s, to represent which polygons
    are currently unhappy (1 should indicate the polygon is unhappy and 0
    means the polygon is happy or the cell is empty)
    '''

    # Start by creating an empty array the same size and shape of the game board
    unhappy_board = np.zeros(game_board.shape, dtype='int')
    
    # Loop over the board
    for i in range(game_board.shape[0]):
        for j in range(game_board.shape[1]):
            # Start off assuming each cell is happy
            happy = True

            # A cell can only be unhappy if it's not empty
            if game_board[i,j] > 0:
                
                neighbors = getNeighborValues(i,j,game_board)
                
                num_neighbors = float(len(neighbors))
                
                num_alike = (np.array(neighbors) == game_board[i,j]).sum()
                
                num_empty = (np.array(neighbors) == 0).sum()
                
                if num_empty < num_neighbors:
                    frac_alike = num_alike/(num_neighbors - num_empty)
                    
                    if frac_alike < happy_min_frac:
                        unhappy_board[i,j] = 1
    
    return unhappy_board
```

```python
def find_unhappy(unhappy_board):
    '''
    Takes in the unhappy board, finds a random unhappy cell, and returns the indices 
    of that unhappy cell.
    '''
    # Find locations [i,j] of unhappy cells
    unhappy_i,unhappy_j = np.nonzero(unhappy_board)
    
    # Random selection of unhappy cell
    num_unhappy = unhappy_board.sum()   # number of unhappy cells, THINK!
    rand_unhappy = random.randint(0, num_unhappy-1)    
    
    # Return location (indices [i,j]) of random unhappy cell
    return unhappy_i[rand_unhappy], unhappy_j[rand_unhappy]
```
