### &#9989;&nbsp; Question 1.2: Generate and Plot Data (4 points)

Consider a function $y=f(x)$ such that $f(x) = 2x$ for $0 \leq x \leq 0.5$ and $f(x)=2-2x$ for $0.5 \leq x \leq 1$. Plot this function over 200 equispaced points ($x$) in the range of $x = $ 0 to 1 (end points included). Then, generate a noisy version of this function by adding Gaussian noise with mean zero and standard deviation 0.2 to all of the $y$ values from this function. Essentially, make $y_{noisy} = f(x) \, + $ Gaussian($x$). Plot the noisy data (versus $x$ values) in a new plot. Label all axes on the plots.

```python
# Put your code here
```

```python
### ANSWER ###
# 1 point for setting up the function correctly, 1 point for plotting it, 1 point for adding noise correctly
# and 1 point for noisy data plot
x=np.linspace(0,1,200)

y=np.zeros_like(x)
for i in range(x.size):
    if x[i] <= 0.5:
        y[i]=2*x[i]
    else:
        y[i]=2 - 2*x[i]
        
plt.figure()
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('f(x)')

y_noisy = y + np.random.normal(0,0.2,x.size)
plt.figure()
plt.plot(x,y_noisy)
plt.xlabel('x')
plt.ylabel('Noisy Data')
```

How would you use NumPy's `polyfit` function to fit a model like $f(x)$ (i.e., polynomial in two pieces) to the corrupted or noisy data? Explain your approach. Use only first order polynomials for this. Print all the fitted model parameters. Then use `poly1d` to help **plot the data together with the best fit model**. Include a legend to make it clear what information represents the data and what information represents the best fit. *Hint: You may fit more than one polynomial to this type of data.*

```python
### ANSWER ###

# 1 pt for explanation of method
# 2 pts for doing poly fit on each segment and 1 pt for printing parameters
# 1 pt for plotting correctly.

#Break the data into two parts in [0,0.5] and (0.5,1] and fit different linear models for each part.

x1=x[x<=0.5]
y1=y_noisy[x<=0.5]

x2=x[x>0.5]
y2=y_noisy[x>0.5]

fit_poly_1 = np.polyfit( x1, y1, 1)
first_order_fit_1 = np.poly1d( fit_poly_1 )
print('The parameters on the first half interval are',fit_poly_1)

fit_poly_2 = np.polyfit( x2, y2, 1)
first_order_fit_2 = np.poly1d( fit_poly_2 )
print('The parameters on the second half interval are',fit_poly_2)

plt.plot( x, y_noisy, "*", label='data' )
plt.plot( x1, first_order_fit_1(x1), 'b',label="first part fit" )
plt.plot( x2,first_order_fit_2(x2),'r', label="second part fit" )
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.legend();
```

### &#9989;&nbsp; Question 1.4: Using Curve Fit for the data (7 points)

Write a function called `fitting_function` that takes the following arguments: $x$, $m_1$, $b_1$, $m_2$, $b_2$, where $x$ is a NumPy array and the rest are all floating point parameter values corresponding to the slopes of the lines/pieces and their intercepts. The function should return an array with values $f(x) = m_1*x + b_1$ when $0 \leq x \leq 0.5$ and $f(x) = m_2*x + b_2$ where $0.5 \leq x \leq 1$.

Use SciPy's `curve_fit` function to find the best fit model parameters for the corrupted data using your `fitting_function`. **Plot the data and the best fit. Label all axes and include a legend. Also print the resulting parameters.**

```python
# Put your pseudocode and code here
```

```python
### ANSWER ###

# 3 pts for writing function correctly (getting function syntax right and including condition to choose between 2 cases)
# 2 pt for calling curve_fit correctly and getting parameters right
# 2 pts for plotting the data and best fit curve including labeling all axes.

def fitting_function(x, m1, b1, m2, b2):
    y=np.zeros_like(x)
    for i in range(x.size):
        if x[i] <= 0.5:
            y[i]=m1*x[i] + b1
        else:
            y[i] = m2*x[i] + b2
    return y

popt, pcov = curve_fit(fitting_function, x, y_noisy)
print('The best parameters are ', popt)

plt.plot( x, y_noisy, "*", label='data' )
plt.plot( x, fitting_function(x,popt[0],popt[1],popt[2],popt[3]), 'r',label="Fit Model" )
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.legend();
```

```python
### ANSWER ###

# 3 pts for using polyfit for the 3 columns (1 pt per column), 1 pt for justification of polynomial degree chosen
# 3 pts for the three plots including labeling axes and adding legends.

# degree 3 polynomial seems to give decent fit to data. Higher order polynomials may overfit.
# degree 2 polynomial gives slightly worse fit to initial dates but works better for future prediction 
# than degree 3 polynomial.

num_points=300
dates_data = weather['date_changed'][0:num_points]
mean_data = weather['actual_mean_temp'][0:num_points]
min_data = weather['actual_min_temp'][0:num_points]
max_data = weather['actual_max_temp'][0:num_points]
deg=2

fit_poly_mean = np.polyfit(dates_data,mean_data,deg)
higher_order_fit_mean = np.poly1d( fit_poly_mean )

fit_poly_min = np.polyfit(dates_data, min_data, deg)
higher_order_fit_min = np.poly1d( fit_poly_min )

fit_poly_max = np.polyfit(dates_data, max_data, deg)
higher_order_fit_max = np.poly1d( fit_poly_max )

plt.figure()
plt.plot_date(dates_data, mean_data, "*", label='data' )
plt.plot(dates_data, higher_order_fit_mean(dates_data), 'b',label="fit model" )
plt.xticks(rotation=30)
plt.xlabel("Date")
plt.ylabel("Mean Temperatures")
plt.grid()
plt.legend();
#plt.savefig('mean_temp_vs_data.png')
plt.figure()
plt.plot_date(dates_data, min_data, "*", label='data' )
plt.plot(dates_data, higher_order_fit_min(dates_data), 'b',label="fit model" )
plt.xticks(rotation=30)
plt.xlabel("Date")
plt.ylabel("Min Temperatures")
plt.grid()
plt.legend();

plt.figure()
plt.plot_date(dates_data, max_data, "*", label='data' )
plt.plot(dates_data, higher_order_fit_max(dates_data), 'b',label="fit model" )
plt.xticks(rotation=30)
plt.xlabel("Date")
plt.ylabel("Max Temperatures")
plt.grid()
plt.legend();
```
