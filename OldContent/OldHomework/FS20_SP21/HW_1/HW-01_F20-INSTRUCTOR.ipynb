```python
### ANSWER

#EXAMPLE SOLUTION

fruit = ["Oranges", "Grapes", "Clementines"]
reasoning = ["They remind me of Holland.","Purple is the best color.","They are the Corgis of oranges."]

for ii in range(len(fruit)):
    print("I like ",fruit[ii]," because ",reasoning[ii])
```

## C. Bringing it All Together (6 Points)

You should have two pieces of code: one that calculates the average of a list and another that goes through all of the dictionary's temperature values.

&#9989;&nbsp; **Use these two pieces to generate a final piece of code that creates a new dictionary with the *average* temperature for each city.** 

```python
### ANSWER 

#EXAMPLE SOLUTION

temperature_dictionary = {"Paris": [84, 70, 64, 72, 81, 69, 62], 
                         "Nairobi": [63, 64, 62, 63, 62, 67],
                         "Goa": [79, 82, 80, 82, 78, 82, 81, 81, 83, 82],
                         "Jakarta": [85, 82, 85, 85, 85, 87],
                         "Manaus": [81, 83, 83, 85, 77, 79, 79]}

average_temp_dict = {}

for city in temperature_dictionary:
    list_of_temperatures = temperature_dictionary[city]
    average_temp = sum(list_of_temperatures)/len(list_of_temperatures)
    average_temp_dict[city] = average_temp

print(average_temp_dict)
```
