```python
### ANSWER ###

# 1 point for identifying that the discount should be applied as a percentage, not subtracted directly
# 2 points for fixing the discount logic by changing subtraction to percentage-based calculation
# 1 point for correctly keeping the conditional structure where the discount applies only when total > 100
# 1 point for ensuring the function returns the correct result with both test cases

##### NOTE TO GRADERS:
'''The original function incorrectly subtracted a flat discount amount instead of applying a percentage.
Students should receive full credit if the bug is fixed, even if the solution slightly differs but works correctly.'''

# Corrected function
def calculate_total(prices, discount_percentage):
    total = sum(prices)
    if total > 100:
        total = total * (1 - discount_percentage / 100)  # Apply discount as a percentage
    return total
```
