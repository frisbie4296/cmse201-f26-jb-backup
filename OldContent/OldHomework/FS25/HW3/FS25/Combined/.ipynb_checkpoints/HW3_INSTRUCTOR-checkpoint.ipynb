## Part 3: Fitting curves to data - Predicting Exam Scores (22 points)

[Back to Top](#toc)

Now that we’ve visualized our data, let’s ask:

What is the relationship between **exam score** and each of these variables:

* **hours_studied**

* **sleep_hours**

Can we predict exam score for both of these variables with a simple linear model?

### Part 3.1: Model
In the above plots we notice that `exam_score` tends to **increase** as `hours_studied` increases in value. Specifically, we can describe this relationship with a linear model:

$$ exam\_score = m \times variable + b $$

Where $variable$ is the predictor variable (e.g., `hours_studied`, or `sleep_hours`), $m$ is the slope of the line, and $b$ is the intercept.

&#9989;&nbsp; **Do this (4 points):**  
- Write a function called `exam_model` that calculates exam score based on a given predictor variable using the equation above.  
- The equation should be structured so that it can be used with the `curve_fit` function in the next section.  


````python
# Put your code here
````

````python
### ANSWER ###

def exam_model(x, m, b):
    return  x*m +b

# 1 point for correct name
# 1 point for correct order of the inputs. No partial credit
# 1 point for correct math. No partial credit
# 1 point for return statement
````

### Part 3.2: Fit the model

We’ll now use the `curve_fit` function from SciPy to find the best-fitting line that predicts exam scores.

&#9989;&nbsp; **Do this (10 points):** 

Use the dataset `student_exam_scores_cleaned.csv` 

For each predictor variable (`hours_studied`, and `sleep_hours`):

- Use the `curve_fit` with your `exam_model` function to find the slope ($m$) and intercept ($b$). (3 pts each, 6 total)

- Print out the value of the slope $m$, write "The value of the slope is..." (1 pt each, 2 total)

- Print out the value of the intercept $b$, write "The value of the intercept is..." (1 pt each, 2 total)

````python
# Put your code here
````

````python
### ANSWER ###

import pandas as pd
from scipy.optimize import curve_fit

# Load the data
exam_scores_df = pd.read_csv("student_exam_scores_cleaned.csv")

# Loop over predictor variables
predictors = ["hours_studied", "sleep_hours"]

for var in predictors:
    popt, pcov = curve_fit(exam_model, exam_scores_df[var], exam_scores_df["exam_score"])
    m, b = popt
    print(f"For {var}: The value of the slope is {m:.3f}, and the value of the intercept is {b:.3f}")

    
# They do NOT have to use a for loop to receive full credit here
# 2 pts for calling curve_fit (1 pt for each variable)
# 2 pts for correct placement of exam_model (1 pt for each variable)
# 2 pts for correct placement of x and y values (1 pt for each variable)

# 2 pts for printing slope (1 pt for each variable)
# 2 pts for printing intercept (1 pt for each variable)
````

### Part 3.3 Check your model (8 points)

&#9989;&nbsp; **Do this (6 points):**  Make a plot comparing your model and the data.

- Plot the actual data (2 points)
- Plot your modeled data (2 points)
- Use x and y axis labels and title (1 point)
- Adjust the color, and size, and alpha of the datapoints. Make sure the plot of your model is a different color than the datapoints so that you can see it. (1 point)

````python
# Put your code here
````

````python
### ANSWER ###
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 12))
plt.subplot(2,1,1)
plt.scatter(exam_scores_df["hours_studied"], exam_scores_df["exam_score"], s=15, alpha=0.3, label="Data")
popt, pcov = curve_fit(exam_model, exam_scores_df["hours_studied"], exam_scores_df["exam_score"])
m, b = popt
plt.plot(exam_scores_df["hours_studied"], exam_model(exam_scores_df["hours_studied"], m, b),
         color="red", linewidth=2, label="Model")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Exam Score vs. Study Hours")
plt.legend()

fig, ax = plt.subplots(figsize=(12, 12))
plt.subplot(2,1,2)
plt.scatter(exam_scores_df["sleep_hours"], exam_scores_df["exam_score"], s=15, alpha=0.3, label="Data")
popt, pcov = curve_fit(exam_model, exam_scores_df["sleep_hours"], exam_scores_df["exam_score"])
m, b = popt
plt.plot(exam_scores_df["sleep_hours"], exam_model(exam_scores_df["sleep_hours"], m, b),
         color="red", linewidth=2, label="Model")
plt.xlabel("Sleep Hours")
plt.ylabel("Exam Score")
plt.title("Exam Score vs. Sleep Hours")
plt.legend()


# 2 points for plotting the data (1 pt each plot)
# 2 points for plotting the fitted line (1 pt each plot)
# 1 point for title and x and y axis labels (0.5 pts each plot)
# 1 point for adjusting color/size/alpha of data and making sure line is different color (0.5 each plot, 0 points if no attempt)
````
