**1.4.1** ✅ **Task (6 points)**

You want to know whether students’ study habits have changed over time, and if so, whether they tend to study **more or less** in recent years.

You decide to analyze the **median number of hours studied** for the following groups based on their **previous_scores** range, as well as for high-performing students:

* 0–20  
* 21–40  
* 41–60  
* 61–80  
* 81–100  

**For each of the above score ranges, calculate the median value of `hours_studied`.  
For each range, print a statement such as:**  
> "The median number of hours studied for students with previous scores between (*insert range here*) is: x hours."

(6 pts)

**Hint:** You can create Boolean masks for value ranges as follows:
```python
data[(data["previous_scores"] >= lower_bound) & (data["previous_scores"] <= upper_bound)]


````python
## your code here
````

````python
### ANSWER ###

median_0_20 = student_df['hours_studied'][(student_df['previous_scores'] >= 0) & (student_df['previous_scores'] <= 20)].median()
median_21_40 = student_df['hours_studied'][(student_df['previous_scores'] >= 21) & (student_df['previous_scores'] <= 40)].median()
median_41_60 = student_df['hours_studied'][(student_df['previous_scores'] >= 41) & (student_df['previous_scores'] <= 60)].median()
median_61_80 = student_df['hours_studied'][(student_df['previous_scores'] >= 61) & (student_df['previous_scores'] <= 80)].median()
median_81_100 = student_df['hours_studied'][(student_df['previous_scores'] >= 81) & (student_df['previous_scores'] <= 100)].median()

print("The median number of study hours for previous scores 0–20 is:", median_0_20)
print("The median number of study hours for previous scores 21–40 is:", median_21_40)
print("The median number of study hours for previous scores 41–60 is:", median_41_60)
print("The median number of study hours for previous scores 61–80 is:", median_61_80)
print("The median number of study hours for previous scores 81–100 is:", median_81_100)
````
