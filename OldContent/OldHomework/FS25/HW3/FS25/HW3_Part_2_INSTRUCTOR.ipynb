```python
### ANSWER ###
bins = 25
alpha = 0.75 # optional

plt.figure(figsize=(9, 3))
plt.suptitle("Distributions of Exam Score, Study Hours, and Sleep Hours")

# exam_score
plt.subplot(1, 3, 1)
plt.hist(exam_scores_df["exam_score"], bins=bins, alpha=alpha)
plt.axvline(exam_scores_df["exam_score"].mean(), linestyle="--", label="mean")
plt.axvline(exam_scores_df["exam_score"].median(), linestyle=":", label="median")
plt.title("Exam Score")
plt.xlabel("score")
plt.ylabel("count")
plt.legend(loc="best")

# hours_studied
plt.subplot(1, 3, 2)
plt.hist(exam_scores_df["hours_studied"], bins=bins, alpha=alpha)
plt.axvline(exam_scores_df["hours_studied"].mean(), linestyle="--", label="mean")
plt.axvline(exam_scores_df["hours_studied"].median(), linestyle=":", label="median")
plt.title("Hours Studied")
plt.xlabel("hours")
plt.ylabel("count")
plt.legend(loc="best")

# sleep_hours
plt.subplot(1, 3, 3)
plt.hist(exam_scores_df["sleep_hours"], bins=bins, alpha=alpha)
plt.axvline(exam_scores_df["sleep_hours"].mean(), linestyle="--", label="mean")
plt.axvline(exam_scores_df["sleep_hours"].median(), linestyle=":", label="median")
plt.title("Sleep Hours")
plt.xlabel("hours/night")
plt.ylabel("count")
plt.legend(loc="best")

plt.tight_layout()
plt.show()
plt.savefig("student_distributions.jpg")
```
