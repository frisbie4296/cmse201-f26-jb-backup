&#9989;&nbsp; **Part 1.2 (8 points)**

For each of the four pathways or arrows in the model, give an interpretation for what actual process might be represented.

<font size=6 color="#009600">&#9998;</font> *Put your answer here*

### ANSWER
Note: student answers might vary slightly but should make sense
- $\alpha A P$ (2pts): rate at which animals eat plants
- $\sigma A$ (2pts): rate at which dead animals decompose.  Students might also talk about animal waste being a source of nitrogen
- $\beta P$ (2pts): rate at which living plants absorb nitrogen 
- $\gamma P$ (2pts): rate at which dead plants decay
