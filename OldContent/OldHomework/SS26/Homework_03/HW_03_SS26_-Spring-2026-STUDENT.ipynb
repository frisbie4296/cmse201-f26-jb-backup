The name "Mexico" itself is derived from the *mexica* people, who established Tehochtitlan which is now Mexico City. [Nahuatl and related languages](https://en.wikipedia.org/wiki/Uto-Aztecan_languages) are spoken throughout Mexico and spread all the way north into what is now the United States of America and all the way south to Central America.  

It is not surprising that domesticated plants from Mexico retain their names in Nahuatl, and that these names--and the cultural uses of the plants--have spread globally. For example, Nahuatl words like ***chīlli*** (🌶️ chile), ***cacaotl*** (🍫 chocolate), ***tomatl*** (🍅 tomato), and ***āhuacatl*** (🥑 avocado) have been incorporated into langauges around the world, just as the uses of these plants have been globally incorporated as well. Remarkably, we have a record of both Nahuatl names for plants and illustrations of them. The [De la Cruz-Badiano Codex of 1552](https://en.wikipedia.org/wiki/Libellus_de_Medicinalibus_Indorum_Herbis), written shortly after Conquest, was written by a Nahuatl physician (*tīcitl*), and demonstrated the medicinal value of plants Nahua people used, as well as documenting botanical details about plants in Mexico. $$

$$

`a` → initial scale (how much production “takes off”)

`b` → growth rate (the star of the show)

`c` → baseline offset (prevents forcing the curve through zero)
