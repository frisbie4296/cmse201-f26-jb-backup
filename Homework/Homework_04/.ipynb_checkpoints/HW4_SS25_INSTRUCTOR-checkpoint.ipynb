## Part 1. Understanding compartmental models (19 total points)

[Back to Top](#toc)





In this part of the assignment, you will:

* Interpret the components and dynamics of a basic SIR (Susceptible-Infectious-Recovered) model
* Reason through what the rates mean and how they shape the system
* Extend the model by incorporating additional real-world factors like birth, death, and immunity loss
* Practice writing ODEs for extended compartmental models (no coding required!)


The **SIR model** is a simple compartmental model used to understand how infectious diseases spread through a population. It divides the population into three groups:

- **S(t)**: Number of susceptible individuals
- **I(t)**: Number of infected individuals
- **R(t)**: Number of recovered (or removed) individuals

<img src="SIR_Basic.jpg">




The system is governed by the following ordinary differential equations:

The equations become:
\begin{align*}
\frac{dS}{dt} &= -\beta {S I} \\
\frac{dI}{dt} &= \beta {S I} - \gamma{I}\\
\frac{dR}{dt} &= \gamma I\\
\end{align*}


#### System Explanation:
- As Susceptible and Infected populations interact, some of the Susceptible population becomes Infected
- Over time, the Infected population begins to Recover from the virus


&#9989;&nbsp; **Part 1.1 (2 pts)** In your own words, explain what each of the terms in the equations represents:
   - What does the term $\beta S I$ mean?
   - What does the term $\gamma I$ mean?

<font size=6 color="#009600">&#9998;</font> *Put your answer here*

### ANSWER

* (1 pt) Anything to indicate the student understands that the term  βSI represents individuals transitioning from Susceptible to Infected.

Example: βSI represents number of people becoming infected, β represents infection rate, etc.

S is the number of people who are susceptible (can catch the disease).
I is the number of people who are infected (can spread the disease).
β is the infection rate constant — it determines how likely a susceptible person is to become infected after coming into contact with an infected person.
Together, βSI models the idea that more infections happen when there are lots of susceptible and infected people interacting, and it scales with how contagious the disease is.

* (1 pt) Anything to indicate the student understands that the term γI represents individuals transitioning from Infected to Recovered.

Example: γI represents number of people recovering, γ represents recovery rate, etc. 

I is the number of people currently infected.
γ is the recovery rate constant, which tells us how quickly infected individuals recover.
So, if γ =0.1, that means 10% of the infected population recovers per unit time (e.g., per day, if time is in days).

## Part 2. Solving differential equations numerically (17 total points)

[Back to Top](#toc)

### Two population model that allows for an endemic.

Here we lay out a two-population model in an SIR format, where each population can be thought of as a separate city. In the model, we note that the two populations interact through people who travel between them for work or visiting. The number of people traveling between the cities is much smaller than the overall population, so transmission rates between the two populations are much smaller than within each population. Finally, we note that population two is assumed to have an environmental factor that enables the illness to mutate, allowing population two to reinfect recovered individuals in both populations.




### We now consider two populations:

- Population 1: $(S_1, I_1, R_1)$
- Population 2: $(S_2, I_2, R_2)$


### Dynamics Explained:
- Initially, infection spreads primarily within Population 1, then slowly to Population 2 (due to small inter-community transmission).
- After infection emerges in Population 2, an infection wave will move through population 2.
- As population 2 becomes ill,    environmental factor that enables the illness to mutate such that it can re-infect the recovered idviduals in either population.  


<img src="SIR_extended.jpg">





### ODE System (Extended Two-Population Model with Reinfection):
The equations become:
\begin{align*}
\frac{dS_1}{dt} &= -\beta_1 \frac{S_1 I_1}{N_1} - \epsilon_{12}\frac{S_1 I_2}{N_2}, \\
\frac{dI_1}{dt} &= \beta_1 \frac{S_1 I_1}{N_1} + \epsilon_{12}\frac{S_1 I_2}{N_2} +  - \gamma_1 I_1 + \alpha_{12} R_1 \frac{I_2}{N_2}, \\
\frac{dR_1}{dt} &= \gamma_1 I_1 - \alpha_{12} R_1 \frac{I_2}{N_2}, \\
\frac{dS_2}{dt} &= -\beta_2 \frac{S_2 I_2}{N_2} - \epsilon_{21}\frac{S_2 I_1}{N_1} , \\
\frac{dI_2}{dt} &= \beta_2 \frac{S_2 I_2}{N_2} + \epsilon_{21}\frac{S_2 I_1}{N_1}  - \gamma_2 I_2 + \alpha_{22} R_2 \frac{I_2}{N_2}, \\
\frac{dR_2}{dt} &= \gamma_2 I_2 - \alpha_{22} R_2 \frac{I_2}{N_2}. \\
\end{align*}

#### Parameters Explained:
- $\beta_1,\beta_2$: Infection transmission rates within each community.
- $\epsilon_{ij}$ (for $i,j=1,2$, $i\neq j$): Inter-community transmission rates, small compared to intra-community transmission rates.
- $\gamma_1,\gamma_2$: Recovery rates for each population.
- $\alpha_{12}, \alpha_{22}$: Reinfection rates for recovered individuals in populations 1 and 2, respectively.
- $N_1,N_2$: Total constant population sizes, $N_i=S_i+I_i+R_i$.





#### Typical Initial Conditions:
- Population 1 initially infected:
$$ S_1(0)=N_1-I_1(0),\quad I_1(0)>0,\quad R_1(0)=0 $$
- Populations 2 and 3 initially uninfected:
$$ S_2(0)=N_2,\quad I_2(0)=0,\quad R_2(0)=0 $$


#### Behavior Observed:
- Initial epidemic in Population 1.
- Delayed spread to Population 2, followed by reinfection of Populations 1 and 2 may occur, creating a more complex, possibly cyclic, epidemic pattern.
- Reinfection dynamics introduce potential long-term oscillations due to SIRS behavior.

Define the derivative function for the two community SIR model in the cell below (to be used as an input later for `solve_ivp` -- so pay attention to the format).

```python
### ANSWER ###

import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# ODE system for the extended three-population SIR model with reinfection
def derivs(t, curr_vals, params):
    S1, I1, R1, S2, I2, R2 = curr_vals
    beta1, beta2,  gamma1, gamma2,  epsilon12,  epsilon21,   alpha12, alpha22, N1, N2 = params

    dS1_dt = -beta1*S1*I1/N1 - epsilon12*S1*I2/N2  
    dI1_dt = beta1*S1*I1/N1 + epsilon12*S1*I2/N2  - gamma1*I1 + alpha12*R2*I2/N2 
    dR1_dt = gamma1*I1 - alpha12*R2*I2/N2

    dS2_dt = -beta2*S2*I2/N2 - epsilon21*S2*I1/N1 
    dI2_dt = beta2*S2*I2/N2 + epsilon21*S2*I1/N1  - gamma2*I2 + alpha22*R2*I2/N2
    dR2_dt = gamma2*I2 - alpha22*R2*I2/N2


    return [dS1_dt, dI1_dt, dR1_dt, dS2_dt, dI2_dt, dR2_dt]

## rubric
# 1 pts for unpacking current_state correctly
# 1 pts for implementing parameters correctly 
    # (they do not have to unpack parameters, they can use a long list of function inputs)
# 2 pts for defining the derivatives correctly
# 2 pts for returning the derivatives correctly
```

&#9989;&nbsp; **Part 2.2 (5 points)**  
Call `solve_ivp` to solve the two-community SIR problem.  

* For initial conditions, use $S_1=999$, $I_1=1$, $R_1=0$, $S_2=1000$, $I_2=0$, and $R_2=0$.
* For parameters, use $\beta_1 = 0.3$, $\beta_2 = 0.15$, $\gamma_1 = 0.1$, $\gamma_2 = 0.075$, $\epsilon_{12} = 0.001$, $\epsilon_{21} = 0.0005$, $\alpha_{12} = 0.025$, $\alpha_{22} = 0.1$, $N_1 = 1000$, $N_2=1000$.
* Use a final time of $300$.

```python
# Put your code here!
```

```python
### ANSWER ###

# Parameters
beta1 = 0.3
beta2 = 0.15
gamma1 = 0.1
gamma2 = 0.075
eps12 = 0.001
eps21 = 0.0005
alpha12 = 0.025
alpha22 = 0.1
N1 = 1000
N2 = 1000

params = [beta1, beta2, gamma1, gamma2, eps12, eps21, alpha12, alpha22, N1, N2]

# Initial conditions
initial_conditions = [999, 1, 0, 1000, 0, 0]

# Time points
tmax = 300
T = np.linspace(0, tmax, 1000)

# Solving the ODE
result = solve_ivp(derivs, (0,tmax),initial_conditions, t_eval=T, args=(params,))

## Rubric
# 1 pts for defining parameters correctly
# 1 pts for defining initial conditions correctly
# 1 pt for defining time array
# 2 pts for calling solve_ivp correctly


```

```python
### ANSWER ###
# Code for Part 3.4
# Change alphas

# Parameters
beta1 = 0.3
beta2 = 0.15
gamma1 = 0.1
gamma2 = 0.075
eps12 = 0.001
eps21 = 0.0005
# alpha12 = 0.025
# alpha22 = 0.1
alpha12 = 0.05
alpha22 = 0.125
N1 = 1000
N2 = 1000

params = [beta1, beta2, gamma1, gamma2, eps12, eps21, alpha12, alpha22, N1, N2]

# Initial conditions
initial_conditions = [999, 1, 0, 1000, 0, 0]

# Time points
tmax = 300
T = np.linspace(0, tmax, 1000)

# Solving the ODE
result = solve_ivp(derivs, (0,tmax),initial_conditions, t_eval=T, args=(params,))

S1 = result.y[0,:]
I1 = result.y[1,:]
R1 = result.y[2,:]
S2 = result.y[3,:]
I2 = result.y[4,:]
R2 = result.y[5,:]

plt.figure(figsize=(12, 8))
width = 4

plt.plot(T, S1, label='S1', color='tab:blue',linestyle='-',linewidth=width)
plt.plot(T, I1, label='I1', color='tab:blue',linestyle='--',linewidth=width)
plt.plot(T, R1, label='R1', color='tab:blue',linestyle=':',linewidth=width)
plt.plot(T, S2, label='S2', color='tab:orange',linestyle='-',linewidth=width)
plt.plot(T, I2, label='I2', color='tab:orange',linestyle='--',linewidth=width)
plt.plot(T, R2, label='R2', color='tab:orange',linestyle=':',linewidth=width)

plt.xlabel('Time')
plt.ylabel('Number of individuals')
plt.title('Extended Two-Population SIR Model with Reinfection')
plt.legend()
plt.grid()

plt.savefig('SIR_extended_result_Alpha.jpg')


```
