# A More Detailed Exploration of Looping by Index

In this very brief module, I want you to explore each part of a `for` loop that loops by index. Here is an example.


```python
list_ex = ['Harry', 'Hermione', 'Frank', 'Luna', 'Ron']
for j in range(len(list_ex)):
    print(list_ex[j])
```

Lets look at each part of the loop declaration (i.e., `for j in range(len(list_ex)):`)

## `len`

We'll start with `len()`. What does `len()` do? Let's print it!

```python
print(len(list_ex))
```

We get 5. Where does that 5 come from? It's the *number of elements in our list.* So, if we were to add another element to our list, like this.

```python
list_ex.append('Lee')
print(len(list_ex))
```

We now get 6, because we've added another element!

## `range`

Now that we have some idea of what `len` is doing, let's look at `range`. What is that doing? Let's print it!

```python
range_var = range(len(list_ex))
print(range_var)
```

That wasn't very helpful :/

Here's a trick: let's tell Python to treat our variable like a `list` and see what we get (no, I wouldn't expect you to have guessed this).

```python
print(list(range_var))
```

Okay! Now we're getting somewhere! So it appears that `range()` gives us a list of numbers in order. 

Now, we know that `len()` gives us the number of elements in a list (in this case, it's 6). Do we need to use `len()`? What if we just put the number 6 into range?

```python
print(list(range(6)))
```

We get the same thing! We get a range of numbers from 0 to 5, going up by 1. What if we try out other numbers in our range function?

```python
print(list(range(12)))
```

```python
print(list(range(2)))
```

Okay! So, what pattern do we notice here? The command `range(x)` function returns **a list of numbers that goes from 0 to `x-1`**. That last part is very important; it doesn't go up to `x`, it goes up to `x-1`.

Cool! Now, there's one other part to the `range()` function that's important but is hidden. Again, not something we (should) expect you to know about. 

`range()` can actually take multiple values! Specifically, the `range()` function can take a start, stop, and interval values. So, for instance, if we wanted to start at 5, go to 13, and go up by 2, we would use the following command: 

```python
print(list(range(5,14,2)))
```

Again, note that we used 14 instead of 13, because `range()` goes to `x-1`. 

## Bringing it all together: Printing out a list in reverse order

OKAY! So, how can we bring this all together and print out a list in reverse order? Well, reverse order means we want to *start* at the *end* and *end* at the *start*. So we could do something like:

```python
print(list(range(14,5,2)))
```

But that doesn't do anything! The reason is because our interval--2--is a *positive number*. So it's like trying to backwards when you can only take steps forward; it's not going to work! Let's tell `range()` to take steps backwards.

```python
print(list(range(14,5,-2)))
```

There we go! We got our list to print out backwards! Coooool (not really, but please let me have this).
