### &#9989;&nbsp; Question (10 points)
## Part 1a) 

You are conducting an experiment where you are adding chemical samples to a containment vessel. The vessel has a safety threshold of **500 grams**. You have a list representing the precise mass of each sample in the order they are to be added. 

**Your goal is to write a `while` loop that prints the cumulative mass after adding each sample.** The process should continue only until the total mass reaches or exceeds the safety limit of 500 grams.

**Task:** Rewrite the provided `for` loop as a **`while` loop**, ensuring that it produces the exact same output as the code below.

```python
# List of sample masses in grams (DO NOT CHANGE!)
sample_masses = [45.24445, 88.11142, 120.59847, 75.30102, 92.47828, 110.13464, 65.45452, 80.99642, 91.10112, 22.97312]
```

```python
# for loop 
# Rewrite this as a while loop
total_mass = 0
for i in range(len(sample_masses)):
    if total_mass < 500:
        total_mass = total_mass + sample_masses[i]
        print(f"After adding Sample {i+1}, the total mass is {total_mass:.5f} grams.")
```
