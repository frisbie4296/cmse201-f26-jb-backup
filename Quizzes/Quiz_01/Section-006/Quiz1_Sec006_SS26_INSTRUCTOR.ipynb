# 1. Problem Context: Converting between `for` loops and `while` loops 

### &#9989;&nbsp; Question (10 points)
## Part 1a) 

You are conducting an experiment where you are adding chemical samples to a containment vessel. The vessel has a safety threshold of **500 grams**. You have a list representing the precise mass of each sample in the order they are to be added. 

**Your goal is to write a `while` loop that prints the cumulative mass after adding each sample.** The process should continue only until the total mass reaches or exceeds the safety limit of 500 grams.

**Task:** Rewrite the provided `for` loop as a **`while` loop**, ensuring that it produces the exact same output as the code below.

```python
# List of sample masses in grams (DO NOT CHANGE!)
sample_masses = [45.24445, 88.11142, 120.59847, 75.30102, 92.47828, 110.13464, 65.45452, 80.99642, 91.10112, 22.97312]

print(f'I have {len(sample_masses)} samples prepared. The total mass is {sum(sample_masses):.5f} grams.')
```

```python
# for loop 
# Rewrite this as a while loop
total_mass = 0
for i in range(len(sample_masses)):
    if total_mass < 500:
        total_mass = total_mass + sample_masses[i]
        print(f"After adding Sample {i+1}, the total mass is {total_mass:.5f} grams.")
```

```python
# put your answer here
```

```python
### ANSWER

total_mass = 0
i = 0
while total_mass < 500:
    total_mass = total_mass + sample_masses[i]
    print(f"After adding Sample {i+1}, the total mass is {total_mass:.5f} grams.")
    i += 1

# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)

You are an astronomer classifying stars based on their **Effective Temperature ($T_{eff}$)** in Kelvin. Depending on the surface temperature of the star, you will assign it to a category.

The categories are defined as follows:
* **"Cool"**: $T_{eff} \le 3,500$ K
* **"Intermediate"**: $3,500 < T_{eff} \le 6,000$ K
* **"Hot"**: $6,000 < T_{eff} \le 10,000$ K
* **"Extreme"**: $T_{eff} > 10,000$ K

In the cell below, you are provided with a list of 10 stars and their recorded surface temperatures. Using **`if`**, **`elif`**, and **`else`** statements inside a loop, print which category each star belongs to.

Some starter code is provided to help you navigate the lists!


```python
# Lists of star names and their temperatures in Kelvin (DO NOT CHANGE!)
stars = ["Proxima Centauri", "Barnard's Star", "The Sun", "Alpha Centauri A", 
         "Sirius A", "Vega", "Achernar", "Rigel", "Bellatrix", "Spica"]

temps = [3042, 3500, 5778, 6000, 9940, 10000, 15000, 12100, 22000, 25000]

# Example starter code
for i in range(len(stars)):
    # Add your if/elif/else logic here to determine the category
    print(f"{stars[i]} has a temperature of {temps[i]} K and is [category]")
```

```python
### ANSWER

for i in range(len(stars)):
    if temps[i] <= 3500:
        category = "Cool"
    elif temps[i] <= 6000:
        category = "Intermediate"
    elif temps[i] <= 10000:
        category = "Hot"
    else:
        category = "Extreme"
    
    print(f"{stars[i]} has a temperature of {temps[i]} K and is {category}")
    
# RUBRIC
# 10 out of 10: Correct use of if/elif/else logic that properly handles the boundary temperatures according to the provided ranges.
#7 out of 10: An attempt is made, but there is a logical error in the thresholds (e.g., using < instead of <= resulting in "off-by-one" errors on the boundaries).
#4 out of 10: Attempt is made but uses the wrong conditional structure or fails to use the list index correctly.
#0 out of 10: No attempt.
```
