## Part 1a) Question (10 points)
**You want to print out the months of the year to schedule your school year.**  
**Rewrite the provided `for` loop as a `while` loop, maintaining the same output.**

```python
# Provided for loop
months = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
for month in months:
    print(month)
```

```python
### YOUR ANSWER

index = 0
while index < len(months):
    print(months[index])
    index += 1

```

# 2. Problem Context: Using if/else Statements to Control Flow (20 points)

## Part 2a) Question (10 points)

Use `if`, `elif`, and `else` statements for the exercise below.

Iterate through each number in the provided list; printing out one of each of the following cases according to the value of the number:\
[`number`] is even and greater than 25\
[`number`] is odd and less than 25\
[`number`] is equal to 25\
[`number`] does not meet any of the specified conditions

For example, if the number on the list is '3', your output should read:\
3 is odd and less than 25


```python
list = [100, 26, 13, 4, 25, 96, 71, 8, 29, 100]
```

```python
# YOUR ANSWER

for num in list:
    if num % 2 == 0 and num > 25:
        print(f"{num} is even and greater than 25")
    elif num % 2 != 0 and num < 25:
        print(f"{num} is odd and less than 25")
    elif num == 25:
        print(f"{num} is equal to 25")
    else:
        print(f"{num} does not meet any of the specified conditions")
```

# 3. Using Loops for Calculations (20 points)

## Part 3a) Question (10 points)

You're managing a small restaurant and need to tally total sales of menu items in the last hour of the day. \
Write a program that computes the total sales based on the item prices and the number of each item sold.

```python
# Example starter code
item_prices = [20.00, 15.00, 35.50, 42.00, 37.50, 20.00, 25.50, 5.00, 6.50]
number_sold = [3, 1, 3, 1, 2, 3, 1, 1, 0]
```

```python
# YOUR ANSWER
total_revenue = 0
for i in range(len(item_prices)):
    total_revenue += item_prices[i] * number_sold[i]

print("Total Revenue:", total_revenue)
```
