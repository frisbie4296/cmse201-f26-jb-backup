````python
# while loop

cities = ["Querétaro", "San Miguel de Allende", "Dolores Hidalgo", "Atotonilco", "Guanajuato"]

i = 0 # set a counter to 0
while i < len(cities): 
    print("We visited", cities[i])
    i += 1 # increment counter +1
````

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)

In San Miguel de Allende, our group goes to a restaurant. In our group of twelve students, each student orders a different dish from the menu: `tacos`, `flautas`, `sopes`, `huaraches`, `tlacoyos`, `quesadillas`, `tamales`, `gorditas`.

For the eight dishes listed above, the two lists below indicate 1) the price per dish (in pesos) and 2) how many students ordered the dish.

Write some code that, for each dish, will multiply out the price in pesos by how many students ordered that dish. Then, add the amount per a dish to a total. That way we will see how much money we spend overall on the receipt (*la cuenta*).

(Note: the name of the dish does not matter, we just need the money spent per dish based on the price per dish and number of students who ordered it)

As a check, if you print out the subtotal each iteration, you should get:
```
100
100
450
1250
1250
2000
2000
2600
```

````python
# Example starter code
price_per_dish = [100, 150, 175, 200, 130, 250, 200, 300] # the price for a single dish (8 dishes total)
number_students = [1, 0, 2, 4, 0, 3, 0, 2] # number of students who ordered each disth (8 dishes total)
````

````python
### ANSWER
# Example starter code
price_per_dish = [100, 150, 175, 200, 130, 250, 200, 300] # the price for a single dish (8 dishes total)
number_students = [1, 0, 2, 4, 0, 3, 0, 2] # number of students who ordered each disth (8 dishes total)

total = 0
for i in range(len(price_per_dish)):
    total += price_per_dish[i]*number_students[i]
    print(total)

# RUBRIC
# points for code that runs and produces the appropriate output
# points for reasoning
# points for resources
````
