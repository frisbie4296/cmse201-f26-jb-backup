# 1. Problem Context: Converting between `for` loops and `while` loops 

### &#9989;&nbsp; Question (10 points)
## Part 1a)
You are planning a travel itinerary and want to visit several cities. Rewrite the given `for` loop as a `while` loop, ensuring that the output remains the same.


```python
cities = ["Detroit", "Chicago", "Las Vegas"]
for city in cities:
    print("Visiting", city)
```

```python
# put your answer here
```

```python
### ANSWER

cities = ["Detroit", "Chicago", "Las Vegas"]
for city in cities:
    print("Visiting", city)
    
# while loop    
i = 0
while i < len(cities):
    print("Visiting", cities[i])
    i += 1



# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)

You are analyzing students’ study habits and want to classify their productivity based on the number of hours studied each day. Below, you will find a list of 12 study hours. Using `if`, `elif`, and `else` statements, determine the productivity level for each student according to the following scale:

* "Very Low": 0–1 hours

* "Low": 1–2.9 hours

* "Moderate": 3–4.9 hours

* "High": 5–6.9 hours

* "Very High": 7+ hours

Some starter code is provided to help you navigate the list and format the output.


```python
# List of study hours for 12 students
study_hours = [0, 1.5, 3, 4.5, 5, 6, 7, 2, 3.5, 0.5, 5.5, 6.5]

# Loop through each student's study hours
for hours in study_hours:
    print("Studied", hours, "hours:", "example", "productivity today!")
```

```python
# put your answer here
```

```python
### ANSWER

study_hours = [0, 1.5, 3, 4.5, 5, 6, 7, 2, 3.5, 0.5, 5.5, 6.5]

for hours in study_hours:
    if hours <= 1:
        productivity = "Very Low"
    elif 1 < hours <= 2.9:
        productivity = "Low"
    elif 3 <= hours <= 4.9:
        productivity = "Moderate"
    elif 5 <= hours <= 6.9:
        productivity = "High"
    else:
        productivity = "Very High"
        
    print("Studied", hours, "hours:", productivity, "productivity today!")
```

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You are helping a movie theater calculate its daily revenue. The theater sells tickets for different movies, and each movie has a specific ticket price and number of tickets sold. Write a program to compute the total revenue for the day.

```python
# Movie ticket prices (in dollars)
ticket_prices = [12, 10, 8]

# Number of tickets sold for each movie
tickets_sold = [150, 200, 180]

# TODO: Use a loop to calculate revenue for each movie
# and accumulate into total_revenue
```

```python
# Put your code here
```

```python
### ANSWER

# Movie ticket prices (in dollars)
ticket_prices = [12, 10, 8]

# Number of tickets sold for each movie
tickets_sold = [150, 200, 180]

# Initialize total revenue
total_revenue = 0  # 2 points

# Method 1: Use a for loop to calculate total revenue
for i in range(len(ticket_prices)):  # 2 points
    revenue = ticket_prices[i]*tickets_sold[i] # 2 points
    total_revenue +=revenue # 2 points
    print(f"Book {i+1}: ${ticket_prices[i]} x {tickets_sold[i]} copies = ${revenue:.2f}") # 2 points
print("Total revenue: $", round(total_revenue,2)) # 2 points
    
#Method 2: Use a while loop to calculate total revenue
c = 0 # 1 points 
total_revenue2 = 0 # 1 points

while c <  len(ticket_prices): # 2 points
    revenue2 = ticket_prices[c]*tickets_sold[c] # 2 points
    total_revenue2 +=revenue2 # 1 points
    print(f"Book {c+1}: ${ticket_prices[c]} x {tickets_sold[c]} copies = ${revenue2:.2f}") #1 points
    c +=1 # 1 points
print("Total revenue: $", round(total_revenue2,2))# 1 points
```
