You are visiting a cat cafe, and you want to make sure you pet all of the cats. Fortunately, you brought your computer with you with Python installed, so you've got this. You need to loop over the list of cats and print out a string that says which cat you just pet. You start by writing a `while` loop to accomplish this task, but after thinking about it for a moment, you decide to rewrite it as a `for` loop.

**Rewrite the provided `while` loop as a `for` loop, maintaining the same output.**

```python
# while loop

cats = ["Buster", "Splat", "Splash", "Bootsie", "Bear","Jeep", "Harriet", "Marian", "Heimdall", "Loki", "Legolas", "Skadi", "BearBear", "Jasper"] # list of cats in Dr. Frisbie's family (past and present)

curr_cat = 0
while curr_cat<len(cats):
    print("I pet",cats[curr_cat]+"!")
    curr_cat+=1
```

```python
# put your answer here
```

```python
### ANSWER

cats = ["Buster", "Splat", "Splash", "Bootsie", "Bear","Jeep", "Harriet", "Marian", "Heimdall", "Loki", "Legolas", "Skadi", "BearBear", "Jasper"] # list of cats in Dr. Frisbie's family (past and present)

curr_cat = 0
while curr_cat<len(cats):
    print("I pet",cats[curr_cat]+"!")
    curr_cat+=1

for cat in cats:
    print("I pet",cat+"!")
# RUBRIC
# 10 out of 10 for a correctly formatted while loop that maintains the same output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 2. Problem Context: Using if/else Statements to Control Flow

### &#9989;&nbsp; Question (10 points)
## Part 2a)
As a college student in Michigan, you are tired of deciding if you should wear a coat or not, so you are going to write some Python code to make those decisions for you. You decide that for temperatures below 32$^\circ$F, you will wear a winter coat, for temperatures between 32$^\circ$F and 50$^\circ$F, you will wear a light jacket, and for temperatures greater than 50$^\circ$F, you won't wear a jacket. 

In the cell below, you will see a list of 10 temperatures. Using `if`, `elif`, and `else` statements, determine what kind of jacket you should wear for each day (and print it out!). Some starter code for navigating the list (and formatting the output) is given!

```python
# Example starter code
temperatures = [42,28,31,40,45,48,48,53,56,49]

for temp in temperatures:
    
    print("You need to wear a winter jacket today!")

    print("You need to wear a light jacket today!")

    print("You don't need to wear jacket today!")
```

```python
### ANSWER

temperatures = [42,28,31,40,45,48,48,53,56,49]

for temp in temperatures:
    if temp <= 32:
        print("You need to wear a winter jacket today!")
    elif temp > 32 and temp < 50:
        print("You need to wear a light jacket today!")
    else:
        print("You don't need to wear jacket today!")

    
# RUBRIC
# 10 out of 10 for a correctly formatted if/elif/else that gives correct output
# 7 out of 10 for an attempt that is not formatted correctly OR output is incorrect
# 4 out of 10 for an attempt that is not formatted correctly AND output is incorrect
# 0 out of 10 for no attempt
```

# 3. Using Loops for Calculations

### &#9989;&nbsp; Question (10 points)
## Part 3a)
You have been watching figure skating at the Olympics, and you decide that you want to keep track of the scores as each skater completes their program. 

You have one list that holds all of the "base values" for a skater's short program and another list (of equal length) that holds the "grade of execution" scores for each element (assume that the 0-th element of the base values list goes with the 0-th element of the grade of exection list, and so on).

Write a program that:
1. Sets up an empty list to fill with each element score (i.e. base value + grade of execution)
2. Initializes a variable to keep track of the total score
3. Loops over the lists and:
   * Adds the base value and grade of execution scores together and appends them to the element score list
   * Adds the total score for the element to the total score variable
4. Prints out the total technical score

```python
# Example starter code
# Ilia Malinin's Olympic individual short program scores (https://skatingscores.com/2526/oly/sr/men/i/short/usa/ilia_malinin/)
base_values = [11.00,8.00,3.20,17.27,2.60,3.30,3.50]
grades_of_execution = [3.77,1.60,0.23,4.76,0.56,1.56,1.00]
```

```python
### ANSWER
# Example starter code
base_values = [11.00,8.00,3.20,17.27,2.60,3.30,3.50]
grades_of_execution = [3.77,1.60,0.23,4.76,0.56,1.56,1.00]
element_totals = []

total = 0
for i in range(len(base_values)):
    element_score = base_values[i]+grades_of_execution[i]
    element_totals.append(element_score)
    total += element_score
print(total)

# RUBRIC
# points for code that runs and produces the appropriate output
# points for reasoning
# points for resources
```
