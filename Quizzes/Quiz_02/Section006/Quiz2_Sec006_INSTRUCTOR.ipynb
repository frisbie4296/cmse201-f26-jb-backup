## Part 1a) (20 points)

Before we dive into the full dataset, we need to build a reliable tool for basic data exploration. 

**Using the test data provided below, write a function** (for example, named `get_health_stats`) that takes an array of values as input and returns the following three statistics:
1. The **mean** of the array.
2. The **standard deviation** of the array.
3. The **median** of the array.

**Demonstrate that your function works** by calling it using the `test_heart_rates` data provided. We will use this function later in the quiz to analyze biometric trends!

```python
import numpy as np

# --- TEST DATA ---
# This array represents a 10-minute window of Heart Rate (BPM) 
# taken from a user during a light jog.
test_heart_rates = np.array([112.5, 115.0, 118.2, 122.1, 120.5, 119.0, 125.4, 128.3, 121.0, 116.8])
```

```python
# --- PUT YOUR FUNCTION HERE ---
```

```python
### ANSWER ###

def get_health_stats(data_array):
    mu = np.mean(data_array)
    sigma = np.std(data_array)
    med = np.median(data_array)
    return mu, sigma, med

# Calling the function to demonstrate it works
print(get_health_stats(test_heart_rates))

# Rubric:
# 10 points for defining the function and having it return values
# 5 points for calculating the statistics (mean, std, median)
# 5 points for testing the function using the provided test data
```

```python
import matplotlib.pyplot as plt

# --- PUT YOUR CODE HERE ---
```

```python
### ANSWER ###

import matplotlib.pyplot as plt

# OPTION 1: Heart Rate (Column 1)
time, hr = np.loadtxt('BioWatch_Data.csv', delimiter=',', skiprows=1, usecols=(0, 1), unpack=True)
plt.figure(figsize=(8, 4))
plt.plot(time, hr, color='crimson')
plt.xlabel('Time (Seconds)')
plt.ylabel('Heart Rate (BPM)')
plt.title('Heart Rate Trends')
plt.show()

# OPTION 2: SpO2 Percent (Column 2)
time, spo2 = np.loadtxt('BioWatch_Data.csv', delimiter=',', skiprows=1, usecols=(0, 2), unpack=True)
plt.figure(figsize=(8, 4))
plt.plot(time, spo2, color='royalblue')
plt.xlabel('Time (Seconds)')
plt.ylabel('Blood Oxygen (%)')
plt.title('Oxygen Saturation (SpO2)')
plt.show()

# OPTION 3: Stress Score (Column 3)
time, stress = np.loadtxt('BioWatch_Data.csv', delimiter=',', skiprows=1, usecols=(0, 3), unpack=True)
plt.figure(figsize=(8, 4))
plt.plot(time, stress, color='darkorange')
plt.xlabel('Time (Seconds)')
plt.ylabel('Stress Level (0-100)')
plt.title('Calculated Stress Metrics')
plt.show()

# Rubric:
# 5 points: Correct use of np.loadtxt (using delimiter=',', skiprows=1, and correct usecols)
# 3 points: Plotting choice of data vs Time_Seconds
# 2 points: Including axis labels and a title
```
