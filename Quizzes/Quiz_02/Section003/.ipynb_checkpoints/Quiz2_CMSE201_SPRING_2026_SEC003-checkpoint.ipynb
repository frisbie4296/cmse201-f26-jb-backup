## Part 1a) (20 points)

Using the test data below, **write a function that takes two arrays of values as input and returns the following statistics**:

- The **mean** of the first array  
- The **standard deviation** of the first array  
- The **median** of the first array  
- The **Pearson correlation coefficient** between the two arrays  

Demonstrate that your function works by calling it using the test data provided below.

We will use this function later in the quiz to analyze the relationship between different variables in the dataset.

**Note:**  
The first three statistics (mean, standard deviation, and median) should be calculated using the **first array only**, while the **correlation coefficient should be calculated using both arrays**.

You are provided with both `x` and `y` test data to demonstrate that your function works correctly.


```python
import numpy as np

# test data
test_x = np.arange(0,10,1.0)

test_y = np.array([
    14., 78., 98., 56., 75.,
    67., 84., 63., 45., 31.
])

# PUT YOUR FUNCTION HERE
```

```python
### ANSWER ###
#test data
test_x = np.arange(0,10,1.0)

test_y = np.array([
    14., 78., 98., 56., 75.,
    67., 84., 63., 45., 31.
])

# function
def summary_stats(x, y):
    
    mean_val = np.mean(x)
    std_val = np.std(x)
    median_val = np.median(x)
    
    corr = np.corrcoef(x, y)[0,1]
    
    return mean_val, std_val, median_val, corr


mean_val, std_val, median_val, corr = summary_stats(test_x, test_y)

print("Mean:", mean_val)
print("Std:", std_val)
print("Median:", median_val)
print("Correlation:", corr)


#Rubric:
# 10 points for defining the function and having it return values
# 5 points for calculating the statistics
# 5 points for testing the function
```

```python
### ANSWER ###
plt.figure(figsize=(8,6))

# Scatter plot
plt.scatter(chosen_variable, exam_score, alpha=0.6, label="Exam Score Data")

# Mean line
plt.axhline(mean_val, color='red', linestyle='--', linewidth=2, label='Mean')

# Median line
plt.axhline(median_val, color='green', linestyle='-.', linewidth=2, label='Median')

# Standard deviation lines
plt.axhline(mean_val + std_val, color='blue', linestyle=':', linewidth=2, label='+1 Std Dev')
plt.axhline(mean_val - std_val, color='blue', linestyle=':', linewidth=2, label='-1 Std Dev')

# Labels and title
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Relationship Between Hours Studied and Exam Score")

# Legend
plt.legend()

plt.show()

#RUBRIC
```
