# 2. Fitting a model to data with `curve_fit` and visualizing the results (40 points)

You’re helping a coastal lab calibrate a **wave buoy**. The buoy’s vertical position oscillates as waves pass. If you can fit a simple sinusoid to the motion, you can estimate the **wave frequency**.

**Data columns**

* `time` — time in seconds
* `height` — buoy height (meters, relative to mean sea level)

---


```python
### RUN THIS CELL to load the dataset

import pandas as pd

buoy_data = pd.read_json(
    '{"time":[0.0,0.5,1.0,1.5,2.0,2.5,3.0,3.5,4.0,4.5,5.0,5.5,6.0,6.5,7.0,7.5,8.0,8.5,9.0,9.5,10.0],'
    '"height":[0.33,0.79,1.25,1.55,1.45,1.90,1.42,1.70,1.05,0.62,0.55,-0.28,-0.72,-0.96,-1.38,-1.10,-1.56,-1.05,-0.83,-0.42,-0.05]}'
)
buoy_data.head()
```

#### &#9989;&nbsp; **Question 2.1 (20 points):** 

A colleague in your lab previously analyzed a similar buoy dataset and suggested that the parameters below are good **starting values** for a simple sinusoidal model of wave motion. Using the data and model below, use `curve_fit` to find best-fit parameters for *a, b,* and *c*. Be sure to pass these starting values via the `p0` argument:

* `a = 1.5`
* `b = 0.6`
* `c = 0.2`

Once you’ve fit the model, **print the best-fit values**. They should be close to, but not exactly, the starting guesses.

```python
# Will need to use NumPy for the function
import numpy as np

# Will need to import curve_fit from SciPy
from scipy.optimize import curve_fit

# This function calculates volts corresponding to time for a sound wave
def sine_wave(t,a,b,c):
    return a*np.sin(b*t)+c
```

```python
# Put your code here
```

```python
### ANSWER
params,ignore = curve_fit(sine_wave,buoy_data['time'], buoy_data['height'],p0=[1.5,0.6,0.2])
print(params)

# rubric
# 5 pts for using model function correctly
# 5 pts for using data correctly
# 5 pts for p0 correct
# 5 pts for printing
```

#### &#9989;&nbsp; **Question 2.2 (10 points)** 

Now that you have a set of best fit parameters, **use those parameters to calculate the expected model values and plot the expected model values on top of the actual data values**. Make sure to label your axes and include a legend to communicate what parts of the plot correspond to the model versus the data.

```python
# Make sure we import matplotlib
import matplotlib.pyplot as plt

# Put your code here
```

```python
### ANSWER ###
plt.scatter(buoy_data['time'],buoy_data['height'],label='data')
plt.plot(buoy_data['time'],sine_wave(buoy_data['time'],params[0],params[1],params[2]),label='fit')
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.title("Fitted Sine Model of Wave Buoy Motion")
plt.legend()
plt.grid(alpha=0.3)
plt.legend();

# Rubric
# 5 pts for data and best fit
# 5 pts for labels, legend, etc.
```
