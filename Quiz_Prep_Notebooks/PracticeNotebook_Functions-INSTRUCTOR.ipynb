```python
# Put your pseudocode and code here
import numpy as np
```

```python
###ANSWER###
def slope_int(num_list=np.array([1,2,3]),m=1,b=0):
    return m*num_list+b

my_array = np.arange(0,10,1)

y_values = slope_int(my_array,m=2,b=1)
print(y_values)
x_values = my_array

import matplotlib.pyplot as plt
plt.plot(x_values,y_values,marker='<')
plt.xlabel('x')
plt.ylabel('y')
```
