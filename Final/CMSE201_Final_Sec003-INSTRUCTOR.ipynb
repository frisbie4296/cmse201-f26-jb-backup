## Part 3: Fit data using NumPy and/or SciPy (27 points)

In this part you will load in some data and manipulate it using different functions.

```python
### ANSWER (actually this is the code to generate the data)
import numpy as np

x = np.linspace(-7,7,200)
y = 19.5*np.sin(0.48*x + 2.2) + 1.3*x - 8
fake_data = np.random.normal(0,4,size=(200)) + y
np.savetxt('mystery.csv',np.stack((x,fake_data)))
```

### Execute the code below to read data from the file `mystery.csv`:

```python
x,y = np.loadtxt('mystery.csv')
```

### 3a) Plot `x` vs `y` (use axis labels). (3 points)

```python
# your code here
```

```python
### ANSWER
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
```

### 3b) Use numpy's `polyfit` function to generate first-, second- and third-order polynomial fits.  Plot these along with the function.  Label the axes and the curves. (6 points)

```python
# your code here
```

```python
### ANSWER
par1 = np.polyfit(x,y,1)
par2 = np.polyfit(x,y,2)
par3 = np.polyfit(x,y,3)

func1 = np.poly1d(par1)
func2 = np.poly1d(par2)
func3 = np.poly1d(par3)

plt.plot(x,y,label='Data')
plt.plot(x,func1(x),label='linear fit')
plt.plot(x,func2(x),label='quadratic fit')
plt.plot(x,func3(x),label='cubic fit')
plt.legend()
```

### 3c) Use scipy's curve fit function to fit the data to a cubic function: $y = A\sin(Bx+C) + Dx + E$.  Print the parameters.  Then plot this fit, as well as the numpy cubic fit, on top of the data. (10 points)

```python
# your code here
```

```python
### ANSWER
def fit(x,A,B,C,D,E):
    return A*np.sin(B*x+C) + D*x + E
```

```python
### ANSWER
from scipy.optimize import curve_fit
params, cov = curve_fit(fit,x,y)
print("Scipy parameters:",params)
```

```python
### ANSWER
plt.plot(x,y,label='Data')
plt.plot(x,func3(x),label='numpy cubic fit')
plt.plot(x,fit(x,params[0],params[1],params[2],params[3],params[4]),label='scipy fit')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```

### 3d) Subtract both fitted curves from the data and store the results in separate arrays.  For each array, print the average value as well as the average squared value. (6 points)

```python
# your code here
```

```python
### ANSWER
arr_np = func3(x) - y
arr_sp = fit(x,params[0],params[1],params[2],params[3],params[4]) - y

print("Numpy fit:")
print("Average: ",arr_np.mean())
print("Average sq: ",np.square(arr_np).mean())
print()
print("Scipy fit:")
print("Average: ",arr_sp.mean())
print("Average sq: ",np.square(arr_sp).mean())
```
