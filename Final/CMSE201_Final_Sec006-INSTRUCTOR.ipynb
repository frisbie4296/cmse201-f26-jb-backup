```python
### ANSWER
par2 = np.polyfit(x,y,2)
par3 = np.polyfit(x,y,3)
par20 = np.polyfit(x,y,20)

func2 = np.poly1d(par2)
func3 = np.poly1d(par3)
func20 = np.poly1d(par20)

plt.scatter(x,y,label='Data')
plt.plot(x,func2(x),label='quadratic fit')
plt.plot(x,func3(x),label='cubic fit')
plt.plot(x,func20(x),label='twentieth order fit')
plt.legend()
```

### 3c) Use scipy's curve fit function to fit the data to a cubic function: $y = A\sin(Bx+C) + Dx + E$.  Print the parameters.  Then plot this fit, as well as the numpy twentieth-order fit, on top of the data. (7 points)

```python
# your code here
```

```python
### ANSWER
def fit(x,A,B,C,D,E):
    return A*np.sin(B*x+C) + D*x + E
```

```python
### ANSWER
from scipy.optimize import curve_fit
params, cov = curve_fit(fit,x,y)
print("Scipy parameters:",params)
```

```python
### ANSWER
plt.scatter(x,y,label='Data')
plt.plot(x,func20(x),label='numpy twentieth order fit', color = 'red')
plt.plot(x,fit(x,params[0],params[1],params[2],params[3],params[4]),label='scipy fit', color = 'black')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```

### 3d) Calculate the Mean squared error (MSE) using the training data (from training.csv) for scipy fit and numpy twentieth order fit. Which one is better in this metric?   (6 points)

```python
### ANSWER
arr_np = func20(x) - y
arr_sp = fit(x,params[0],params[1],params[2],params[3],params[4]) - y

print("Numpy fit:")
print("Average: ",arr_np.mean())
print("Average sq: ",np.square(arr_np).mean())
print()
print("Scipy fit:")
print("Average: ",arr_sp.mean())
print("Average sq: ",np.square(arr_sp).mean())
```
